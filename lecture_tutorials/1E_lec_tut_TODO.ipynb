## 12.1. The Conditions for Equilibrium

In mechanics, many real-world systems — from bridges and buildings to human limbs — remain motionless even though multiple forces act on them.  
Such systems are said to be in **equilibrium**, a state where the combined effects of all external forces and torques produce no motion.

Understanding equilibrium is essential for analyzing stationary structures and mechanical systems in static balance.

---

### Definition of Equilibrium

A body is in **equilibrium** when both its **translational** and **rotational** motions are absent or constant.  
That is, there is no linear acceleration and no angular acceleration.

Thus, equilibrium implies both:

1. **Translational equilibrium**  
   The vector sum of all external forces acting on the body is zero:
   $$
   \sum \vec{F} = 0.
   $$

2. **Rotational equilibrium**  
   The vector sum of all external torques about any axis is zero:
   $$
   \sum \vec{\tau} = 0.
   $$

When these two conditions are simultaneously satisfied, the object is in complete mechanical equilibrium.

---

### Static Equilibrium

If, in addition, the object is **at rest** (not moving at constant velocity), the equilibrium is called **static equilibrium**.

Examples include:
- A ladder leaning against a wall,
- A bridge supporting its own weight,
- A person standing still.

---

### Translational Equilibrium in Components

The condition $\sum \vec{F} = 0$ means that the net force must vanish in every coordinate direction:

$$
\sum F_x = 0, \quad \sum F_y = 0, \quad \sum F_z = 0.
$$

For planar (two-dimensional) problems, the $z$-component is zero by definition, leaving two scalar equations.

---

### Rotational Equilibrium in Components

The second equilibrium condition, $\sum \vec{\tau} = 0$, ensures no net angular acceleration.  
Torque about a point or axis is defined as:

$$
\vec{\tau} = \vec{r} \times \vec{F},
$$

with magnitude:

$$
\tau = r F \sin\theta,
$$

where $\theta$ is the angle between $\vec{r}$ (the position vector from the axis) and $\vec{F}$.

The rotational equilibrium condition means:

$$
\sum \tau = 0,
$$

which must hold for any axis chosen.

---

### Choice of Axis

The torque equilibrium equation is true for **any** axis, but it is often convenient to choose one that simplifies the calculation — for example, through a point where an unknown force acts, so that its torque contribution is zero.

---

### Example 12.1. Equilibrium of a Beam Supported by a Cable

A uniform horizontal beam of weight $W_b = 200~\text{N}$ and length $L = 4.0~\text{m}$ is hinged to a wall at one end and supported by a cable making an angle $\theta = 30°$ with the beam.  
Find the tension in the cable and the horizontal and vertical components of the hinge force.

#### Solution

**Step 1. Identify forces**

- Weight of the beam acts downward at its center ($L/2$).  
- Tension $T$ acts upward and to the left at the beam’s end.  
- Hinge exerts reaction forces $H_x$ and $H_y$.

**Step 2. Torque equilibrium**

Take torques about the hinge:

$$
\sum \tau = 0 \Rightarrow T\sin\theta(L) - W_b(L/2) = 0.
$$

Solve for $T$:

$$
T = \frac{W_b/2}{\sin\theta} = \frac{200/2}{\sin30°} = \frac{100}{0.5} = 200~\text{N}.
$$

**Step 3. Force equilibrium**

Horizontal:
$$
\sum F_x = 0 \Rightarrow H_x - T\cos\theta = 0 \Rightarrow H_x = T\cos30° = 200(0.866) = 173~\text{N}.
$$

Vertical:
$$
\sum F_y = 0 \Rightarrow H_y + T\sin\theta - W_b = 0.
$$

$$
H_y = W_b - T\sin30° = 200 - 200(0.5) = 100~\text{N}.
$$

**Results:**
- $T = 200~\text{N}$  
- $H_x = 173~\text{N}$  
- $H_y = 100~\text{N}$

---

### Example 12.2. Two-Force Member

A light bar is pinned at both ends with equal and opposite forces $F$.  
Show that the forces must act along the line of the bar.

#### Solution

For equilibrium:
- $\sum \vec{F} = 0$ requires the two forces be equal and opposite.
- $\sum \vec{\tau} = 0$ can only hold if the forces produce **no torque**, which occurs only when their lines of action are **collinear** with the bar.

Therefore, any **two-force member** (a rigid body acted on by only two forces) must have those forces **equal, opposite, and collinear**.

---

### Example 12.3. Ladder Against a Wall

A uniform ladder of length $L = 6.0~\text{m}$ and weight $W = 180~\text{N}$ rests against a smooth wall at an angle $\theta = 53°$ with the floor.  
The floor is rough enough to prevent slipping.  
Find the forces on the ladder.

#### Solution

Forces:
- Weight $W$ acts at the midpoint downward.
- Normal force from wall $N_w$ acts horizontally.
- Normal force from floor $N_f$ acts vertically.
- Friction $f$ acts horizontally at the base.

**Torque equilibrium about the base:**
$$
N_w(L)\sin\theta - W(L/2)\cos\theta = 0.
$$
$$
N_w = \frac{W}{2}\cot\theta = \frac{180}{2}\cot53° = 90(0.75) = 67.5~\text{N}.
$$

**Horizontal equilibrium:**
$$
f = N_w = 67.5~\text{N}.
$$

**Vertical equilibrium:**
$$
N_f = W = 180~\text{N}.
$$

Hence, the floor exerts a vertical normal force of **180 N** and a horizontal friction force of **67.5 N** to prevent slipping.

---

### General Strategy for Equilibrium Problems

1. **Draw a free-body diagram (FBD)** showing all external forces.  
2. **Choose a convenient coordinate system.**  
3. **Apply the equilibrium conditions:**
   $$
   \sum F_x = 0, \quad \sum F_y = 0, \quad \sum \tau = 0.
   $$
4. **Choose the axis** to simplify torque calculations (preferably through unknown forces).  
5. **Solve the resulting equations** for the unknown quantities.

---

### Conceptual Notes

- Both translational and rotational equilibrium must hold simultaneously.  
- Torques depend on the **choice of reference point**, but if $\sum \vec{F} = 0$, $\sum \tau = 0$ about one point implies it is zero about all points.  
- In static systems, static friction often prevents motion but does **no work** if there is no slipping.  
- The ability to choose the pivot point freely is a powerful simplification tool in statics.

---

In the next section, we will develop systematic procedures for **solving statics problems**, applying these equilibrium conditions to a wide variety of mechanical systems.


## 12.2. Solving Statics Problems

Statics problems involve determining unknown forces, torques, or reactions that keep an object in equilibrium.  
Such analyses are fundamental in engineering, biomechanics, and architecture, where loads must be supported without motion or failure.

This section presents a systematic procedure for solving equilibrium problems efficiently and accurately.

---

### The Two Equilibrium Conditions

A rigid body is in **equilibrium** when both conditions are satisfied:

1. **Translational equilibrium**
   $$
   \sum \vec{F} = 0.
   $$

   This means there is no net linear acceleration — the center of mass is at rest or moves at constant velocity.

2. **Rotational equilibrium**
   $$
   \sum \vec{\tau} = 0.
   $$

   There is no net torque; the object does not begin to rotate or change its rotational speed.

For planar systems, we typically resolve forces into two components ($x$ and $y$) and write one torque equation, giving three independent scalar equations.

---

### Step-by-Step Procedure

#### Step 1. Identify the Object in Equilibrium
Determine which object or subsystem is to be analyzed.  
If the problem involves several connected parts, isolate one part at a time for analysis.

#### Step 2. Draw the Free-Body Diagram (FBD)
Draw the object and show **all external forces** acting on it, including:
- Weights,
- Tensions,
- Normal forces,
- Friction forces,
- Reaction forces at supports or hinges.

Internal forces between connected parts are omitted in each FBD because they cancel pairwise when analyzing the full system.

#### Step 3. Choose a Convenient Coordinate System
Align axes with the directions of known forces (for example, horizontal and vertical).  
This minimizes the need for trigonometric components.

#### Step 4. Choose a Convenient Axis for Torques
Select a pivot point to simplify torque calculations.  
If possible, choose a point through which one or more unknown forces pass — their torques will be zero.

#### Step 5. Write Out the Equilibrium Equations
Use the scalar forms:
$$
\sum F_x = 0, \quad \sum F_y = 0, \quad \sum \tau = 0.
$$

Include signs carefully:
- Forces to the right or upward are positive.
- Counterclockwise torques are positive (by convention).

#### Step 6. Solve for Unknowns
Use the resulting algebraic equations to find the unknown forces or torques.  
Check that the number of independent equations equals the number of unknowns.

#### Step 7. Verify Results
Confirm that all equilibrium conditions are satisfied and that the answers are physically reasonable (e.g., normal forces cannot be negative unless the contact is lost).

---

### Example 12.4. Hanging Sign

A $3.0~\text{m}$ uniform beam of mass $20~\text{kg}$ supports a $30~\text{kg}$ sign at its end.  
The beam is hinged to a wall and supported by a cable attached to its end, making an angle of $37°$ with the beam.  
Find the tension in the cable and the reaction forces at the hinge.

#### Solution

**Step 1. Identify forces**

- Beam’s weight: $W_b = 20(9.8) = 196~\text{N}$, acts at midpoint.  
- Sign’s weight: $W_s = 30(9.8) = 294~\text{N}$, acts at the beam’s end.  
- Tension $T$ in the cable acts upward at $37°$.  
- Hinge reactions: $H_x$, $H_y$.

**Step 2. Torque equilibrium about hinge**

$$
\sum \tau = 0 \Rightarrow T\sin37°(3.0) - W_b(1.5) - W_s(3.0) = 0.
$$

$$
T = \frac{W_b(1.5) + W_s(3.0)}{3.0\sin37°} = \frac{(196)(1.5) + (294)(3.0)}{3.0(0.601)} = 676~\text{N}.
$$

**Step 3. Force equilibrium**

Horizontal:
$$
\sum F_x = 0 \Rightarrow H_x = T\cos37° = 676(0.799) = 540~\text{N}.
$$

Vertical:
$$
\sum F_y = 0 \Rightarrow H_y + T\sin37° - W_b - W_s = 0.
$$

$$
H_y = W_b + W_s - T\sin37° = (196 + 294) - (676)(0.601) = 490 - 406 = 84~\text{N}.
$$

**Results:**
- $T = 676~\text{N}$  
- $H_x = 540~\text{N}$  
- $H_y = 84~\text{N}$

---

### Example 12.5. Person on a Diving Board

A uniform $4.0~\text{m}$ diving board of mass $20~\text{kg}$ is supported by two pillars, one at the left end (A) and one $1.5~\text{m}$ from that end (B).  
A $70~\text{kg}$ diver stands at the free end.  
Find the forces on the supports.

#### Solution

**Step 1. Forces**
- Board weight: $W_b = 20(9.8) = 196~\text{N}$, acts at $2.0~\text{m}$ from A.
- Diver weight: $W_d = 70(9.8) = 686~\text{N}$, acts at $4.0~\text{m}$.
- Reactions: $N_A$ at A, $N_B$ at B.

**Step 2. Torque equilibrium about A**
$$
\sum \tau_A = 0 \Rightarrow N_B(1.5) - W_b(2.0) - W_d(4.0) = 0.
$$
$$
N_B = \frac{W_b(2.0) + W_d(4.0)}{1.5} = \frac{(196)(2.0) + (686)(4.0)}{1.5} = 2142~\text{N}.
$$

**Step 3. Vertical force equilibrium**
$$
\sum F_y = 0 \Rightarrow N_A + N_B - W_b - W_d = 0.
$$
$$
N_A = W_b + W_d - N_B = (196 + 686) - 2142 = -1260~\text{N}.
$$

A **negative sign** indicates that the left support exerts a **downward force** (it pushes down on the board).  
Thus:
- $N_B = 2140~\text{N}$ upward,  
- $N_A = 1260~\text{N}$ downward.

---

### Example 12.6. Supported Rod with Hanging Mass

A $2.0~\text{m}$ uniform rod of mass $5.0~\text{kg}$ is supported at its left end and by a cable attached $1.5~\text{m}$ from the left end.  
A $10~\text{kg}$ mass hangs from the right end.  
The cable makes an angle of $30°$ with the horizontal.  
Find the tension in the cable and the reaction at the support.

#### Solution

Weights:
$$
W_r = 5.0(9.8) = 49~\text{N}, \quad W_m = 10(9.8) = 98~\text{N}.
$$

Torque equilibrium about the left end:
$$
T\sin30°(1.5) - W_r(1.0) - W_m(2.0) = 0.
$$
$$
T = \frac{W_r(1.0) + W_m(2.0)}{1.5\sin30°} = \frac{(49)(1.0) + (98)(2.0)}{0.75} = \frac{245}{0.75} = 327~\text{N}.
$$

Horizontal equilibrium:
$$
H_x = T\cos30° = 327(0.866) = 283~\text{N}.
$$

Vertical equilibrium:
$$
H_y = W_r + W_m - T\sin30° = (49 + 98) - 327(0.5) = 147 - 164 = -17~\text{N}.
$$

A small negative value implies that the cable nearly supports all the weight, slightly pulling the hinge upward.

---

### Common Mistakes in Statics Problems

- **Forgetting reaction components:** At a hinge or pin, both horizontal and vertical components must be considered.  
- **Incorrect torque sign:** Always maintain a consistent convention (counterclockwise positive).  
- **Using wrong lever arm:** The perpendicular distance from the pivot to the line of action of the force determines torque, not the direct distance to the point of application.  
- **Neglecting weight of the object:** Even light members can contribute significant torque if long.

---

### Conceptual Notes

- The number of independent equilibrium equations equals the number of unknowns in planar systems (typically three).  
- A system in equilibrium can still move at constant velocity — static equilibrium requires the velocity to be **zero**.  
- For multiple-body systems, consider each component separately but ensure that action–reaction forces between them cancel internally.  
- The free-body diagram is the most critical tool — it must include all forces and moments acting on the chosen body.

---

## 12.3. Applications to Muscles and Joints

The principles of static equilibrium are not limited to bridges and beams — they also apply to the **human body**.  
Bones act as rigid levers, and muscles apply forces to maintain balance or movement.  
Because of the geometry of joints and the need for precise control, muscles often exert forces much larger than the weights they support.

In this section, we apply the conditions of equilibrium to analyze typical biomechanical systems such as the forearm, the upper arm, and the lower leg.

---

### The Human Body as a System of Levers

A **lever** consists of:
1. A **pivot point** (the joint),
2. A **rigid arm** (the bone), and
3. One or more **forces** (muscles, weights, contact forces).

Muscles usually attach close to the joint, while external loads act farther away.  
As a result, muscles must generate large forces to maintain equilibrium, despite producing only small displacements.

---

### Example 12.7. The Forearm and Biceps

The forearm is supported horizontally by the biceps muscle.  
The biceps attaches to the forearm at a point $d = 5.0~\text{cm}$ from the elbow joint.  
The forearm and hand together have a weight of $W_a = 20~\text{N}$ acting at the center of mass, $15~\text{cm}$ from the elbow.  
A $100~\text{N}$ weight is held in the hand, $35~\text{cm}$ from the elbow.  
Find the tension $T$ in the biceps and the force at the elbow joint.

#### Solution

**Step 1. Free-body diagram**

Forces acting:
- Biceps tension $T$ upward at $d = 5.0~\text{cm}$,
- Arm weight $W_a = 20~\text{N}$ downward at $r_1 = 15~\text{cm}$,
- Hand weight $W_h = 100~\text{N}$ downward at $r_2 = 35~\text{cm}$,
- Elbow joint reaction $R$ (with components $R_x$ and $R_y$).

**Step 2. Torque equilibrium about the elbow**

Take counterclockwise as positive:

$$
T(d) - W_a(r_1) - W_h(r_2) = 0.
$$

Substitute distances in meters:

$$
T(0.05) = (20)(0.15) + (100)(0.35) = 3.0 + 35.0 = 38.0~\text{N·m}.
$$

Solve for $T$:

$$
T = \frac{38.0}{0.05} = 760~\text{N}.
$$

The biceps exerts a tension of **760 N**, about eight times the weight being held.

**Step 3. Force equilibrium**

Vertical forces:

$$
R_y + T - W_a - W_h = 0 \Rightarrow R_y = W_a + W_h - T = 120 - 760 = -640~\text{N}.
$$

A negative sign means the elbow joint exerts a **downward** force of **640 N** on the forearm.

**Result:**
- $T = 760~\text{N}$ (biceps tension),
- $R_y = 640~\text{N}$ (downward joint force).

---

### Mechanical Disadvantage of Muscles

Although the biceps force is large, its small lever arm provides **fine control** and rapid response.  
Muscles trade mechanical advantage for **speed and precision**.

The **mechanical advantage** (MA) is the ratio of load arm to effort arm:

$$
\text{MA} = \frac{d_{\text{load}}}{d_{\text{muscle}}} = \frac{35}{5} = 7.
$$

Thus, the muscle must exert seven times the force of the load to maintain equilibrium.

---

### Example 12.8. Forces in the Shoulder Joint

A 3.0-kg weight is held with the arm extended horizontally, the center of gravity of the arm being 30 cm from the shoulder joint.  
The deltoid muscle attaches 12 cm from the joint at an angle of $15°$ above the arm.  
Find the tension in the deltoid and the vertical force at the shoulder joint.

#### Solution

**Step 1. Forces**

- Arm weight: $W_a = 3.0(9.8) = 29.4~\text{N}$, acts at $0.30~\text{m}$.
- Muscle tension $T$ acts upward at $0.12~\text{m}$, at $15°$ above the horizontal.
- Shoulder reaction $R_y$ acts vertically.

**Step 2. Torque equilibrium about the shoulder**

$$
T\sin15°(0.12) = W_a(0.30).
$$

$$
T = \frac{W_a(0.30)}{0.12\sin15°} = \frac{(29.4)(0.30)}{0.12(0.259)} = 285~\text{N}.
$$

**Step 3. Vertical equilibrium**

$$
R_y + T\sin15° - W_a = 0.
$$

$$
R_y = W_a - T\sin15° = 29.4 - 285(0.259) = 29.4 - 74 = -45~\text{N}.
$$

The negative sign indicates the shoulder joint exerts a **downward** force of **45 N** on the arm.

---

### Example 12.9. The Triceps and the Forearm

When the forearm extends, the **triceps muscle** applies a force near the elbow to counteract weights acting farther away.  
Suppose the triceps attaches 2.5 cm from the joint, the forearm weighs 20 N acting 15 cm away, and a 30 N weight is held 35 cm away.  
Find the force the triceps must exert to keep the forearm horizontal.

#### Solution

Torque equilibrium about the elbow:

$$
T(0.025) = (20)(0.15) + (30)(0.35) = 3.0 + 10.5 = 13.5~\text{N·m}.
$$

$$
T = \frac{13.5}{0.025} = 540~\text{N}.
$$

Thus, the triceps must exert a force of **540 N** — much greater than the weights being supported.

---

### Muscular Efficiency

Even though muscles operate at a **mechanical disadvantage**, they achieve **energetic efficiency** by working through small displacements.  
For every small shortening of the muscle, the hand moves through a much larger distance, allowing precise control of motion.

If the biceps shortens by $\Delta x_m$ while the hand rises by $\Delta x_h$, the work done by both is equal:

$$
T\Delta x_m = W\Delta x_h.
$$

Since $\Delta x_h / \Delta x_m = d_{\text{load}} / d_{\text{muscle}}$, the large mechanical disadvantage is compensated by a large motion advantage.

---

### Example 12.10. Lifting a Weight: Energy and Work

A person holds a 5-kg weight in the hand, 30 cm from the elbow joint.  
The biceps attaches 4 cm from the joint and exerts a force of 400 N.  
If the forearm rotates through $60°$, find the work done by the biceps and by the weight.

#### Solution

Arc distance of hand:
$$
s_h = (0.30)(60° \times \pi/180) = 0.314~\text{m}.
$$

Arc distance of biceps insertion:
$$
s_m = (0.04)(60° \times \pi/180) = 0.042~\text{m}.
$$

Work by muscle:
$$
W_m = T s_m = 400(0.042) = 16.8~\text{J}.
$$

Work by weight (negative):
$$
W_w = -mgh = -(5)(9.8)(0.314) = -15.4~\text{J}.
$$

The work magnitudes are nearly equal, showing that most of the muscle’s effort balances the gravitational torque.

---

### Conceptual Notes

- Muscles typically operate at **mechanical disadvantage**, producing large forces over short distances.  
- The **torque** generated by a muscle depends on both the **magnitude** of its tension and its **lever arm** about the joint.  
- Even though the forces in joints can be very large, they are balanced by corresponding reaction forces, keeping the system in equilibrium.  
- These static analyses assume steady positions; in dynamic motion, additional inertial torques must be considered.

---

## 12.4. Stability and Balance

Equilibrium can exist under different conditions of **stability**.  
An object may return to equilibrium after a small disturbance, move farther away from it, or remain in its new position.  
Understanding these behaviors is crucial in designing structures, vehicles, and in analyzing the balance of the human body.

---

### Types of Equilibrium

There are **three types** of mechanical equilibrium, depending on how the center of gravity (CG) moves when the object is displaced slightly.

1. **Stable equilibrium:**  
   When displaced, the object experiences a restoring torque or force that returns it to its original position.

   - The center of gravity rises when displaced.  
   - Potential energy increases, then decreases back to the minimum when released.

   Example: A pendulum at its lowest point, or a cone resting on its base.

2. **Unstable equilibrium:**  
   A small displacement causes the object to move farther from equilibrium.

   - The center of gravity lowers when displaced.  
   - Potential energy decreases, so the object continues to move away.

   Example: A pencil balanced on its tip, or a cone resting on its vertex.

3. **Neutral equilibrium:**  
   When displaced, the object neither returns nor moves farther away — it stays in the new position.

   - The center of gravity remains at the same height.  
   - Potential energy stays constant.

   Example: A ball on a flat horizontal surface.

---

### Center of Gravity and Stability

The **center of gravity (CG)** is the point at which the entire weight of an object can be considered to act.  
An object remains in equilibrium as long as the vertical line through its CG falls **within its base of support**.

If this line passes outside the base, a **torque due to gravity** causes the object to tip over.

---

### Condition for Stability

Consider an object of weight $W$ resting on a base.  
When tilted by a small angle $\theta$, the CG moves horizontally by a distance $x = h\sin\theta$, where $h$ is the height of the CG above the pivot point.

The restoring torque about the pivot is:

$$
\tau = W x = W h \sin\theta.
$$

- If the torque acts to restore the object to equilibrium, the position is **stable**.
- If it acts to increase the displacement, the position is **unstable**.

---

### Example 12.11. Stability of a Tipping Block

A uniform block of height $h$ and base width $b$ rests on a horizontal surface.  
At what tilt angle $\theta_c$ will it begin to tip over?

#### Solution

When the block just begins to tip, the line of action of its weight passes through the pivot at the lower edge.

From geometry:

$$
\tan\theta_c = \frac{b/2}{h}.
$$

Thus, the critical angle is:

$$
\theta_c = \tan^{-1}\left(\frac{b}{2h}\right).
$$

For example, if $b = 0.20~\text{m}$ and $h = 0.50~\text{m}$:

$$
\theta_c = \tan^{-1}\left(\frac{0.20}{1.00}\right) = 11.3°.
$$

So the block will tip when tilted more than **11.3°**.

---

### Potential Energy and Stability

The condition of equilibrium can be analyzed in terms of potential energy $U$.  
For small displacements:

- **Stable equilibrium:** $U$ has a **minimum** (second derivative positive).
- **Unstable equilibrium:** $U$ has a **maximum** (second derivative negative).
- **Neutral equilibrium:** $U$ is **constant**.

This is analogous to a ball resting at the bottom, top, or flat portion of a curved surface.

---

### Example 12.12. Ball in a Bowl

A ball of mass $m$ rests in a hemispherical bowl of radius $R$.  
If displaced horizontally by a small distance $x$, find the restoring force.

#### Solution

When displaced, the center of mass rises by $h \approx \frac{x^2}{2R}$ for small $x$.  
The increase in potential energy is:

$$
U = mg h = \frac{mgx^2}{2R}.
$$

The restoring force is:

$$
F = -\frac{dU}{dx} = -\frac{mg}{R}x.
$$

This is the equation of a **simple harmonic oscillator**, showing the motion is stable and oscillatory about equilibrium.

---

### Stability in Human Balance

The human body maintains balance by keeping its **center of gravity** above its **support base** (the area between the feet).

- Standing upright: CG is above the midpoint of the feet.  
- Leaning: CG shifts toward the edge of the base; muscles generate restoring torques to keep balance.  
- Falling: Occurs when CG passes outside the base of support.

Increasing the **base width** or **lowering the CG** improves stability — a principle used in athletics and engineering design alike.

---

### Example 12.13. Balance While Carrying a Load

A person of mass $70~\text{kg}$ holds a $20~\text{kg}$ weight at arm’s length, 0.70 m in front of the body.  
Where must the person’s center of gravity shift to maintain balance?

#### Solution

The combined torques about the feet must sum to zero.

Let $x$ be the horizontal shift of the body’s center of gravity backward.

$$
(70)(9.8)x = (20)(9.8)(0.70).
$$

$$
x = \frac{20(0.70)}{70} = 0.20~\text{m}.
$$

Thus, the person must lean **20 cm backward** to remain balanced.

---

### Design Principles for Stability

To improve stability:
1. **Lower the center of gravity ($h$):** decreases the potential for tipping.
2. **Increase the base width ($b$):** increases $\theta_c$ and the support area.
3. **Widen the mass distribution:** increases moment of inertia, resisting rotational acceleration.
4. **Ensure weight acts within the base area:** prevents net torque from developing.

---

### Example 12.14. Stability of a Car on a Hill

A car of width $1.6~\text{m}$ and height of CG $0.80~\text{m}$ drives on a slope.  
At what maximum angle $\theta$ will it begin to tip sideways?

#### Solution

Critical condition:

$$
\tan\theta_c = \frac{b/2}{h} = \frac{0.80}{1.6} = 0.50.
$$

$$
\theta_c = \tan^{-1}(0.5) = 26.6°.
$$

Thus, the car will tip sideways at slopes greater than **26.6°**.

---

### Conceptual Notes

- The **stability** of an equilibrium depends on the position of the center of gravity relative to the base of support.  
- When the vertical projection of the CG passes **outside** the base, the system becomes **unstable** and tips.  
- Lowering the CG or widening the base enhances stability.  
- In energy terms, equilibrium is stable when potential energy is at a **minimum**.  
- Human posture, animal stance, and engineered designs (cars, towers, robots) all rely on maintaining stable equilibrium.

---

In the next section, we will examine how materials deform under stress, introducing the concepts of **elasticity, stress, and strain**, which determine the strength and flexibility of physical structures.


## 12.5. Elasticity; Stress and Strain

In real materials, applied forces do not always cause motion — they often produce **deformation** instead.  
When the force is removed, the material may return to its original shape.  
This property of returning to the original configuration after being deformed is called **elasticity**.  
If the deformation remains, the material has undergone **plastic (permanent) deformation**.

Understanding elasticity is essential for engineering, architecture, and biomechanics, as it determines how structures and materials respond to forces without breaking.

---

### Deformation and Elastic Behavior

When an external force acts on a solid body, the object changes its dimensions slightly.  
The type of deformation depends on the nature of the applied force:

- **Tensile (stretching)** – forces act to elongate the body.  
- **Compressive** – forces act to shorten or squeeze it.  
- **Shear** – forces act tangentially, distorting the shape.  
- **Torsional** – forces twist the body about an axis.

If the force is removed and the body returns to its original shape, the deformation is **elastic**.  
For small deformations, most materials obey **Hooke’s Law**, where deformation is proportional to the applied force.

---

### Hooke’s Law (General Form)

Within the elastic limit, the **stress** is directly proportional to the **strain**:

$$
\text{Stress} \propto \text{Strain}, \quad \text{or} \quad \frac{\text{Stress}}{\text{Strain}} = \text{constant}.
$$

The constant of proportionality is the **elastic modulus**, which depends on the type of deformation and the material.

---

### Stress and Strain Defined

1. **Stress (Force per Unit Area)**  
   Stress quantifies the intensity of the internal forces that resist deformation:

   $$
   \text{Stress} = \frac{F}{A},
   $$

   where:
   - $F$ is the external force,
   - $A$ is the cross-sectional area.

   Units: **N/m² = Pa (Pascal)**.

2. **Strain (Relative Deformation)**  
   Strain measures the fractional change in the object’s dimensions:

   $$
   \text{Strain} = \frac{\Delta L}{L_0},
   $$

   where:
   - $\Delta L$ is the change in length,
   - $L_0$ is the original length.

   Strain is dimensionless.

---

### Types of Elastic Moduli

There are three principal types of elasticity constants depending on the type of deformation:

1. **Young’s Modulus (E)** — longitudinal stretching or compression  
   $$
   E = \frac{\text{Stress}}{\text{Strain}} = \frac{F/A}{\Delta L / L_0} = \frac{F L_0}{A \Delta L}.
   $$

2. **Shear Modulus (G)** — deformation due to tangential (shear) forces  
   $$
   G = \frac{\text{Shear Stress}}{\text{Shear Strain}} = \frac{F/A}{x/h} = \frac{Fh}{A x}.
   $$

3. **Bulk Modulus (B)** — deformation due to uniform compression (change in volume)  
   $$
   B = -\frac{\text{Stress}}{\text{Volume Strain}} = -\frac{\Delta P}{\Delta V / V_0}.
   $$

The negative sign indicates that an increase in pressure ($\Delta P > 0$) leads to a **decrease** in volume ($\Delta V < 0$).

---

### Example 12.15. Stretching a Steel Wire

A steel wire of diameter $1.0~\text{mm}$ and length $2.0~\text{m}$ supports a load of $10~\text{kg}$.  
How much does the wire stretch?  
(Take $E_{\text{steel}} = 2.0\times10^{11}~\text{Pa}$.)

#### Solution

Given:
$$
F = mg = (10)(9.8) = 98~\text{N}, \quad A = \pi r^2 = \pi(0.0005)^2 = 7.85\times10^{-7}~\text{m}^2.
$$

$$
\Delta L = \frac{F L_0}{A E} = \frac{(98)(2.0)}{(7.85\times10^{-7})(2.0\times10^{11})} = 1.25\times10^{-3}~\text{m} = 1.25~\text{mm}.
$$

The wire stretches by **1.25 mm** under the 10-kg load.

---

### Example 12.16. Shear Deformation of a Bolt

A bolt with cross-sectional area $A = 8.0\times10^{-5}~\text{m}^2$ holds two plates together.  
A shear force of $F = 400~\text{N}$ is applied.  
If the distance between plates is $h = 5.0~\text{mm}$ and $G = 8.0\times10^{10}~\text{Pa}$, find the relative displacement of the plates.

#### Solution

Shear strain:
$$
\frac{x}{h} = \frac{F/A}{G} = \frac{400 / (8.0\times10^{-5})}{8.0\times10^{10}} = \frac{5.0\times10^6}{8.0\times10^{10}} = 6.25\times10^{-5}.
$$

Displacement:
$$
x = h \times 6.25\times10^{-5} = (5.0\times10^{-3})(6.25\times10^{-5}) = 3.13\times10^{-7}~\text{m}.
$$

The top plate shifts by only **0.31 μm**, showing the rigidity of steel under shear.

---

### Example 12.17. Volume Compression of Water

The bulk modulus of water is $B = 2.2\times10^9~\text{Pa}$.  
Find the fractional decrease in volume of a water sample under a pressure increase of $3.0\times10^6~\text{Pa}$.

#### Solution

$$
\frac{\Delta V}{V_0} = -\frac{\Delta P}{B} = -\frac{3.0\times10^6}{2.2\times10^9} = -1.36\times10^{-3}.
$$

Thus, the volume decreases by only **0.136%**, confirming that water is nearly incompressible.

---

### Elastic Energy Stored in a Stretched Wire

The work done in stretching an elastic material is stored as **potential energy**.  
If the force increases linearly with elongation (Hooke’s law), the elastic energy is:

$$
U = \frac{1}{2}F\Delta L = \frac{1}{2}\frac{E A}{L_0}(\Delta L)^2.
$$

The corresponding **energy density** (energy per unit volume) is:

$$
u = \frac{U}{A L_0} = \frac{1}{2}E\left(\frac{\Delta L}{L_0}\right)^2 = \frac{1}{2}E(\text{strain})^2.
$$

---

### Example 12.18. Energy in a Guitar String

A steel string ($L_0 = 0.60~\text{m}$, diameter $1.0~\text{mm}$, $E = 2.0\times10^{11}~\text{Pa}$) is stretched by $1.2~\text{mm}$.  
Find the tension and elastic energy stored.

#### Solution

$$
A = \pi(0.0005)^2 = 7.85\times10^{-7}~\text{m}^2.
$$

Tension:
$$
F = \frac{E A \Delta L}{L_0} = \frac{(2.0\times10^{11})(7.85\times10^{-7})(1.2\times10^{-3})}{0.60} = 314~\text{N}.
$$

Energy:
$$
U = \frac{1}{2}F\Delta L = 0.5(314)(1.2\times10^{-3}) = 0.19~\text{J}.
$$

The stretched string stores about **0.19 J** of potential energy.

---

### Stress–Strain Curve

When a material is tested by gradually increasing stress, the **stress–strain graph** typically exhibits several regions:

1. **Linear (Elastic) region:**  
   Stress and strain are proportional (Hooke’s law applies).  
   The slope equals the elastic modulus.

2. **Yield point:**  
   Material begins to deform permanently; Hooke’s law no longer holds.

3. **Plastic region:**  
   Permanent deformation occurs; material hardens until fracture.

4. **Fracture point:**  
   Material breaks when maximum stress is exceeded.

---

### Typical Elastic Moduli

| Material | Young’s Modulus (Pa) | Shear Modulus (Pa) | Bulk Modulus (Pa) |
|-----------|----------------------|--------------------|-------------------|
| Steel | 2.0×10¹¹ | 8.0×10¹⁰ | 1.6×10¹¹ |
| Copper | 1.1×10¹¹ | 4.2×10¹⁰ | 1.4×10¹¹ |
| Aluminum | 7.0×10¹⁰ | 2.5×10¹⁰ | 7.5×10¹⁰ |
| Glass | 7.0×10¹⁰ | 3.0×10¹⁰ | 3.5×10¹⁰ |
| Concrete | 2.0×10¹⁰ | — | — |
| Bone (human) | 1.5×10¹⁰ | — | — |
| Rubber | 5.0×10⁶ | 3.0×10⁶ | 1.6×10⁹ |
| Water | — | — | 2.2×10⁹ |

---

### Conceptual Notes

- **Stress** measures internal forces; **strain** measures relative deformation.  
- Within the **elastic limit**, stress and strain are proportional (Hooke’s law).  
- Beyond the **yield point**, deformation becomes permanent.  
- Stiff materials (like steel) have high elastic moduli; soft materials (like rubber) have low moduli.  
- Elastic energy is stored during deformation and recovered when the material returns to equilibrium.

## 12.6. Fracture

When a material is subjected to increasing stress, it eventually reaches a point where it can no longer sustain the load — it **fractures**.  
Fracture marks the transition from elastic or plastic deformation to complete structural failure.  
Understanding the factors that lead to fracture is essential for the design of safe structures, tools, and biological tissues.

---

### The Breaking Stress

Every material has a characteristic **breaking stress** (also called **ultimate tensile strength**).  
This is the maximum stress that the material can withstand before breaking.

$$
\sigma_{\text{break}} = \frac{F_{\text{max}}}{A}.
$$

Beyond this limit, cracks propagate rapidly through the material, and failure occurs.

| Material | Ultimate Tensile Strength (Pa) |
|-----------|--------------------------------|
| Steel (high-carbon) | 1.8×10⁹ |
| Aluminum | 5.0×10⁸ |
| Concrete | 2.0×10⁷ |
| Bone (human femur) | 1.3×10⁸ |
| Spider silk | 1.0×10⁹ |

---

### Stress–Strain Behavior and Fracture Point

A **stress–strain curve** provides a visual representation of a material’s behavior under tension.

1. **Elastic region:**  
   The curve is linear; stress is proportional to strain (Hooke’s law).  
   The slope gives Young’s modulus $E$.

2. **Yield point:**  
   Marks the onset of plastic deformation; strain increases faster than stress.

3. **Plastic region:**  
   Material elongates permanently; internal microcracks begin to form.

4. **Ultimate tensile stress (UTS):**  
   The highest point on the curve — maximum stress the material can bear.

5. **Fracture point:**  
   Material fails and breaks apart; stress suddenly drops to zero.

---

### Ductile vs. Brittle Materials

**Ductile materials** (e.g., steel, copper, aluminum) exhibit significant plastic deformation before breaking.  
They can be drawn into wires or reshaped under stress.

**Brittle materials** (e.g., glass, ceramics, concrete) fracture suddenly with little or no plastic deformation.  
They tend to fail catastrophically once cracks initiate.

---

### Example 12.19. Tensile Stress in a Steel Cable

A steel cable of cross-sectional area $A = 2.5~\text{cm}^2$ supports a $2000~\text{kg}$ load.  
Determine the stress in the cable and whether it is safe if the breaking stress is $1.5\times10^9~\text{Pa}$.

#### Solution

$$
A = 2.5\times10^{-4}~\text{m}^2, \quad F = mg = (2000)(9.8) = 1.96\times10^4~\text{N}.
$$

Stress:
$$
\sigma = \frac{F}{A} = \frac{1.96\times10^4}{2.5\times10^{-4}} = 7.84\times10^7~\text{Pa}.
$$

The stress is about **78 MPa**, which is well below the breaking stress of **1.5 GPa**, so the cable is safe with a large safety margin.

---

### The Safety Factor

Engineers never design structures to operate near the breaking stress.  
Instead, they apply a **safety factor** (SF) to provide a margin against unexpected loads, imperfections, or fatigue.

$$
\text{Allowable Stress} = \frac{\sigma_{\text{break}}}{\text{SF}}.
$$

Typical safety factors:
- **Bridges, cranes:** SF = 5 to 10  
- **Aircraft parts:** SF = 1.5 to 3  
- **Biomedical implants:** SF ≈ 3  

---

### Example 12.20. Safety Factor in a Suspension Cable

A bridge cable has a breaking stress of $1.2\times10^9~\text{Pa}$ and is designed with SF = 6.  
What maximum load per unit area can it safely support?

#### Solution

$$
\sigma_{\text{safe}} = \frac{\sigma_{\text{break}}}{\text{SF}} = \frac{1.2\times10^9}{6} = 2.0\times10^8~\text{Pa}.
$$

Hence, the cable must be designed so that the stress never exceeds **200 MPa** under normal load.

---

### Fatigue and Microcracks

Repeated or cyclic loading below the breaking stress can still cause failure over time — a phenomenon known as **fatigue**.

- Microscopic cracks form and propagate with each stress cycle.  
- Once cracks reach a critical size, fracture occurs suddenly.  
- Fatigue failure explains why rotating machinery, airplane wings, or bridges can fail unexpectedly even when stresses are moderate.

Factors that affect fatigue:
- Frequency and amplitude of load cycles,  
- Temperature and corrosion,  
- Surface roughness and imperfections.

---

### Brittle Fracture and Crack Propagation

In brittle materials, fracture begins at microscopic flaws or **stress concentrators** (scratches, voids, or sharp corners).  
The local stress near a crack tip can be much larger than the average applied stress:

$$
\sigma_{\text{local}} \approx \sigma_{\text{applied}}\left(1 + 2\sqrt{\frac{a}{r}}\right),
$$

where:
- $a$ = half-length of the crack,
- $r$ = radius of curvature at the crack tip.

Reducing sharp edges or surface imperfections helps prevent premature fracture.

---

### Example 12.21. Estimating Stress Near a Crack

A glass plate under tension experiences an average stress of $2.0\times10^7~\text{Pa}$.  
If it contains a surface crack of half-length $a = 0.5~\text{mm}$ and tip radius $r = 1~\mu\text{m}$, estimate the local stress at the crack tip.

#### Solution

$$
\sigma_{\text{local}} = 2.0\times10^7\left(1 + 2\sqrt{\frac{a}{r}}\right)
= 2.0\times10^7\left(1 + 2\sqrt{\frac{5.0\times10^{-4}}{1.0\times10^{-6}}}\right)
= 2.0\times10^7(1 + 2\sqrt{500}) = 2.0\times10^7(1 + 44.7) = 9.1\times10^8~\text{Pa}.
$$

This local stress exceeds the breaking stress of glass (~$7\times10^7~\text{Pa}$), explaining why even small cracks can cause sudden failure.

---

### Brittle vs. Ductile Fracture Patterns

- **Brittle fracture:**  
  Occurs suddenly, with little warning.  
  Fracture surfaces are smooth and flat, often following grain boundaries.

- **Ductile fracture:**  
  Preceded by necking and plastic deformation.  
  Fracture surfaces are rough and fibrous, absorbing much more energy before failure.

---

### Example 12.22. Bone Fracture under Bending

A human femur (cross-section $A = 5.0~\text{cm}^2$) breaks when the compressive stress reaches $1.3\times10^8~\text{Pa}$.  
Estimate the maximum force it can support.

#### Solution

$$
A = 5.0\times10^{-4}~\text{m}^2, \quad \sigma_{\text{break}} = 1.3\times10^8~\text{Pa}.
$$

$$
F_{\text{max}} = \sigma_{\text{break}} A = (1.3\times10^8)(5.0\times10^{-4}) = 6.5\times10^4~\text{N}.
$$

Thus, the femur can support about **65,000 N** (roughly six times body weight) before breaking.

---

### Conceptual Notes

- **Breaking stress** is a material constant independent of geometry (for a given material and temperature).  
- The **safety factor** ensures that operational stresses remain well below breaking limits.  
- **Fatigue** and **crack growth** can lead to failure even when stresses are within the elastic range.  
- Materials with greater **ductility** can absorb more energy before fracture.  
- **Brittle materials** fail suddenly without warning — safety margins must be larger.

---

## 12.7. Trusses and Bridges

Large structures such as roofs, cranes, and bridges rely on a network of slender members joined together — called **trusses** — to support heavy loads efficiently.  
Each member in a truss is designed to carry either **tension** or **compression**, while the joints are assumed to act as **frictionless pins**.  
In this section, we apply the equilibrium conditions to determine the forces in truss members and understand how such structures remain stable.

---

### The Truss: Basic Assumptions

A **truss** is an assembly of straight members connected at their ends by frictionless pins.  
The main simplifying assumptions are:

1. All members are **connected only at their ends**.  
2. **Loads and reactions** act only at the joints.  
3. Each member is either in **tension** (pulling apart) or **compression** (pushing together).  
4. The **weight of members** is small compared with the applied loads and is often neglected.

Under these conditions, every joint of the truss can be treated as a **particle in equilibrium**, and the forces in the members can be found using the laws of statics.

---

### Methods of Analysis

There are two principal approaches:

1. **Method of Joints:**  
   - Apply $\sum F_x = 0$ and $\sum F_y = 0$ at each joint.  
   - Useful for finding the forces in all members of a small truss.

2. **Method of Sections:**  
   - Cut through the truss to expose a few members of interest.  
   - Apply $\sum F_x = 0$, $\sum F_y = 0$, and $\sum \tau = 0$ to the resulting section.  
   - More efficient for finding only a few specific member forces.

---

### Example 12.23. Simple Triangular Truss

A simple triangular truss supports a $1000~\text{N}$ load at its apex $C$.  
It is pinned at $A$ and supported by a roller at $B$.  
Each side of the triangle is $3.0~\text{m}$.

Find the forces in members $AC$, $BC$, and $AB$.

#### Step 1. Free-body diagram of the whole truss

Because the structure is symmetric:
$$
R_A = R_B = \frac{1000}{2} = 500~\text{N}.
$$

#### Step 2. Analyze joint $C$

Forces at $C$:
- Downward load $1000~\text{N}$,
- Tension or compression in $AC$ and $BC$ (assume tension initially).

Geometry: sides form $60°$ angles with the base.

Vertical equilibrium:
$$
2F_{AC}\sin60° = 1000.
$$

$$
F_{AC} = F_{BC} = \frac{1000}{2\sin60°} = \frac{1000}{1.732} = 577~\text{N}.
$$

Horizontal equilibrium:
$$
F_{AB} = F_{AC}\cos60° - F_{BC}\cos60° = 0,
$$
so the base member $AB$ is **in compression or zero force**, depending on external loads.

Result:
- $F_{AC} = F_{BC} = 577~\text{N}$ (tension),
- $F_{AB} = 0$ (no load).

---

### Example 12.24. Asymmetric Truss

A truss with base $AB = 4.0~\text{m}$ and height $2.0~\text{m}$ supports a $1200~\text{N}$ load at the apex $C$.  
Supports: pin at $A$, roller at $B$.  
Find the forces in $AC$, $BC$, and $AB$.

#### Step 1. Reaction forces

Because the load acts vertically at $C$, symmetry does not apply.  
Taking torques about $A$:

$$
R_B(4.0) = 1200(2.0) \Rightarrow R_B = 600~\text{N}.
$$
$$
R_A = 1200 - 600 = 600~\text{N}.
$$

#### Step 2. Joint $C$

Vertical equilibrium:
$$
F_{AC}\sin\theta_1 + F_{BC}\sin\theta_2 = 1200.
$$

Horizontal equilibrium:
$$
F_{AC}\cos\theta_1 = F_{BC}\cos\theta_2.
$$

From geometry:
$$
\tan\theta_1 = \frac{2.0}{2.0} = 1.0 \Rightarrow \theta_1 = 45°, \quad
\tan\theta_2 = \frac{2.0}{2.0} = 45°.
$$

Thus, $\theta_1 = \theta_2 = 45°$ (symmetrical).  
So:
$$
2F_{AC}\sin45° = 1200 \Rightarrow F_{AC} = 424~\text{N}.
$$

Hence $F_{BC} = 424~\text{N}$, both in **tension**.

---

### Example 12.25. Bridge Truss Section (Method of Sections)

In a bridge truss, a diagonal member connects joints $B$ and $D$.  
The truss is loaded with a $400~\text{N}$ downward force at $C$.  
Each span is $3.0~\text{m}$.  
Find the force in the diagonal $BD$.

#### Step 1. Reaction forces

By symmetry, $R_A = R_E = 200~\text{N}$.

#### Step 2. Cut through the section

Cut through members $BD$, $BC$, and $CD$.  
Take moments about point $C$ (to eliminate $BC$ and $CD$):

$$
\sum \tau_C = 0 = R_A(6.0) - F_{BD}(3.0\sin45°) - 400(3.0).
$$

$$
F_{BD} = \frac{R_A(6.0) - 400(3.0)}{3.0\sin45°}
= \frac{(200)(6.0) - 1200}{2.12} = \frac{0}{2.12} = 0.
$$

Thus, $F_{BD} = 0$.  
In some loading configurations, certain truss members experience **no force** — they serve to stabilize the structure geometrically.

---

### Tension and Compression in Members

- **Tension:** Members are being stretched; internal forces pull **away** from the joint.  
- **Compression:** Members are being shortened; internal forces push **toward** the joint.

In free-body diagrams:
- Tension arrows point **away** from the joint.
- Compression arrows point **toward** the joint.

---

### Example 12.26. Determining Member Forces by Inspection

In a simple truss, any member connected to a joint that has only **two collinear members and no external load** is a **zero-force member**.

Such members are important for **stability** under different load conditions but carry no load in the given configuration.

---

### Design Considerations for Trusses and Bridges

1. **Triangular Geometry:**  
   Triangles are inherently stable — their shape cannot change without altering member lengths.

2. **Load Path:**  
   Loads are transmitted from the point of application to the supports through members under tension or compression.

3. **Efficiency:**  
   Using slender members reduces weight while maintaining strength.

4. **Safety Factor:**  
   Each member is designed to carry stresses below a safe fraction of its yield strength.

5. **Redundancy:**  
   Extra members are often added to provide alternative load paths in case of failure.

---

### Example 12.27. Determining the Load on a Bridge Cable

A suspension bridge cable spans $120~\text{m}$ between supports and carries a uniformly distributed load of $w = 20,000~\text{N/m}$.  
Find the horizontal component of the tension at the lowest point if the cable forms a parabolic arc with a central sag $h = 12~\text{m}$.

#### Solution

For a parabolic cable under uniform load, the horizontal tension is:

$$
T_H = \frac{w L^2}{8h}.
$$

$$
T_H = \frac{(20,000)(120)^2}{8(12)} = \frac{(20,000)(14,400)}{96} = 3.0\times10^9~\text{N}.
$$

Hence, the horizontal tension at midspan is **3.0×10⁹ N**, illustrating the enormous forces in bridge cables.

---

### Conceptual Notes

- Trusses efficiently distribute loads using members under pure tension or compression.  
- Triangular arrangements prevent shape distortion and provide high rigidity.  
- The **method of joints** and **method of sections** are systematic ways to solve for member forces.  
- Members carrying **zero force** still play an essential role in maintaining structural stability.  
- Safety and redundancy are built into all large truss and bridge designs to prevent catastrophic failure.

---

## 12.8. Arches and Domes

Arches and domes are curved structural forms that efficiently support heavy loads primarily through **compression**.  
Unlike beams and trusses, which experience both tension and bending, arches and domes are designed so that most of the internal forces are **compressive**, minimizing tensile stress and allowing materials such as stone, brick, or concrete to be used effectively.

These structures have been employed since antiquity — from Roman aqueducts to modern bridges and cathedrals — and remain fundamental in architecture and engineering.

---

### The Principle of the Arch

An **arch** is a curved structure that spans an opening and supports a load from above.  
The forces from the load are transmitted along the curve of the arch into its **supports (abutments)**, which provide equal and opposite reaction forces.

Because the material is mainly in compression, arches can support much greater loads than straight beams of the same material.

---

### Load Transmission in an Arch

When a vertical load acts on an arch, each small element of the arch transmits the force to its neighbors.  
This produces a **compressive force** along the curve and a small **horizontal thrust** at the supports.

The supports must resist both:
- The **vertical reaction** (weight of the structure and applied load), and  
- The **horizontal thrust**, which tends to push the supports outward.

If not properly contained, this horizontal component can cause the arch to collapse.

---

### Conditions for Equilibrium of an Arch Segment

For a segment of an arch to be in equilibrium:
$$
\sum F_x = 0, \quad \sum F_y = 0, \quad \sum \tau = 0.
$$

These conditions ensure that:
1. The **compressive forces** within the arch balance the applied loads, and  
2. The **reactions at the supports** counter the outward thrust.

---

### Example 12.28. Horizontal Thrust in a Semicircular Arch

A semicircular stone arch of radius $R = 5.0~\text{m}$ supports a uniform load (its own weight plus additional) of $w = 20,000~\text{N/m}$.  
Find the horizontal reaction at each support.

#### Solution

For a semicircular arch carrying a uniform load along its span, the horizontal thrust $H$ at each support is approximately:

$$
H = \frac{w R}{\pi}.
$$

Substitute:

$$
H = \frac{(20,000)(5.0)}{3.1416} = 3.18\times10^4~\text{N}.
$$

Thus, each support must resist a **horizontal thrust of 31.8 kN** in addition to the vertical load.

---

### Example 12.29. Parabolic Arch with Uniform Load

A parabolic arch with a span $L = 40~\text{m}$ and central height $h = 10~\text{m}$ carries a uniformly distributed load $w = 5.0\times10^3~\text{N/m}$.  
Find the horizontal thrust at the supports.

#### Solution

For a parabolic arch under uniform load:

$$
H = \frac{w L^2}{8h}.
$$

Substitute:

$$
H = \frac{(5.0\times10^3)(40)^2}{8(10)} = \frac{(5.0\times10^3)(1600)}{80} = 1.0\times10^5~\text{N}.
$$

Hence, the horizontal thrust at each support is **100 kN**.

---

### Stability of Arches

The **stability** of an arch depends on how well the internal compressive forces follow the **line of thrust** — the path of resultant forces through the arch.

- If the line of thrust lies entirely **within the arch**, it remains stable.  
- If it passes **outside** the structure, tensile stresses develop and cracks may appear, leading to failure.

Architectural designs (e.g., Gothic pointed arches) help ensure the line of thrust stays within the arch geometry under typical loads.

---

### Buttresses and Tie Rods

To resist the **outward horizontal thrust**, engineers use:

- **Buttresses:** Massive supports built at the sides of the arch.  
  - Common in cathedrals and bridges.  
- **Tie rods:** Steel rods connecting the arch bases, converting horizontal thrust into **tension** rather than outward push.  
  - Common in modern arch bridges.

These structural features allow slender arches to carry large loads without spreading at their bases.

---

### Example 12.30. Thrust Reduction by a Tie Rod

A parabolic arch of span $L = 30~\text{m}$ and height $h = 6.0~\text{m}$ carries a uniform load $w = 3.0\times10^4~\text{N/m}$.  
If a horizontal tie rod connects the two ends, what is the tension in the rod?

#### Solution

The tie rod must carry the horizontal thrust $H$ that each support would otherwise resist:

$$
H = \frac{w L^2}{8h} = \frac{(3.0\times10^4)(30)^2}{8(6)} = \frac{(3.0\times10^4)(900)}{48} = 5.63\times10^5~\text{N}.
$$

Hence, the tie rod must sustain a tension of **563 kN**.

---

### Domes: Three-Dimensional Arches

A **dome** can be thought of as a three-dimensional arch rotated around its vertical axis.  
Like an arch, it carries loads primarily through **compression**.

- The **vertical component** of compression supports the load.  
- The **horizontal component** produces an outward thrust along the base, which must be resisted by the supporting ring or foundations.

If the horizontal thrust is not properly balanced, the dome may crack or collapse outward.

---

### Example 12.31. Horizontal Force at the Base of a Dome

A hemispherical concrete dome of radius $R = 10~\text{m}$ weighs $5.0\times10^6~\text{N}$.  
Estimate the horizontal thrust at its base.

#### Solution

For a hemispherical dome under its own weight, the horizontal thrust at the base is approximately:

$$
H = \frac{W}{\pi R}.
$$

$$
H = \frac{5.0\times10^6}{3.1416(10)} = 1.59\times10^5~\text{N}.
$$

Each meter of the base circumference must resist an **outward thrust of about 1.6×10⁵ N/m**.

---

### Structural Advantages of Arches and Domes

1. **Efficient load distribution:**  
   Forces flow smoothly through compression without generating bending moments.

2. **Material economy:**  
   Since tension is minimized, materials with high compressive strength (stone, brick, concrete) can be used effectively.

3. **Aesthetic and functional design:**  
   Curved forms provide both structural strength and architectural elegance.

4. **Scalability:**  
   From small doorways to massive bridges or stadium roofs, the same principles apply.

---

### Example 12.32. The Arch Bridge

An arch bridge of span $L = 100~\text{m}$ and rise $h = 20~\text{m}$ supports a uniform load of $w = 2.5\times10^4~\text{N/m}$.  
Find the horizontal thrust at the supports and comment on its significance.

#### Solution

$$
H = \frac{w L^2}{8h} = \frac{(2.5\times10^4)(100)^2}{8(20)} = \frac{(2.5\times10^4)(10,000)}{160} = 1.56\times10^6~\text{N}.
$$

The supports must resist a **horizontal thrust of 1.56 MN**, showing that the foundations must be extremely strong or reinforced by tie elements.

---

### Historical and Modern Examples

- **Roman aqueducts:** Early semicircular stone arches.  
- **Gothic cathedrals:** Pointed arches and flying buttresses.  
- **Modern concrete bridges:** Parabolic arches with internal steel reinforcement.  
- **Geodesic domes:** Lightweight frameworks distributing compression through interconnected triangles.

These structures illustrate the enduring utility of the arch and dome principle.

---

### Conceptual Notes

- Arches and domes carry loads mainly by **compression**, minimizing bending stresses.  
- The **horizontal thrust** at the supports must be balanced by abutments or tie rods.  
- The **line of thrust** should remain within the structure for stability.  
- Curved geometry distributes forces more uniformly than flat beams.  
- Modern materials (reinforced concrete, steel) allow for thinner and larger arches than those possible with stone alone.
