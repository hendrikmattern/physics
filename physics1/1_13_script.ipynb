## 17.1. Atomic Theory of Matter

The atomic theory of matter states that all matter is composed of extremely small particles called atoms. Atoms typically have sizes on the order of $10^{-10}\,\text{m}$. Although atoms are far too small to be observed directly with ordinary optical microscopes, a wide range of physical and chemical evidence confirms their existence.

Atoms can combine to form molecules, which are the smallest units of a substance that retain its chemical properties. For example, a water molecule consists of two hydrogen atoms and one oxygen atom. The properties of matter in different phases arise primarily from the arrangement, motion, and interaction strength of atoms or molecules.

In a solid, atoms or molecules are closely packed and held together by strong intermolecular forces. They vibrate about fixed equilibrium positions but do not move freely through the material. In a liquid, the particles are still close together, but the intermolecular forces are weaker, allowing the particles to move past one another so the substance can flow. In a gas, atoms or molecules are much farther apart, the intermolecular forces are very weak, and the particles move freely throughout the container.

Atomic and molecular masses are measured using a relative scale. By convention, the most common isotope of carbon, carbon-12, is assigned an atomic mass of exactly $12.0000\,\text{u}$, where $\text{u}$ is the atomic mass unit. One atomic mass unit is defined as
$$
1\,\text{u} = \frac{1}{12} \text{ of the mass of a } {}^{12}\text{C atom}.
$$

On this scale, the atomic mass of an element represents the weighted average of the masses of its naturally occurring isotopes. The molecular mass of a molecule is the sum of the atomic masses of all the atoms that make it up. For example, the molecular mass of carbon dioxide is
$$
12 + 2(16) = 44\,\text{u}.
$$

The atomic theory provides a microscopic explanation for many macroscopic properties of matter. Density depends on the mass and spacing of atoms. Pressure in a gas arises from collisions of gas molecules with the walls of a container. Temperature is related to the average kinetic energy of the atoms or molecules. Phase changes occur when thermal energy alters the balance between molecular motion and intermolecular forces.

Although atoms were once thought to be indivisible, modern physics shows that atoms themselves consist of smaller particles such as electrons, protons, and neutrons. However, for the study of thermal phenomena and the behavior of gases in this chapter, it is entirely adequate to treat atoms and molecules as fundamental particles.

### Conceptual Notes

The atomic theory connects microscopic particle behavior to macroscopic thermal phenomena. It forms the foundation for understanding temperature, thermal expansion, pressure, and the laws governing gases.


## 17.2. Temperature and Thermometers

Temperature is a physical quantity that characterizes the **thermal state** of a system and determines the direction of heat flow between systems. When two objects at different temperatures are placed in thermal contact, energy flows from the object at higher temperature to the one at lower temperature until equilibrium is reached.

Although temperature is closely related to thermal energy, it is not the same as energy. Temperature measures the **degree of hotness or coldness**, whereas thermal energy depends on both temperature and the amount of matter present.

---

### 1. Qualitative Meaning of Temperature

At the microscopic level, temperature is associated with the **random motion of atoms and molecules**. A higher temperature corresponds to greater average kinetic energy of the particles. However, temperature itself is defined operationally, based on measurable physical properties, rather than directly from microscopic motion.

Two systems have the same temperature if no net heat flows between them when they are placed in thermal contact.

---

### 2. Thermometric Properties

A **thermometer** measures temperature by exploiting a physical property that changes reproducibly with temperature. Such a property is called a **thermometric property**. Examples include:

- Length of a liquid column (liquid-in-glass thermometer),
- Volume or pressure of a gas,
- Electrical resistance of a metal (resistance thermometer),
- Voltage across a thermocouple.

For a thermometer to be useful, the thermometric property must vary **monotonically** with temperature.

---

### 3. Liquid-in-Glass Thermometers

In a liquid-in-glass thermometer, temperature changes cause the liquid (mercury or alcohol) to expand or contract. The height of the liquid column serves as the thermometric property.

Mercury is often used because:
- It expands nearly uniformly over a wide temperature range,
- It does not wet glass,
- It remains liquid over a large range of temperatures.

Alcohol is used for very low temperatures because it does not freeze as easily as mercury.

---

### 4. Temperature Scales

To assign numerical values to temperature, a **temperature scale** must be defined using fixed reference points.

#### Celsius Scale

On the Celsius scale:
- The freezing point of water is defined as $0^\circ\text{C}$,
- The boiling point of water (at standard atmospheric pressure) is defined as $100^\circ\text{C}$.

The scale is divided into 100 equal intervals.

#### Fahrenheit Scale

On the Fahrenheit scale:
- Water freezes at $32^\circ\text{F}$,
- Water boils at $212^\circ\text{F}$.

---

### 5. Conversion Between Celsius and Fahrenheit

The relationship between Celsius and Fahrenheit temperatures is linear:

$$
T_F = \frac{9}{5} T_C + 32,
$$

or equivalently,

$$
T_C = \frac{5}{9}(T_F - 32).
$$

---

### Example 17.1. Temperature Conversion

Convert $25^\circ\text{C}$ to Fahrenheit.

$$
T_F = \frac{9}{5}(25) + 32 = 45 + 32 = 77^\circ\text{F}.
$$

---

### 6. Gas Thermometers

Gas thermometers are particularly important because gas properties behave in a simple and predictable way at low pressures.

At constant volume, the **pressure** of a gas increases with temperature. At constant pressure, the **volume** increases with temperature. These relationships form the basis of gas temperature measurement and later lead to the definition of the absolute temperature scale.

---

### 7. Calibration of Thermometers

A thermometer must be calibrated by placing it in thermal contact with systems at known temperatures. Common calibration points include:
- The ice–water mixture (freezing point),
- Steam above boiling water at standard pressure.

Once calibrated, the thermometer can be used to measure unknown temperatures by comparing its thermometric property to these reference points.

---

### 8. Limitations of Ordinary Thermometers

Different thermometers may not agree exactly at all temperatures because different thermometric properties do not vary identically with temperature. Agreement improves when:
- The thermometer interacts weakly with the system,
- The system is in thermal equilibrium,
- The thermometer is properly calibrated.

The need for a universal temperature scale motivates the introduction of the **absolute temperature scale** in later sections.

---

### Conceptual Notes

Temperature determines the direction of heat flow between systems. It is defined operationally through measurable physical properties rather than directly from microscopic motion. Thermometers rely on properties that change reproducibly with temperature, such as length, pressure, volume, or resistance. Different temperature scales exist, but a more fundamental scale is required for precise thermodynamic analysis.

---

## 17.3. Thermal Equilibrium and the Zeroth Law of Thermodynamics

The concept of **thermal equilibrium** provides the fundamental basis for defining and measuring temperature.  
It describes the condition under which no net heat flows between systems that are in contact.  
This idea is formalized by the **Zeroth Law of Thermodynamics**, which establishes temperature as a meaningful and measurable physical quantity.

---

### 1. Thermal Equilibrium

Two systems are said to be in **thermal equilibrium** if, when they are placed in thermal contact, **no net heat flows** between them.  
This condition indicates that the two systems have the same temperature.

Thermal contact allows energy transfer in the form of heat but does not permit work to be done or matter to be exchanged.

---

### 2. The Need for a Fundamental Law

Before the Zeroth Law was explicitly stated, thermodynamics relied implicitly on the idea that temperature comparison is transitive.  
However, this property is not obvious and must be postulated as a fundamental principle.

Without such a law, the concept of temperature would lack logical consistency.

---

### 3. Statement of the Zeroth Law of Thermodynamics

**Zeroth Law of Thermodynamics:**  
If system A is in thermal equilibrium with system C, and system B is also in thermal equilibrium with system C, then systems A and B are in thermal equilibrium with each other.

This law implies that all three systems have the same temperature.

---

### 4. Physical Meaning of the Zeroth Law

The Zeroth Law establishes that **temperature is a property** shared by systems in thermal equilibrium.  
It allows the comparison of temperatures without requiring direct contact between every pair of systems.

System C in the Zeroth Law acts as a **thermometer**.  
If C is in equilibrium with both A and B, then A and B have the same temperature.

---

### 5. Thermometers and the Zeroth Law

A thermometer works by reaching thermal equilibrium with the system whose temperature is being measured.  
Once equilibrium is established, the thermometer’s thermometric property corresponds to the temperature of the system.

The Zeroth Law guarantees that the thermometer reading is meaningful and reproducible, independent of the thermometer’s composition, provided it is properly calibrated.

---

### Example 17.2. Using a Thermometer

A mercury thermometer is placed first in contact with system A and then with system B.  
If the thermometer shows the same reading in both cases, then by the Zeroth Law, systems A and B are in thermal equilibrium and have the same temperature.

---

### 6. Thermal Equilibrium vs. Mechanical Equilibrium

Thermal equilibrium concerns **heat flow** and temperature.  
Mechanical equilibrium concerns **forces** and pressure.  
A system can be in thermal equilibrium without being in mechanical equilibrium, and vice versa.

For example, two gas containers at the same temperature but different pressures are in thermal equilibrium but not mechanical equilibrium.

---

### 7. Importance in Thermodynamics

The Zeroth Law underpins all of thermodynamics by:

- Justifying the use of thermometers,
- Allowing the definition of temperature scales,
- Enabling meaningful comparison of temperatures across systems.

It is called the “zeroth” law because it was recognized after the First and Second Laws but is logically more fundamental.

---

### 8. Thermal Equilibrium and Heat Flow

Heat flows spontaneously from higher temperature to lower temperature systems until thermal equilibrium is reached.  
Once equilibrium is established, microscopic energy exchanges continue, but there is **no net energy transfer**.

This behavior motivates the later introduction of entropy and the Second Law of Thermodynamics.

---

### Conceptual Notes

Thermal equilibrium is the condition of no net heat flow between systems in contact. The Zeroth Law asserts the transitive nature of thermal equilibrium and establishes temperature as a well-defined physical quantity. This law provides the logical foundation for thermometry and all subsequent thermodynamic laws.

---

## 17.4. Thermal Expansion

When the temperature of a substance changes, its dimensions usually change as well.  
This phenomenon is known as **thermal expansion** and occurs because increasing temperature raises the average kinetic energy of atoms or molecules, causing them to vibrate more vigorously and, on average, to move farther apart.

Thermal expansion is important in engineering and everyday life, affecting bridges, buildings, pipelines, and precision instruments.

---

### 1. Microscopic Origin of Thermal Expansion

Atoms in a solid are bound to one another by intermolecular forces and vibrate about equilibrium positions.  
The potential energy curve describing these forces is generally **asymmetric**.  
As temperature increases, the average position of vibration shifts slightly, leading to an increase in average atomic separation and therefore an expansion of the material.

If the potential were perfectly symmetric, no expansion would occur.

---

### 2. Linear Thermal Expansion

For a solid object whose length $L$ changes with temperature, the change in length $\Delta L$ is given by

$$
\Delta L = \alpha L_0 \Delta T,
$$

where:
- $L_0$ is the original length,
- $\Delta T$ is the temperature change,
- $\alpha$ is the **coefficient of linear expansion**.

The coefficient $\alpha$ depends on the material and has units of $\text{K}^{-1}$.

---

### Example 17.3. Expansion of a Steel Rod

A steel rod of length $L_0 = 2.00~\text{m}$ is heated from $20^\circ\text{C}$ to $70^\circ\text{C}$.  
If $\alpha = 1.2\times10^{-5}~\text{K}^{-1}$, find the increase in length.

#### Solution

$$
\Delta T = 50~\text{K},
$$

$$
\Delta L = (1.2\times10^{-5})(2.00)(50) = 1.2\times10^{-3}~\text{m}.
$$

The rod increases in length by **1.2 mm**.

---

### 3. Area Expansion

For a two-dimensional object such as a metal plate, the change in area $\Delta A$ is approximately

$$
\Delta A = 2\alpha A_0 \Delta T,
$$

where $A_0$ is the original area.  
The factor of 2 arises because expansion occurs independently in two perpendicular directions.

---

### 4. Volume Expansion of Solids

For three-dimensional solids, the change in volume $\Delta V$ is

$$
\Delta V = \beta V_0 \Delta T,
$$

where:
- $V_0$ is the initial volume,
- $\beta$ is the **coefficient of volume expansion**.

For isotropic solids,

$$
\beta \approx 3\alpha.
$$

---

### 5. Volume Expansion of Liquids

Liquids also expand when heated, but their expansion is usually much greater than that of solids.  
Because liquids do not have a fixed shape, only **volume expansion** is defined for them.

The volume expansion of a liquid is described by the same form:

$$
\Delta V = \beta V_0 \Delta T,
$$

but the coefficient $\beta$ is generally larger than for solids.

---

### 6. Anomalous Expansion of Water

Water exhibits **anomalous behavior** between $0^\circ\text{C}$ and $4^\circ\text{C}$.  
As water is heated from $0^\circ\text{C}$ to $4^\circ\text{C}$, it **contracts** instead of expanding.  
Above $4^\circ\text{C}$, water expands normally.

This anomaly explains why ice floats on water and why lakes freeze from the top down, allowing aquatic life to survive in cold climates.

---

### 7. Practical Applications and Consequences

Thermal expansion must be accounted for in engineering design:

- Expansion joints in bridges and railways prevent buckling.
- Gaps in sidewalks allow for expansion in hot weather.
- Overhead power lines sag more in summer than in winter.
- Precision instruments use materials with very low expansion coefficients.

---

### Example 17.4. Expansion Gap in a Bridge

A steel bridge section is $30~\text{m}$ long at $0^\circ\text{C}$.  
If the temperature can reach $40^\circ\text{C}$ and $\alpha = 1.2\times10^{-5}~\text{K}^{-1}$, find the required expansion gap.

$$
\Delta L = (1.2\times10^{-5})(30)(40) = 0.0144~\text{m}.
$$

The expansion gap should be at least **1.4 cm**.

---

### 8. Limits of the Linear Approximation

The formulas for thermal expansion are accurate only for **small temperature changes**.  
For very large temperature variations, the expansion coefficients themselves may vary with temperature, and more detailed models are required.

---

### Conceptual Notes

Thermal expansion occurs because atomic vibrations increase with temperature, shifting average particle separation. Solids expand in length, area, and volume, while liquids expand primarily in volume. Water shows anomalous contraction between 0°C and 4°C, a property with major environmental significance. Thermal expansion plays a crucial role in engineering design and must be carefully managed.

---

## 17.5. Thermal Stresses

When a material undergoes a temperature change, it naturally tends to expand or contract.  
If this change in dimension is **constrained** by external forces or rigid supports, the material cannot freely change length or volume.  
As a result, **thermal stresses** develop within the material.  
These stresses can be large enough to cause deformation, cracking, or structural failure.

---

### 1. Origin of Thermal Stress

Thermal stress arises when thermal expansion is **prevented**.  
If a solid were free to expand, no stress would develop.  
However, when expansion is restricted, internal forces appear that oppose the attempted change in length.

Common situations include:
- Railroad rails fastened rigidly to the ground,
- Steel beams embedded in concrete,
- Glass constrained in a rigid frame.

---

### 2. Linear Thermal Strain

The **thermal strain** $\epsilon$ that would occur in a free object is

$$
\epsilon = \frac{\Delta L}{L_0} = \alpha \Delta T,
$$

where:
- $\alpha$ is the coefficient of linear expansion,
- $\Delta T$ is the temperature change.

If the object is constrained so that $\Delta L = 0$, this strain cannot occur and instead produces stress.

---

### 3. Stress–Strain Relation (Hooke’s Law)

For elastic deformations, stress and strain are related by **Young’s modulus** $Y$:

$$
\sigma = Y \epsilon,
$$

where:
- $\sigma$ is the stress (force per unit area),
- $Y$ is Young’s modulus of the material.

Substituting the thermal strain gives the **thermal stress**:

$$
\sigma = Y \alpha \Delta T.
$$

This formula applies when expansion is completely prevented.

---

### 4. Sign of Thermal Stress

- If temperature **increases** and expansion is prevented, the material experiences **compressive stress**.
- If temperature **decreases** and contraction is prevented, the material experiences **tensile stress**.

Tensile stresses are particularly dangerous for brittle materials such as glass or concrete.

---

### Example 17.5. Thermal Stress in a Steel Rod

A steel rod is rigidly fixed between two walls so it cannot expand.  
The temperature increases by $\Delta T = 40~\text{K}$.  
Given:
- $\alpha = 1.2\times10^{-5}~\text{K}^{-1}$,
- $Y = 2.0\times10^{11}~\text{Pa}$.

Find the thermal stress.

#### Solution

$$
\sigma = Y \alpha \Delta T
       = (2.0\times10^{11})(1.2\times10^{-5})(40)
       = 9.6\times10^{7}~\text{Pa}.
$$

The rod experiences a compressive stress of **96 MPa**.

---

### 5. Thermal Stress and Force

The force exerted by thermal stress on a cross-sectional area $A$ is

$$
F = \sigma A = Y \alpha \Delta T\, A.
$$

This force can be extremely large for thick structural elements.

---

### Example 17.6. Force Due to Thermal Stress

A steel beam with cross-sectional area $A = 5.0\times10^{-4}~\text{m}^2$ is rigidly constrained.  
Using the stress from the previous example:

$$
F = (9.6\times10^{7})(5.0\times10^{-4}) = 4.8\times10^{4}~\text{N}.
$$

The beam exerts a force of **48 kN** on its supports.

---

### 6. Engineering Implications

Thermal stresses must be carefully managed in design:

- Expansion joints allow free expansion in bridges and buildings.
- Gaps are left between concrete slabs and rails.
- Bimetallic strips exploit different expansion coefficients for temperature sensing.

Failure to account for thermal stress can lead to buckling, fracture, or long-term fatigue damage.

---

### 7. Bimetallic Strips

A **bimetallic strip** consists of two bonded metals with different expansion coefficients.  
When heated, one side expands more than the other, causing the strip to bend.

This principle is used in:
- Thermostats,
- Circuit breakers,
- Temperature-controlled switches.

---

### 8. Partial Constraints

If expansion is only partially constrained, the resulting stress is less than $Y\alpha\Delta T$.  
In such cases, mechanical deformation and thermal expansion share the strain.

A full analysis requires combining elasticity with thermal expansion effects.

---

### Conceptual Notes

Thermal stresses arise when thermal expansion or contraction is prevented. The resulting stress is proportional to Young’s modulus, the coefficient of expansion, and the temperature change. Large thermal stresses can develop even for modest temperature changes, making expansion allowances essential in engineering design. Tensile thermal stress is especially dangerous for brittle materials.

---

## 17.6. The Gas Laws and Absolute Temperature

Gases exhibit simple and well-defined relationships between pressure, volume, and temperature when the amount of gas is fixed.  
These relationships, known as the **gas laws**, were established experimentally and later unified into the ideal gas law.  
Their formulation also led to the concept of **absolute temperature**, which provides a fundamental and universal temperature scale.

---

### 1. Pressure, Volume, and Temperature of a Gas

A gas consists of a large number of rapidly moving molecules that collide with the walls of their container.  
The macroscopic properties of a gas arise from this microscopic motion:

- **Pressure** results from molecular collisions with container walls.  
- **Volume** is the space available for molecular motion.  
- **Temperature** reflects the average kinetic energy of the molecules.

By controlling one variable while holding others constant, several simple empirical laws are obtained.

---

### 2. Boyle’s Law (Pressure–Volume Law)

**Boyle’s Law:**  
For a fixed amount of gas at constant temperature, the pressure is inversely proportional to the volume:

$$
P \propto \frac{1}{V}.
$$

Equivalently,

$$
PV = \text{constant}.
$$

If a gas changes from state 1 to state 2 at constant temperature:

$$
P_1 V_1 = P_2 V_2.
$$

This law shows that compressing a gas increases its pressure.

---

### Example 17.7. Boyle’s Law

A gas occupies $2.0~\text{L}$ at a pressure of $1.0~\text{atm}$.  
If it is compressed to $0.50~\text{L}$ at constant temperature, the new pressure is

$$
P_2 = P_1 \frac{V_1}{V_2} = (1.0)\frac{2.0}{0.50} = 4.0~\text{atm}.
$$

---

### 3. Charles’s Law (Volume–Temperature Law)

**Charles’s Law:**  
For a fixed amount of gas at constant pressure, the volume is directly proportional to the temperature:

$$
V \propto T.
$$

Experiments show that extrapolating the $V$–$T$ relationship leads to zero volume at approximately $-273^\circ\text{C}$.

This observation motivates the definition of an absolute temperature scale.

---

### 4. Gay-Lussac’s Law (Pressure–Temperature Law)

**Gay-Lussac’s Law:**  
For a fixed amount of gas at constant volume, the pressure is directly proportional to the temperature:

$$
P \propto T.
$$

Thus,

$$
\frac{P}{T} = \text{constant}.
$$

---

### 5. Absolute Temperature and the Kelvin Scale

The **absolute temperature scale** is defined so that zero temperature corresponds to zero average molecular kinetic energy.  
This temperature is called **absolute zero**.

The Kelvin temperature $T$ is related to the Celsius temperature $T_C$ by

$$
T = T_C + 273.15.
$$

On this scale:
- $0~\text{K}$ corresponds to $-273.15^\circ\text{C}$,
- Temperature differences are the same in kelvin and degrees Celsius.

---

### 6. Combined Gas Law

Combining Boyle’s, Charles’s, and Gay-Lussac’s laws yields the **combined gas law**:

$$
\frac{P V}{T} = \text{constant}.
$$

For two equilibrium states of the same gas:

$$
\frac{P_1 V_1}{T_1} = \frac{P_2 V_2}{T_2}.
$$

This relation is valid provided the amount of gas remains constant.

---

### Example 17.8. Combined Gas Law

A gas has $P_1 = 1.0~\text{atm}$, $V_1 = 3.0~\text{L}$, and $T_1 = 300~\text{K}$.  
If it changes to $V_2 = 2.0~\text{L}$ and $T_2 = 400~\text{K}$, find $P_2$.

$$
P_2 = P_1 \frac{V_1}{V_2} \frac{T_2}{T_1}
    = (1.0)\frac{3.0}{2.0}\frac{400}{300}
    = 2.0~\text{atm}.
$$

---

### 7. Physical Meaning of Absolute Zero

At absolute zero, molecular motion would cease entirely in the classical picture.  
In reality, quantum effects prevent complete rest, but absolute zero represents the lowest attainable temperature.

All gas laws require temperature to be measured on the **absolute (Kelvin) scale** to remain valid.

---

### Conceptual Notes

The gas laws describe how pressure, volume, and temperature are related for a fixed amount of gas. Boyle’s law applies at constant temperature, Charles’s law at constant pressure, and Gay-Lussac’s law at constant volume. Their combination leads naturally to the concept of absolute temperature, measured in kelvin, which provides a universal and physically meaningful temperature scale.

---

In the next section, these empirical laws are unified into a single equation: the **ideal gas law**.


## 17.7. The Ideal Gas Law

The experimental gas laws relating pressure, volume, and temperature can be combined into a single equation that accurately describes the behavior of many gases under ordinary conditions.  
This relation is known as the **ideal gas law**.  
It provides a simple yet powerful model connecting macroscopic gas properties to microscopic motion.

---

### 1. Formulation of the Ideal Gas Law

For a fixed amount of gas, experiments show that

$$
PV \propto T.
$$

Introducing a proportionality constant that depends on the amount of gas leads to

$$
PV = nRT,
$$

where:
- $P$ is the pressure,
- $V$ is the volume,
- $T$ is the absolute temperature (in kelvin),
- $n$ is the number of moles of gas,
- $R$ is the **universal gas constant**.

---

### 2. The Universal Gas Constant

The gas constant has the value

$$
R = 8.31~\text{J/(mol·K)}.
$$

This constant is the same for all ideal gases and reflects the universal nature of molecular thermal motion.

---

### 3. Conditions for Ideal Gas Behavior

A real gas behaves approximately as an ideal gas when:
- The gas density is low,
- The pressure is not too high,
- The temperature is well above the condensation point.

Under these conditions, the volume of individual molecules and intermolecular forces can be neglected.

---

### 4. Units in the Ideal Gas Law

To use the ideal gas law correctly, consistent units must be used:

- Pressure in pascals (Pa),
- Volume in cubic meters (m³),
- Temperature in kelvin (K),
- Amount of substance in moles (mol).

One atmosphere is related to SI units by

$$
1~\text{atm} = 1.01\times10^5~\text{Pa}.
$$

---

### Example 17.9. Gas Volume from the Ideal Gas Law

A sample of $1.00~\text{mol}$ of gas is at a temperature of $300~\text{K}$ and a pressure of $1.01\times10^5~\text{Pa}$.  
Find the volume.

#### Solution

$$
V = \frac{nRT}{P}
  = \frac{(1.00)(8.31)(300)}{1.01\times10^5}
  = 2.47\times10^{-2}~\text{m}^3.
$$

This corresponds to **24.7 L**, close to the molar volume of an ideal gas at room temperature.

---

### 5. Molar Volume of an Ideal Gas

At standard temperature and pressure (STP):
- $T = 273~\text{K}$,
- $P = 1.01\times10^5~\text{Pa}$.

For $n = 1.00~\text{mol}$:

$$
V = \frac{RT}{P} = \frac{(8.31)(273)}{1.01\times10^5}
  = 2.24\times10^{-2}~\text{m}^3.
$$

Thus, the molar volume at STP is **22.4 L**.

---

### 6. Solving the Ideal Gas Law

The ideal gas law can be rearranged to solve for any variable:

$$
P = \frac{nRT}{V}, \quad
V = \frac{nRT}{P}, \quad
T = \frac{PV}{nR}, \quad
n = \frac{PV}{RT}.
$$

These forms are used extensively in thermodynamics and chemistry.

---

### Example 17.10. Gas Pressure Calculation

A gas occupies $0.010~\text{m}^3$ at $T = 350~\text{K}$ and contains $0.50~\text{mol}$.  
Find the pressure.

$$
P = \frac{nRT}{V}
  = \frac{(0.50)(8.31)(350)}{0.010}
  = 1.45\times10^5~\text{Pa}.
$$

---

### 7. Microscopic Interpretation (Preview)

The ideal gas law reflects the microscopic behavior of gas molecules.  
Later, it will be shown that pressure arises from molecular collisions and that temperature is proportional to the average kinetic energy per molecule.

This microscopic viewpoint leads to the kinetic theory of gases.

---

### 8. Limitations of the Ideal Gas Law

The ideal gas law fails when:
- Pressure is very high,
- Temperature is very low,
- Gas molecules are large or strongly interacting.

Under these conditions, corrections for molecular volume and intermolecular forces are required.

---

### Conceptual Notes

The ideal gas law unifies all empirical gas laws into a single equation. It requires the use of absolute temperature and introduces the mole as a measure of the amount of substance. Although it is an approximation, it accurately describes the behavior of many gases over a wide range of conditions and forms the basis for the kinetic theory of gases.

---

## 17.8. Problem Solving with the Ideal Gas Law

The ideal gas law provides a powerful and flexible tool for solving a wide variety of problems involving gases.  
Successful problem solving requires careful attention to **units**, **absolute temperature**, and the **amount of gas** involved.  
In this section, we outline a systematic approach and work through representative examples.

---

### 1. General Strategy

When applying the ideal gas law,

$$
PV = nRT,
$$

the following steps should always be followed:

1. Identify the known and unknown quantities.  
2. Convert all temperatures to **kelvin**.  
3. Convert pressures and volumes to **SI units** where required.  
4. Solve algebraically for the desired variable before inserting numbers.  
5. Check the result for physical plausibility.

---

### 2. Importance of Absolute Temperature

The ideal gas law is valid only when temperature is measured on the **absolute (Kelvin) scale**.  
Using Celsius temperatures directly will lead to incorrect results.

The conversion is:

$$
T = T_C + 273.15.
$$

---

### Example 17.11. Temperature Conversion Error

A gas at $27^\circ\text{C}$ is mistakenly treated as $T = 27$ in the ideal gas law instead of $300~\text{K}$.  
This results in an error of more than a factor of ten, illustrating the necessity of using kelvin.

---

### 3. Solving for an Unknown Quantity

The ideal gas law can be rearranged depending on which variable is unknown:

$$
P = \frac{nRT}{V}, \quad
V = \frac{nRT}{P}, \quad
T = \frac{PV}{nR}, \quad
n = \frac{PV}{RT}.
$$

Each form corresponds to a different physical situation.

---

### Example 17.12. Finding the Amount of Gas

A gas occupies $0.030~\text{m}^3$ at a pressure of $2.0\times10^5~\text{Pa}$ and a temperature of $400~\text{K}$.  
Find the number of moles.

#### Solution

$$
n = \frac{PV}{RT}
  = \frac{(2.0\times10^5)(0.030)}{(8.31)(400)}
  = 1.80~\text{mol}.
$$

---

### 4. Gases Under Changing Conditions

If a gas changes state while the amount of gas remains constant, the combined gas law applies:

$$
\frac{P_1 V_1}{T_1} = \frac{P_2 V_2}{T_2}.
$$

This is particularly useful when comparing initial and final states.

---

### Example 17.13. Heating a Confined Gas

A gas initially has $P_1 = 1.0\times10^5~\text{Pa}$, $V_1 = 1.5\times10^{-2}~\text{m}^3$, and $T_1 = 300~\text{K}$.  
It is heated at constant volume to $T_2 = 450~\text{K}$.  
Find the final pressure.

#### Solution

$$
P_2 = P_1 \frac{T_2}{T_1}
    = (1.0\times10^5)\frac{450}{300}
    = 1.5\times10^5~\text{Pa}.
$$

---

### 5. Density of an Ideal Gas

Gas density $\rho$ is mass per unit volume:

$$
\rho = \frac{m}{V}.
$$

Using $m = nM$, where $M$ is the molar mass:

$$
\rho = \frac{PM}{RT}.
$$

This expression shows that gas density increases with pressure and decreases with temperature.

---

### Example 17.14. Density of Air

Find the density of air at $P = 1.01\times10^5~\text{Pa}$ and $T = 300~\text{K}$.  
Assume molar mass $M = 0.029~\text{kg/mol}$.

$$
\rho = \frac{(1.01\times10^5)(0.029)}{(8.31)(300)} = 1.17~\text{kg/m}^3.
$$

---

### 6. Mixtures of Ideal Gases

For a mixture of ideal gases, each component obeys the ideal gas law independently.  
The total pressure is the sum of the **partial pressures**:

$$
P = P_1 + P_2 + P_3 + \dots
$$

This principle is known as **Dalton’s law of partial pressures**.

---

### 7. Partial Pressure and Mole Fraction

The partial pressure of component $i$ is:

$$
P_i = x_i P,
$$

where $x_i = n_i / n_{\text{total}}$ is the mole fraction.

---

### Example 17.15. Partial Pressure in Air

If oxygen makes up 21% of dry air at $1.0\times10^5~\text{Pa}$:

$$
P_{\text{O}_2} = 0.21(1.0\times10^5) = 2.1\times10^4~\text{Pa}.
$$

---

### 8. Common Pitfalls

- Forgetting to convert to kelvin,  
- Mixing SI and non-SI units,  
- Using the ideal gas law outside its valid range,  
- Confusing mass with amount of substance.

Careful dimensional analysis helps prevent these errors.

---

### Conceptual Notes

Problem solving with the ideal gas law requires consistent units, absolute temperature, and careful algebraic manipulation. The law applies to individual gases and mixtures alike and can be used to determine pressure, volume, temperature, or amount of gas. Density and partial pressure relationships follow naturally from the ideal gas model.

---

## 17.9. Ideal Gas Law in Terms of Molecules; Avogadro’s Number

The ideal gas law can be expressed not only in terms of moles but also directly in terms of **individual molecules**.  
This reformulation introduces **Avogadro’s number**, which provides the crucial link between macroscopic thermodynamic quantities and microscopic particle behavior.

---

### 1. Avogadro’s Number

**Avogadro’s number** $N_A$ is defined as the number of particles in one mole of a substance:

$$
N_A = 6.02\times10^{23}~\text{mol}^{-1}.
$$

Thus:
- 1 mol of atoms contains $6.02\times10^{23}$ atoms,  
- 1 mol of molecules contains $6.02\times10^{23}$ molecules.

This constant bridges the atomic scale and the macroscopic scale.

---

### 2. Number of Molecules and Amount of Substance

If a gas contains $n$ moles, the total number of molecules $N$ is:

$$
N = n N_A.
$$

Conversely,

$$
n = \frac{N}{N_A}.
$$

This relation allows the ideal gas law to be written in molecular form.

---

### 3. Ideal Gas Law in Molecular Form

Starting from

$$
PV = nRT,
$$

and substituting $n = N / N_A$, we obtain

$$
PV = \frac{N}{N_A} RT.
$$

Defining the **Boltzmann constant** $k_B$ as

$$
k_B = \frac{R}{N_A} = 1.38\times10^{-23}~\text{J/K},
$$

the ideal gas law becomes

$$
PV = N k_B T.
$$

This equation relates macroscopic variables directly to the number of molecules.

---

### 4. Physical Meaning of the Boltzmann Constant

The Boltzmann constant sets the energy scale for thermal motion at the molecular level.  
It connects temperature to energy:

$$
\text{Typical thermal energy per molecule} \sim k_B T.
$$

At room temperature ($T \approx 300~\text{K}$):

$$
k_B T \approx 4.1\times10^{-21}~\text{J}.
$$

---

### 5. Example 17.16. Number of Molecules in a Gas Sample

A gas occupies a volume of $0.020~\text{m}^3$ at a pressure of $1.0\times10^5~\text{Pa}$ and a temperature of $300~\text{K}$.  
Find the number of molecules.

#### Solution

Using the molecular form of the ideal gas law:

$$
N = \frac{PV}{k_B T}
  = \frac{(1.0\times10^5)(0.020)}{(1.38\times10^{-23})(300)}
  = 4.8\times10^{23}.
$$

The gas contains approximately **$5\times10^{23}$ molecules**.

---

### 6. Molecular Interpretation of Pressure

From the molecular viewpoint, pressure arises from countless collisions of gas molecules with the walls of the container.  
Each collision transfers momentum to the wall, and the collective effect of many such collisions produces a steady force per unit area.

The ideal gas law summarizes the average result of these microscopic interactions.

---

### 7. Average Molecular Spacing

The average volume per molecule is

$$
\frac{V}{N}.
$$

The characteristic distance between molecules is approximately

$$
d \sim \left(\frac{V}{N}\right)^{1/3}.
$$

For air at atmospheric pressure and room temperature, this distance is on the order of a few nanometers, much larger than the size of individual molecules, justifying the ideal gas approximation.

---

### 8. Molecules, Mass, and Density

If each molecule has mass $m_0$, the total mass of the gas is

$$
m = N m_0.
$$

The density is therefore

$$
\rho = \frac{m}{V} = \frac{N m_0}{V}.
$$

Using $PV = N k_B T$, this can be written as

$$
\rho = \frac{P m_0}{k_B T}.
$$

This expression links density directly to molecular mass and temperature.

---

### 9. Example 17.17. Molecular Mass from Gas Density

A gas has density $\rho = 1.2~\text{kg/m}^3$ at $P = 1.01\times10^5~\text{Pa}$ and $T = 300~\text{K}$.  
Find the mass of one molecule.

$$
m_0 = \frac{\rho k_B T}{P}
    = \frac{(1.2)(1.38\times10^{-23})(300)}{1.01\times10^5}
    = 4.9\times10^{-26}~\text{kg}.
$$

This value is consistent with the mass of an air molecule.

---

### 10. Significance of the Molecular Form

The molecular form of the ideal gas law:

$$
PV = N k_B T
$$

- Applies to any number of molecules,  
- Makes no reference to moles,  
- Is fundamental in statistical mechanics and kinetic theory.

It emphasizes that temperature is a measure of molecular energy and that macroscopic gas behavior arises from microscopic motion.

---

### Conceptual Notes

Avogadro’s number connects the mole to individual particles. Rewriting the ideal gas law in molecular form introduces the Boltzmann constant, which links temperature to energy at the microscopic level. This formulation provides a direct bridge between thermodynamics and molecular physics and sets the stage for the kinetic theory of gases.

---

## 17.10. Ideal Gas Temperature Scale — a Standard

The behavior of an ideal gas provides a natural and precise way to define a **temperature scale** that is independent of the properties of any particular substance.  
This leads to the **ideal gas temperature scale**, which forms the basis of the modern **Kelvin scale** and serves as the standard for thermodynamic temperature.

---

### 1. Motivation for a Standard Temperature Scale

Different thermometers rely on different thermometric properties, such as length, pressure, or electrical resistance.  
While these devices agree closely over moderate temperature ranges, small discrepancies arise because no real material behaves perfectly linearly with temperature.

A universal temperature scale must therefore be defined using a system whose behavior is as simple and reproducible as possible.  
An ideal gas, which obeys simple and universal laws, provides such a reference.

---

### 2. Gas Thermometer at Constant Volume

Consider a fixed amount of gas confined to a constant volume.  
Experiments show that, as the pressure approaches zero, the pressure of any gas becomes directly proportional to temperature:

$$
P \propto T \quad (\text{as } P \to 0).
$$

This behavior is nearly identical for all gases at sufficiently low pressures.

---

### 3. Definition of the Ideal Gas Temperature

Using a constant-volume gas thermometer, temperature is defined by

$$
T = \lim_{P \to 0} \left( \frac{P}{P_0} \right) T_0,
$$

where:
- $P$ is the pressure of the gas at the temperature being measured,
- $P_0$ is the pressure at a reference temperature $T_0$,
- the limit ensures that deviations from ideal behavior are eliminated.

This definition does not depend on the type of gas used.

---

### 4. Absolute Zero

Extrapolating the pressure–temperature relationship for a gas at constant volume shows that pressure approaches zero at a finite temperature.  
This temperature is identified as **absolute zero**.

Experimentally, absolute zero corresponds to approximately

$$
T = -273.15^\circ\text{C}.
$$

At this temperature, the average kinetic energy of gas molecules approaches its minimum possible value.

---

### 5. The Kelvin Temperature Scale

The **Kelvin scale** is defined so that:
- Absolute zero corresponds to $0~\text{K}$,
- Temperature intervals are the same size as those on the Celsius scale.

Thus, the conversion between Celsius and Kelvin is

$$
T = T_C + 273.15.
$$

The kelvin is the SI unit of thermodynamic temperature.

---

### 6. Triple Point of Water

To establish a precise and reproducible standard, a fixed reference point is required.  
The **triple point of water** is the unique temperature and pressure at which ice, liquid water, and water vapor coexist in equilibrium.

By definition:

$$
T_{\text{triple}} = 273.16~\text{K}.
$$

This single point defines the Kelvin scale without reference to any other temperatures.

---

### 7. Ideal Gas Law and Temperature Scale

The ideal gas law,

$$
PV = nRT,
$$

is valid only when temperature is measured on the **ideal gas (Kelvin) scale**.  
Using Celsius temperatures directly would violate the proportionality between pressure and temperature required by the gas laws.

The ideal gas temperature scale therefore ensures internal consistency across all thermodynamic relations.

---

### 8. Universality of the Ideal Gas Scale

At sufficiently low pressures, all real gases approach ideal behavior.  
As a result, the ideal gas temperature scale is:

- Independent of gas species,  
- Independent of the amount of gas,  
- Reproducible in any laboratory.

This universality makes it the foundation of all precise temperature measurements.

---

### 9. Relation to Microscopic Physics

On the ideal gas temperature scale, temperature is directly related to molecular motion.  
In later developments of kinetic theory, temperature is shown to be proportional to the **average translational kinetic energy** of gas molecules:

$$
\langle K \rangle \propto k_B T.
$$

This confirms the deep physical meaning of thermodynamic temperature.

---

### 10. Practical Temperature Measurement

Modern thermometers, such as platinum resistance thermometers and thermocouples, are calibrated against the Kelvin scale.  
Although they do not use gases directly, their readings are traceable to the ideal gas temperature definition.

---

### Conceptual Notes

The ideal gas temperature scale provides a universal and substance-independent definition of temperature. By extrapolating gas behavior to zero pressure, deviations from real-gas effects are eliminated. Absolute zero emerges naturally from this construction, and the Kelvin scale formalizes thermodynamic temperature as a fundamental physical quantity tied directly to molecular energy.

---

## 18.1. The Ideal Gas Law and the Molecular Interpretation of Temperature

Kinetic theory analyzes the behavior of gases by modeling them as large collections of molecules in continuous random motion. Instead of tracking individual molecules, which is impossible in practice, the theory connects **statistical averages of microscopic quantities** to **macroscopic observables** such as pressure, volume, and temperature.

### Basic Assumptions of the Kinetic Theory of an Ideal Gas

The kinetic theory of an ideal gas is based on the following assumptions.

There is a very large number $N$ of identical molecules, each of mass $m$, moving randomly in all directions.  
The molecules are, on average, far apart compared to their own size, so the volume occupied by the molecules themselves is negligible compared to the volume of the container.  
The molecules obey the laws of classical mechanics and interact with one another only during brief collisions.  
All collisions between molecules, and between molecules and the container walls, are perfectly elastic.  
Any intermolecular forces between collisions are negligible compared to the kinetic energy of the molecules.

Under these conditions, real gases at low pressure and sufficiently high temperature behave very closely like ideal gases.

---

### Microscopic Origin of Gas Pressure

Gas pressure arises from collisions of molecules with the walls of the container. Consider a rectangular container of length $l$ and wall area $A$. A molecule with velocity component $v_x$ collides elastically with a wall perpendicular to the $x$-direction.

During a collision, the $x$-component of momentum changes from $-mv_x$ to $+mv_x$, so the change in momentum is
$$
\Delta p_x = 2mv_x.
$$

The time between successive collisions of this molecule with the same wall is the time required to travel a distance $2l$:
$$
\Delta t = \frac{2l}{v_x}.
$$

The average force exerted by this single molecule on the wall is
$$
F = \frac{\Delta p_x}{\Delta t}
= \frac{2mv_x}{2l/v_x}
= \frac{mv_x^2}{l}.
$$

To find the total force on the wall, we sum over all $N$ molecules:
$$
F = \frac{m}{l} \left( v_{x1}^2 + v_{x2}^2 + \cdots + v_{xN}^2 \right).
$$

Defining the average of $v_x^2$ as
$$
\overline{v_x^2} = \frac{1}{N}\sum_{i=1}^N v_{xi}^2,
$$
the total force becomes
$$
F = \frac{Nm\overline{v_x^2}}{l}.
$$

---

### Isotropy of Molecular Motion

Because molecular motion is random, there is no preferred direction. The average squared velocity components satisfy
$$
\overline{v_x^2} = \overline{v_y^2} = \overline{v_z^2}.
$$

Since
$$
v^2 = v_x^2 + v_y^2 + v_z^2,
$$
taking averages gives
$$
\overline{v^2} = 3\overline{v_x^2},
$$
or
$$
\overline{v_x^2} = \frac{1}{3}\overline{v^2}.
$$

Substituting into the force expression yields
$$
F = \frac{1}{3}\frac{Nm\overline{v^2}}{l}.
$$

---

### Pressure in Terms of Molecular Motion

Pressure is defined as force per unit area:
$$
P = \frac{F}{A}.
$$

Using $V = lA$ for the volume of the container, we obtain
$$
P = \frac{1}{3}\frac{Nm\overline{v^2}}{V}.
$$

Multiplying both sides by $V$ gives
$$
PV = \frac{2}{3}N\left( \frac{1}{2}m\overline{v^2} \right).
$$

The quantity
$$
\frac{1}{2}m\overline{v^2}
$$
is the **average translational kinetic energy** of a molecule.

---

### Molecular Interpretation of Temperature

The macroscopic ideal gas law is
$$
PV = NkT,
$$
where $k = 1.38 \times 10^{-23}\,\text{J/K}$ is Boltzmann’s constant.

Comparing this with the kinetic theory result leads to
$$
\frac{1}{2}m\overline{v^2} = \frac{3}{2}kT.
$$

This fundamental relation shows that:

The absolute temperature of a gas is directly proportional to the average translational kinetic energy of its molecules.

Temperature is therefore a **measure of molecular motion**, not a measure of total energy or speed of any single molecule.

---

### Root-Mean-Square Speed

The square root of the average of $v^2$ is called the **root-mean-square speed**:
$$
v_{\text{rms}} = \sqrt{\overline{v^2}}.
$$

From the kinetic energy relation,
$$
v_{\text{rms}} = \sqrt{\frac{3kT}{m}}.
$$

Heavier molecules move more slowly on average than lighter ones at the same temperature, even though their average kinetic energies are equal.

---

### Numerical Example: Molecular Kinetic Energy

At body temperature $T = 310\,\text{K}$,
$$
\overline{K} = \frac{3}{2}kT
= \frac{3}{2}(1.38 \times 10^{-23})(310)
= 6.4 \times 10^{-21}\,\text{J}.
$$

Although this energy is extremely small for a single molecule, the total kinetic energy of one mole of gas is macroscopic.

---

### Conceptual Notes

All gases at the same temperature have the same average translational kinetic energy per molecule, independent of molecular mass.  
The average **velocity** of molecules in a gas is zero, because motion is equally likely in all directions, but the average **speed** is not zero.  
As temperature approaches absolute zero, classical kinetic theory predicts vanishing kinetic energy, although quantum effects prevent motion from fully ceasing.

This microscopic interpretation of temperature forms the foundation for the Maxwell speed distribution and the behavior of real gases discussed in later sections.


## 18.2. Distribution of Molecular Speeds

Although all molecules in an ideal gas at a given temperature have the same **average kinetic energy**, they do not all move with the same speed.  
Instead, molecular speeds are distributed over a wide range of values.  
This distribution is described by the **Maxwell–Boltzmann speed distribution**, which arises naturally from statistical considerations applied to large numbers of particles.

---

### 1. Need for a Speed Distribution

At any instant, gas molecules undergo frequent elastic collisions that continuously change their speeds and directions.  
Some molecules move very slowly, others extremely fast, and most move at intermediate speeds.  
Macroscopic gas laws describe average behavior, but a complete microscopic description requires knowing how speeds are distributed.

---

### 2. Maxwell–Boltzmann Speed Distribution

For an ideal gas of molecules with mass $m$ at temperature $T$, the probability density $f(v)$ that a molecule has speed between $v$ and $v + dv$ is

$$
f(v) = 4\pi \left( \frac{m}{2\pi kT} \right)^{3/2} v^2
       e^{-mv^2/(2kT)}.
$$

Key features of this distribution:

- $f(v) = 0$ at $v = 0$,  
- $f(v)$ rises to a maximum at an intermediate speed,  
- $f(v)$ decreases exponentially for large $v$.

The factor $v^2$ accounts for the number of velocity states with speed $v$, while the exponential factor suppresses very high speeds.

---

### 3. Normalization of the Distribution

The distribution is normalized so that the total probability is 1:

$$
\int_0^\infty f(v)\,dv = 1.
$$

This ensures that every molecule has some speed between 0 and infinity.

---

### 4. Characteristic Speeds

Several characteristic speeds are commonly defined.

#### Most Probable Speed

The **most probable speed** $v_p$ is the speed at which $f(v)$ is maximum:

$$
v_p = \sqrt{\frac{2kT}{m}}.
$$

#### Average Speed

The **mean (average) speed** is

$$
\overline{v} = \sqrt{\frac{8kT}{\pi m}}.
$$

#### Root-Mean-Square Speed

The **root-mean-square speed** is

$$
v_{\text{rms}} = \sqrt{\overline{v^2}} = \sqrt{\frac{3kT}{m}}.
$$

These speeds satisfy

$$
v_p < \overline{v} < v_{\text{rms}}.
$$

---

### 5. Physical Interpretation

- $v_p$ corresponds to the peak of the speed distribution curve.  
- $\overline{v}$ gives the average magnitude of molecular velocity.  
- $v_{\text{rms}}$ is directly related to the average kinetic energy:
  $$
  \frac{1}{2}m v_{\text{rms}}^2 = \frac{3}{2}kT.
  $$

---

### Example 18.1. Characteristic Speeds of Nitrogen

Nitrogen molecules ($\text{N}_2$) have molecular mass  
$m = 4.65\times10^{-26}~\text{kg}$.

At $T = 300~\text{K}$:

$$
v_p = \sqrt{\frac{2kT}{m}} = 422~\text{m/s},
$$

$$
\overline{v} = \sqrt{\frac{8kT}{\pi m}} = 476~\text{m/s},
$$

$$
v_{\text{rms}} = \sqrt{\frac{3kT}{m}} = 517~\text{m/s}.
$$

---

### 6. Effect of Temperature

Increasing the temperature:

- Shifts the distribution to **higher speeds**,  
- Broadens the distribution,  
- Increases all characteristic speeds proportional to $\sqrt{T}$.

Doubling the temperature increases characteristic speeds by a factor of $\sqrt{2}$.

---

### 7. Effect of Molecular Mass

For gases at the same temperature:

- Heavier molecules move more slowly on average,  
- Lighter molecules move faster.

All characteristic speeds scale as

$$
v \propto \frac{1}{\sqrt{m}}.
$$

---

### Example 18.2. Oxygen vs. Hydrogen

At the same temperature, hydrogen molecules ($\text{H}_2$) move about four times faster than oxygen molecules ($\text{O}_2$), because the molecular mass of $\text{O}_2$ is about 16 times larger.

---

### 8. Speed Distribution vs. Velocity Distribution

The speed distribution describes the **magnitude** of velocity only.  
The full velocity distribution depends on three components $(v_x, v_y, v_z)$ and is isotropic.

Although the average velocity vector is zero, the average of $v^2$ is nonzero and determines temperature.

---

### 9. High-Speed Tail and Escape Phenomena

The exponential tail of the distribution means that a small fraction of molecules always have speeds far above average.  
This explains phenomena such as:

- Atmospheric escape of light gases,  
- Enhanced reaction rates at high temperatures,  
- Diffusion and effusion processes.

---

### Conceptual Notes

Molecular speeds in a gas are distributed according to the Maxwell–Boltzmann distribution, not concentrated at a single value. Temperature determines the overall scale of molecular speeds, while molecular mass controls how fast molecules move at a given temperature. The existence of a high-speed tail in the distribution has important physical consequences, even though only a small fraction of molecules occupy it.

---

## 18.3. Real Gases and Changes of Phase

The ideal gas model provides an excellent approximation for many gases under ordinary conditions, but real gases deviate from ideal behavior at **high pressures** and **low temperatures**.  
These deviations arise because real gas molecules have **finite size** and **intermolecular forces**, neither of which is included in the ideal gas assumptions.  
Understanding these effects also explains how gases undergo **phase changes** into liquids and solids.

---

### 1. Limitations of the Ideal Gas Model

The ideal gas law assumes that:
- Gas molecules occupy negligible volume,  
- There are no intermolecular forces except during collisions,  
- Collisions are perfectly elastic.

In reality:
- Molecules have finite size, reducing the free volume available for motion,  
- Attractive forces act between molecules, especially at short distances.

These effects become significant when molecules are close together, i.e., at high pressure or low temperature.

---

### 2. Deviations at High Pressure

At high pressures, the volume occupied by the molecules themselves is no longer negligible compared to the container volume.  
As a result, the **effective volume** available for molecular motion is smaller than the measured volume.

This causes the pressure predicted by the ideal gas law to be **too low**, since collisions with the container walls occur more frequently than the ideal model predicts.

---

### 3. Deviations at Low Temperature

At low temperatures, molecular speeds decrease, and **intermolecular attractions** become important.  
These attractive forces pull molecules slightly inward, reducing the momentum transferred to the container walls during collisions.

As a consequence, the measured pressure is **lower** than that predicted by the ideal gas law.

---

### 4. Competing Effects

The two deviation mechanisms act in opposite directions:

- Finite molecular volume tends to **increase** pressure relative to the ideal gas prediction.  
- Intermolecular attraction tends to **decrease** pressure.

Depending on temperature and pressure, one effect may dominate.

---

### 5. Compressibility Factor

The degree to which a real gas deviates from ideal behavior is described by the **compressibility factor** $Z$:

$$
Z = \frac{PV}{nRT}.
$$

- For an ideal gas, $Z = 1$.  
- For real gases, $Z \ne 1$.

At moderate pressures, many gases have $Z < 1$ due to attractive forces.  
At very high pressures, $Z > 1$ due to finite molecular volume.

---

### 6. Phase Changes: Gas to Liquid

As temperature decreases or pressure increases, a gas may undergo a **phase change** into a liquid.  
During condensation:
- Molecules lose kinetic energy,  
- Intermolecular attractions bind molecules together,  
- The gas collapses into a much denser liquid phase.

This process cannot be described by the ideal gas law.

---

### 7. Critical Temperature and Critical Pressure

Each substance has a **critical temperature** $T_c$ and **critical pressure** $P_c$.

- Above $T_c$, no amount of pressure can liquefy the gas.  
- Below $T_c$, increasing pressure can cause condensation.

At the **critical point**, the distinction between gas and liquid disappears.

---

### Example 18.3. Carbon Dioxide

For carbon dioxide:
- $T_c = 304~\text{K}$,  
- $P_c = 7.4~\text{MPa}$.

Above $304~\text{K}$, CO$_2$ cannot exist as a liquid regardless of pressure.

---

### 8. Phase Diagrams

A **phase diagram** shows the regions of pressure and temperature for which a substance exists as a solid, liquid, or gas.

Key features include:
- Phase boundaries (solid–liquid, liquid–gas, solid–gas),  
- The **triple point**, where all three phases coexist,  
- The **critical point**, where liquid and gas become indistinguishable.

---

### 9. Latent Heat and Phase Transitions

During a phase change, temperature remains constant while energy is absorbed or released.  
This energy is called **latent heat**.

For vaporization:
$$
Q = m L_v,
$$

where $L_v$ is the latent heat of vaporization.

This energy is used to overcome intermolecular attractions, not to increase kinetic energy.

---

### 10. Microscopic Picture of Condensation

As a gas cools:
- Average molecular speed decreases,  
- A growing fraction of molecules have speeds too low to escape attractive forces,  
- Clusters form and grow into a liquid.

The Maxwell–Boltzmann distribution explains why condensation begins gradually rather than abruptly.

---

### Conceptual Notes

Real gases deviate from ideal behavior because molecules have finite size and experience intermolecular forces. These deviations become important near phase transitions. The concepts of critical temperature, critical pressure, and phase diagrams describe the conditions under which gases liquefy or solidify. Phase changes involve latent heat and reflect changes in molecular arrangement rather than temperature.

---

In the next section, we examine **vapor pressure and humidity**, which describe the behavior of gases in equilibrium with their liquid phase.


## 18.4. Vapor Pressure and Humidity

When a liquid is in contact with its vapor in a closed container, molecules continuously leave the liquid phase and return to it.  
This dynamic balance leads to the concept of **vapor pressure**, which plays a central role in evaporation, boiling, weather, and humidity in the atmosphere.

---

### 1. Evaporation and Condensation

At any temperature, molecules within a liquid have a range of kinetic energies.  
Some molecules near the surface have enough energy to overcome intermolecular attractions and escape into the gas phase.  
This process is called **evaporation**.

At the same time, vapor molecules collide with the liquid surface and may lose energy and return to the liquid phase.  
This reverse process is called **condensation**.

---

### 2. Dynamic Equilibrium and Vapor Pressure

In a closed container, evaporation initially dominates, increasing the number of vapor molecules.  
As vapor density increases, the rate of condensation rises until it equals the rate of evaporation.

At this point, the system reaches **dynamic equilibrium**, and the pressure exerted by the vapor is called the **vapor pressure** $P_v$.

Vapor pressure depends only on the **temperature** and the **nature of the liquid**, not on the volume of the container or the amount of liquid present.

---

### 3. Temperature Dependence of Vapor Pressure

As temperature increases:
- More molecules have sufficient energy to escape the liquid,
- The rate of evaporation increases,
- The equilibrium vapor pressure increases rapidly.

For most liquids, vapor pressure rises exponentially with temperature.

---

### Example 18.4. Vapor Pressure of Water

At:
- $20^\circ\text{C}$, water has $P_v \approx 2.3~\text{kPa}$,  
- $100^\circ\text{C}$, water has $P_v = 101~\text{kPa}$.

At $100^\circ\text{C}$, the vapor pressure equals atmospheric pressure, and boiling occurs.

---

### 4. Boiling and Vapor Pressure

**Boiling** occurs when the vapor pressure of a liquid equals the external pressure:

$$
P_v = P_{\text{ext}}.
$$

At this point, vapor bubbles can form throughout the liquid, not just at the surface.

If external pressure decreases, boiling occurs at a lower temperature.

---

### 5. Saturated and Unsaturated Vapor

- **Saturated vapor:** Vapor in equilibrium with its liquid at a given temperature; its pressure equals the vapor pressure.  
- **Unsaturated vapor:** Vapor whose pressure is below the equilibrium vapor pressure.

Air can hold only a limited amount of water vapor at a given temperature.

---

### 6. Humidity

Humidity describes the amount of water vapor present in the air.

#### Absolute Humidity

Absolute humidity is the mass of water vapor per unit volume of air:

$$
\text{Absolute humidity} = \frac{m_{\text{vapor}}}{V}.
$$

It is rarely used in weather reports.

---

### 7. Relative Humidity

**Relative humidity (RH)** compares the actual water vapor content of air to the maximum possible at the same temperature:

$$
\text{RH} = \frac{P_v}{P_{v,\text{sat}}} \times 100\%.
$$

Here:
- $P_v$ is the partial pressure of water vapor in the air,  
- $P_{v,\text{sat}}$ is the saturation vapor pressure at that temperature.

---

### Example 18.5. Relative Humidity Calculation

At $25^\circ\text{C}$, the saturation vapor pressure of water is $3.2~\text{kPa}$.  
If the actual water vapor pressure is $1.6~\text{kPa}$:

$$
\text{RH} = \frac{1.6}{3.2}\times100\% = 50\%.
$$

---

### 8. Dew Point

The **dew point** is the temperature at which air becomes saturated when cooled at constant pressure.  
At this temperature:

$$
P_v = P_{v,\text{sat}}.
$$

Cooling air below the dew point causes condensation (dew, fog, clouds).

---

### 9. Evaporative Cooling

Evaporation removes the highest-energy molecules from a liquid, reducing the average kinetic energy of the remaining molecules.  
This results in **cooling** of the liquid.

Evaporative cooling explains:
- Cooling of skin by sweating,  
- Cooling of liquids in open containers,  
- Refrigeration processes.

---

### 10. Atmospheric Applications

Humidity influences:
- Weather and cloud formation,  
- Human comfort and heat regulation,  
- Efficiency of evaporative cooling.

High humidity reduces evaporation rates, making hot environments feel hotter.

---

### Conceptual Notes

Vapor pressure is the pressure exerted by a vapor in equilibrium with its liquid and depends only on temperature. Boiling occurs when vapor pressure equals external pressure. Humidity describes the amount of water vapor in air, with relative humidity comparing actual vapor content to saturation. The dew point indicates when condensation begins, and evaporative cooling arises from the escape of high-energy molecules.

---

In the next section, we examine why the **boiling temperature of water decreases with altitude**, using vapor pressure and atmospheric pressure.


## 18.5. Temperature Decrease of Boiling Water with Altitude

The temperature at which a liquid boils depends on the **external pressure** acting on it.  
As altitude increases, atmospheric pressure decreases, and therefore the **boiling temperature of water decreases**.  
This effect has important consequences for cooking, weather, and physiology at high elevations.

---

### 1. Boiling Condition Revisited

A liquid boils when its **vapor pressure** equals the surrounding pressure:

$$
P_v(T_b) = P_{\text{ext}}.
$$

At sea level, the external pressure is approximately

$$
P_{\text{atm}} = 1.01\times10^5~\text{Pa},
$$

and water boils at

$$
T_b = 100^\circ\text{C}.
$$

If the external pressure is reduced, the temperature at which this equality is satisfied is lower.

---

### 2. Atmospheric Pressure and Altitude

Atmospheric pressure decreases with altitude because there is less air above exerting weight.  
Over moderate altitude ranges, pressure decreases approximately exponentially:

$$
P(h) = P_0 e^{-h/H},
$$

where:
- $P_0$ is sea-level pressure,
- $h$ is altitude,
- $H \approx 8.0~\text{km}$ is the atmospheric scale height.

Thus, even modest increases in altitude produce noticeable pressure reductions.

---

### 3. Boiling Temperature as a Function of Pressure

Because vapor pressure increases rapidly with temperature, a small decrease in pressure corresponds to a significant decrease in boiling temperature.

For water:
- At $P = 101~\text{kPa}$ → $T_b = 100^\circ\text{C}$,
- At $P = 80~\text{kPa}$ → $T_b \approx 93^\circ\text{C}$,
- At $P = 60~\text{kPa}$ → $T_b \approx 86^\circ\text{C}$.

---

### Example 18.6. Boiling Water at High Altitude

At an altitude where atmospheric pressure is $75~\text{kPa}$, find the approximate boiling temperature of water.

Using vapor pressure data for water, the boiling temperature is approximately

$$
T_b \approx 92^\circ\text{C}.
$$

Water therefore boils well below $100^\circ\text{C}$ at this altitude.

---

### 4. Physical Explanation

At lower external pressure:
- Less energy is required for vapor bubbles to expand against the surroundings,  
- Vapor bubbles form at lower temperatures,  
- The liquid transitions to the gas phase more easily.

Boiling does **not** mean that the liquid is hot in an absolute sense; it only means that vapor pressure equals ambient pressure.

---

### 5. Consequences for Cooking

Because boiling water at high altitude has a lower temperature:
- Food cooks more slowly,  
- Chemical and biological reactions proceed at reduced rates,  
- Pressure cookers are often used to raise the boiling temperature by increasing pressure.

This explains why cooking instructions differ for high-altitude locations.

---

### 6. Pressure Cookers

A pressure cooker increases the pressure inside a sealed container, raising the boiling temperature of water.

For example:
- At $P \approx 200~\text{kPa}$, water boils at about $120^\circ\text{C}$.

Higher boiling temperature allows food to cook faster and more thoroughly.

---

### 7. Extreme Altitudes

At very high altitudes, such as on Mount Everest:
- Atmospheric pressure is about $34~\text{kPa}$,
- Water boils at approximately $70^\circ\text{C}$.

At this temperature, boiling water is not hot enough to cook many foods properly.

---

### 8. Human Physiology and Boiling

Reduced pressure at high altitude affects not only boiling water but also **evaporation in the human body**.

- Lower pressure enhances evaporation, increasing dehydration risk,  
- Reduced oxygen partial pressure affects respiration,  
- Boiling of bodily fluids does not occur due to internal pressure regulation.

---

### 9. Relation to Phase Diagrams

On a phase diagram, lowering pressure moves the system horizontally toward the vapor region.  
The intersection with the liquid–gas boundary occurs at lower temperature.

This graphical representation clarifies why boiling temperature depends on pressure.

---

### Conceptual Notes

Boiling occurs when a liquid’s vapor pressure equals the surrounding pressure. Because atmospheric pressure decreases with altitude, the boiling temperature of water decreases as well. This effect influences cooking, weather, and human activity at high elevations. Pressure cookers exploit the same principle in reverse by increasing pressure to raise the boiling temperature.

---

## 18.6. Van der Waals Equation of State

The ideal gas law provides an excellent approximation for many gases at low pressure and high temperature, but it fails when gas molecules are close together or when intermolecular forces become significant.  
The **van der Waals equation of state** modifies the ideal gas law to account for **finite molecular size** and **intermolecular attractions**, providing a more realistic description of real gases.

---

### 1. Motivation for Corrections to the Ideal Gas Law

The ideal gas law assumes that:
- Gas molecules have negligible volume,
- There are no intermolecular forces except during collisions.

In real gases:
- Molecules occupy a finite volume,
- Attractive forces act between molecules, especially at short distances.

These effects become important at **high pressures** and **low temperatures**, where molecules are closer together and move more slowly.

---

### 2. Correction for Finite Molecular Volume

Because gas molecules occupy space, the volume available for molecular motion is less than the container volume $V$.

If $b$ represents the **effective volume per mole** excluded by the molecules themselves, then the available volume is

$$
V_{\text{eff}} = V - nb.
$$

This correction reduces the volume appearing in the ideal gas law.

---

### 3. Correction for Intermolecular Attractions

Attractive forces between molecules reduce the pressure exerted on the container walls because molecules near the walls are pulled inward by neighboring molecules.

This pressure reduction is proportional to the square of the molecular density:

$$
\Delta P \propto \left(\frac{n}{V}\right)^2.
$$

Introducing a constant $a$ to quantify the strength of intermolecular attraction, the corrected pressure becomes

$$
P_{\text{eff}} = P + a\frac{n^2}{V^2}.
$$

---

### 4. Van der Waals Equation

Including both corrections leads to the **van der Waals equation of state**:

$$
\left( P + a\frac{n^2}{V^2} \right)(V - nb) = nRT.
$$

Here:
- $a$ measures the strength of attractive forces,
- $b$ accounts for the finite volume of molecules,
- $n$ is the number of moles,
- $R$ is the universal gas constant.

Each gas has its own characteristic values of $a$ and $b$.

---

### 5. Interpretation of the Constants $a$ and $b$

- A larger value of $a$ indicates stronger intermolecular attractions and easier liquefaction.  
- A larger value of $b$ indicates larger molecular size.

For example, gases with strong attractions, such as ammonia, have large $a$ values.

---

### 6. Reduction to the Ideal Gas Law

At low pressure and high temperature:
- $V \gg nb$,  
- $a n^2 / V^2 \ll P$.

Under these conditions, the van der Waals equation reduces to

$$
PV \approx nRT,
$$

recovering the ideal gas law.

---

### 7. Critical Constants from the Van der Waals Equation

The van der Waals equation predicts the existence of a **critical point**.  
For one mole ($n = 1$), the critical temperature, pressure, and volume are

$$
T_c = \frac{8a}{27Rb},
$$

$$
P_c = \frac{a}{27b^2},
$$

$$
V_c = 3b.
$$

These values mark the end of the liquid–gas coexistence curve.

---

### Example 18.7. Critical Constants of Carbon Dioxide

For carbon dioxide:
- $a = 0.364~\text{Pa·m}^6/\text{mol}^2$,
- $b = 4.27\times10^{-5}~\text{m}^3/\text{mol}$.

Using the formulas above yields values close to the experimentally observed critical temperature and pressure.

---

### 8. Isotherms and Phase Behavior

Van der Waals isotherms plotted on a $P$–$V$ diagram show characteristic behavior:
- At high temperatures, smooth curves similar to ideal gas behavior,
- At temperatures below $T_c$, regions where pressure decreases with decreasing volume, indicating phase instability.

These unstable regions are replaced in reality by horizontal segments corresponding to phase coexistence.

---

### 9. Successes and Limitations

The van der Waals equation qualitatively explains:
- Deviations from ideal gas behavior,
- Existence of a critical point,
- Gas–liquid phase transitions.

However, it is still an approximation and does not provide precise quantitative agreement for all gases.

---

### Conceptual Notes

The van der Waals equation introduces simple corrections to the ideal gas law to account for molecular size and intermolecular attraction. It bridges the gap between ideal gas behavior and real gas phenomena, including liquefaction and critical behavior. Despite its limitations, it provides deep physical insight into why real gases deviate from ideal behavior.

---

## 18.7. Mean Free Path

In a gas, molecules move freely between collisions with other molecules.  
The **mean free path** is the average distance a molecule travels between successive collisions.  
It plays a central role in understanding diffusion, viscosity, thermal conductivity, and the validity of the ideal gas model.

---

### 1. Definition of Mean Free Path

The **mean free path** $\lambda$ is defined as the average distance a molecule travels before colliding with another molecule.

It depends on:
- The number density of molecules,
- The effective size of molecules,
- The frequency of collisions.

---

### 2. Molecular Collision Model

Molecules are modeled as hard spheres of effective diameter $d$.  
A collision occurs when the centers of two molecules come within a distance $d$.

As a molecule moves, it sweeps out a cylindrical volume. Any other molecule whose center lies within this volume will collide with it.

---

### 3. Number Density

The **number density** $n$ is the number of molecules per unit volume:

$$
n = \frac{N}{V}.
$$

Using the ideal gas law in molecular form,

$$
PV = NkT,
$$

the number density can be written as

$$
n = \frac{P}{kT}.
$$

---

### 4. Expression for Mean Free Path

Taking into account the relative motion of molecules, the mean free path is

$$
\lambda = \frac{1}{\sqrt{2}\,\pi d^2 n}.
$$

Substituting $n = P/(kT)$ gives

$$
\lambda = \frac{kT}{\sqrt{2}\,\pi d^2 P}.
$$

This shows that:
- Mean free path increases with temperature,
- Mean free path decreases with pressure.

---

### Example 18.8. Mean Free Path of Air Molecules

For air at room temperature:
- $T = 300~\text{K}$,
- $P = 1.01\times10^5~\text{Pa}$,
- Molecular diameter $d \approx 3.7\times10^{-10}~\text{m}$.

$$
\lambda = \frac{(1.38\times10^{-23})(300)}{\sqrt{2}\pi(3.7\times10^{-10})^2(1.01\times10^5)}
         \approx 6.8\times10^{-8}~\text{m}.
$$

The mean free path is about **70 nm**, much larger than molecular size.

---

### 5. Dependence on Pressure and Altitude

As pressure decreases with altitude, the mean free path increases rapidly.  
At very high altitudes, the mean free path becomes comparable to or larger than the size of containers or spacecraft, and the gas no longer behaves as a continuous fluid.

This regime is important in:
- Upper-atmosphere physics,
- Spacecraft reentry,
- Vacuum systems.

---

### 6. Mean Free Path and Gas Properties

Mean free path influences:
- **Viscosity:** Momentum transport between layers of gas,
- **Thermal conductivity:** Energy transport,
- **Diffusion:** Mass transport.

These transport properties depend more on molecular speed and mean free path than on pressure alone.

---

### 7. Validity of the Ideal Gas Approximation

The ideal gas model is valid when:
- Mean free path is much larger than molecular size,
- Collisions are infrequent compared to molecular spacing,
- Intermolecular forces are negligible except during collisions.

These conditions are satisfied at low pressures and moderate temperatures.

---

### Conceptual Notes

The mean free path characterizes how often gas molecules collide. It depends inversely on pressure and directly on temperature. Under ordinary conditions, the mean free path is much larger than molecular dimensions, justifying the ideal gas approximation. At low pressures, mean free paths become large and gas behavior deviates from continuum assumptions.

---

## 18.8. Diffusion

**Diffusion** is the spontaneous spreading of particles from regions of high concentration to regions of low concentration due to random molecular motion.  
It occurs in gases, liquids, and solids, but is most rapid in gases because molecules move freely and collide frequently.

---

### 1. Physical Origin of Diffusion

Diffusion arises from the random thermal motion of molecules.  
Although each molecule moves randomly, a net flow occurs whenever there is a concentration gradient.

No external force is required; diffusion is a purely statistical process.

---

### 2. Diffusion in Gases

In gases, diffusion is relatively fast because:
- Molecules have high speeds,
- Mean free paths are relatively long.

Diffusion explains everyday phenomena such as:
- Smells spreading through air,
- Mixing of gases without stirring.

---

### 3. Fick’s First Law (Qualitative Form)

The rate of diffusion is proportional to the concentration gradient:

$$
\text{Flux} \propto -\frac{dC}{dx}.
$$

The negative sign indicates that diffusion occurs in the direction of decreasing concentration.

---

### 4. Diffusion Coefficient

Quantitatively, Fick’s first law is written as

$$
J = -D \frac{dC}{dx},
$$

where:
- $J$ is the particle flux,
- $D$ is the **diffusion coefficient**,
- $C$ is concentration.

The diffusion coefficient depends on:
- Mean free path,
- Molecular speed,
- Temperature,
- Pressure.

---

### 5. Relation to Molecular Motion

In kinetic theory, the diffusion coefficient for a gas is approximately

$$
D \sim \frac{1}{3} \lambda \overline{v},
$$

where:
- $\lambda$ is the mean free path,
- $\overline{v}$ is the average molecular speed.

This shows that diffusion is faster at higher temperatures and lower pressures.

---

### Example 18.9. Why Diffusion Is Faster in Gases

Because gases have:
- Larger mean free paths,
- Higher molecular speeds,

their diffusion coefficients are much larger than those of liquids or solids, where molecular motion is strongly constrained.

---

### 6. Graham’s Law of Diffusion

For gases at the same temperature and pressure, the diffusion rate is inversely proportional to the square root of molecular mass:

$$
\text{Rate} \propto \frac{1}{\sqrt{m}}.
$$

This follows from the dependence of molecular speed on mass.

---

### Example 18.10. Hydrogen and Oxygen Diffusion

Hydrogen molecules diffuse about four times faster than oxygen molecules because oxygen has roughly 16 times the molecular mass.

---

### 7. Diffusion vs. Effusion

- **Diffusion:** Spreading through random motion and collisions,  
- **Effusion:** Flow through a tiny hole without collisions.

Effusion obeys the same mass dependence as diffusion but occurs under different physical conditions.

---

### 8. Biological and Physical Applications

Diffusion plays a crucial role in:
- Gas exchange in lungs,
- Transport of nutrients and waste in cells,
- Mixing of atmospheric gases,
- Semiconductor fabrication.

---

### Conceptual Notes

Diffusion is a natural consequence of random molecular motion and concentration gradients. Its rate depends on molecular speed and mean free path, making it strongly temperature- and pressure-dependent. Diffusion provides a clear example of how macroscopic transport phenomena emerge from microscopic particle dynamics.

## 19.1. Heat as Energy Transfer

Heat is a form of **energy transfer** that occurs solely because of a **temperature difference** between a system and its surroundings.  
Unlike work, which is associated with forces acting through distances, heat is associated with microscopic energy transfer arising from random molecular motion.

Heat is **not a property** of a system. A system does not “contain heat.” Instead, heat is energy **in transit** across the boundary of a system due to a temperature difference.

---

### 1. Heat and Temperature

Temperature characterizes the thermal state of a system, whereas heat describes energy transfer that occurs when two systems at different temperatures are placed in thermal contact.

If:
- $T_{\text{hot}} > T_{\text{cold}}$,

then heat flows spontaneously from the hotter system to the colder one until thermal equilibrium is reached.

This direction of heat flow is an observed fact of nature and does not depend on the material or size of the systems.

---

### 2. Microscopic Interpretation of Heat Transfer

At the microscopic level, heat transfer occurs through interactions between atoms and molecules:

- Faster-moving particles in a hotter object collide with slower-moving particles in a colder object.
- Energy is transferred during these collisions.
- The average kinetic energy of the colder system increases, while that of the hotter system decreases.

This process continues until the average kinetic energies, and therefore the temperatures, become equal.

---

### 3. Units of Heat

Because heat is a form of energy, its SI unit is the **joule (J)**.

Historically, heat was measured in calories. One calorie is defined as the amount of heat required to raise the temperature of $1~\text{g}$ of water by $1^\circ\text{C}$.

The mechanical equivalent of heat is:

$$
1~\text{cal} = 4.186~\text{J}.
$$

The use of joules emphasizes that heat and work are fundamentally the same physical quantity: energy.

---

### 4. Heat vs. Work

Both heat and work transfer energy, but they differ in mechanism:

- **Heat:** Energy transfer due to temperature difference.
- **Work:** Energy transfer due to forces acting through distances.

For example:
- Compressing a gas transfers energy as work.
- Placing a hot object in contact with a cold one transfers energy as heat.

In many thermodynamic processes, both heat and work may occur simultaneously.

---

### 5. Sign Convention for Heat

By convention:
- Heat **added to** a system is taken as positive ($Q > 0$).
- Heat **removed from** a system is taken as negative ($Q < 0$).

This convention will be used consistently in applying the first law of thermodynamics.

---

### Example 19.1. Heat Transfer Between Two Objects

A hot metal block at $80^\circ\text{C}$ is placed in contact with a cooler block at $20^\circ\text{C}$.  
Heat flows from the hotter block to the cooler block until both reach the same final temperature.

During this process:
- The hot block loses energy as heat ($Q < 0$).
- The cool block gains energy as heat ($Q > 0$).

The total energy is conserved.

---

### 6. Calorimetry Preview

The measurement of heat transfer is called **calorimetry**.  
In calorimetry problems, heat gained by one part of a system is balanced by heat lost by another part, assuming no energy escapes to the surroundings.

This principle will be used extensively in later sections.

---

### 7. Heat Transfer Does Not Imply Temperature Change

Heat transfer does not always cause a temperature change.  
During a **phase change** (such as melting or boiling), heat is transferred while the temperature remains constant.

This distinction highlights the difference between heat and temperature.

---

### Conceptual Notes

Heat is energy in transit due to temperature difference and is not a stored quantity. It flows spontaneously from higher to lower temperature systems. Both heat and work transfer energy, but through different physical mechanisms. Using joules as the unit of heat emphasizes the equivalence of thermal and mechanical energy.

---

## 19.2. Internal Energy

The **internal energy** of a system is the total microscopic energy associated with the random motion and interactions of its atoms and molecules.  
It includes all forms of energy at the molecular level but excludes the macroscopic kinetic and potential energy of the system as a whole.

Internal energy is a **state function**, meaning its value depends only on the current state of the system and not on how the system reached that state.

---

### 1. Components of Internal Energy

The internal energy $U$ of a substance consists of several contributions:

- **Translational kinetic energy** of molecules,
- **Rotational and vibrational kinetic energy** (especially in molecules),
- **Potential energy** due to intermolecular forces,
- **Internal electronic energy**.

For many applications, especially ideal gases, only translational kinetic energy plays a significant role.

---

### 2. Internal Energy and Temperature

For an **ideal monatomic gas**, the internal energy depends only on temperature:

$$
U = \frac{3}{2} nRT.
$$

This result follows from the kinetic theory relation:

$$
\frac{1}{2}m\overline{v^2} = \frac{3}{2}kT.
$$

Because internal energy depends only on temperature for an ideal gas, any process that changes temperature changes internal energy, regardless of the path taken.

---

### 3. Change in Internal Energy

The change in internal energy is denoted by

$$
\Delta U = U_f - U_i.
$$

Since internal energy is a state function, $\Delta U$ depends only on the initial and final states, not on the details of the process.

---

### Example 19.2. Internal Energy Change of an Ideal Gas

A sample of $2.0~\text{mol}$ of an ideal monatomic gas is heated from $300~\text{K}$ to $450~\text{K}$.

$$
\Delta U = \frac{3}{2}nR\Delta T
         = \frac{3}{2}(2.0)(8.31)(150)
         = 3740~\text{J}.
$$

The internal energy increases by **3.7 kJ**.

---

### 4. Internal Energy vs. Heat

Internal energy is **not** the same as heat.

- Internal energy is a property of the system.
- Heat is energy transferred due to temperature difference.

A system can have its internal energy changed by heat transfer, by work done on or by the system, or by both.

---

### 5. Internal Energy vs. Mechanical Energy

Internal energy excludes:
- Macroscopic kinetic energy of the system’s center of mass,
- Macroscopic gravitational potential energy.

For example, a moving gas-filled container has kinetic energy of motion in addition to the internal energy of the gas inside.

---

### 6. Solids and Liquids

For solids and liquids, internal energy includes significant contributions from:
- Vibrational motion of atoms,
- Potential energy associated with intermolecular forces.

As a result, the internal energy of solids and liquids does not depend on temperature alone, and simple formulas like $U = \frac{3}{2}nRT$ do not apply.

---

### 7. Reference Level of Internal Energy

Only **changes** in internal energy can be measured experimentally.  
The absolute value of $U$ cannot be directly measured, so an arbitrary reference level is chosen.

Thermodynamic laws depend only on differences $\Delta U$.

---

### 8. Microscopic Interpretation

Increasing internal energy corresponds to:
- Higher molecular speeds,
- Larger vibrational amplitudes,
- Weaker binding between molecules (at high energies).

During phase changes, internal energy changes significantly even though temperature remains constant.

---

### Conceptual Notes

Internal energy represents the total microscopic energy of a system and is a state function. For ideal gases, it depends only on temperature, while for solids and liquids it also depends on intermolecular interactions. Internal energy changes can result from heat transfer, work, or both, but internal energy itself is not heat.

---

In the next section, we quantify how much heat is required to change the temperature of a substance by introducing **specific heat**.


## 19.3. Specific Heat

Different substances require different amounts of heat to produce the same change in temperature.  
This property is quantified by the **specific heat**, which measures how much energy must be transferred as heat to change the temperature of a given amount of material.

---

### 1. Definition of Specific Heat

The **specific heat** $c$ of a substance is defined as the amount of heat $Q$ required to raise the temperature of a mass $m$ by an amount $\Delta T$:

$$
Q = mc\Delta T.
$$

Here:
- $Q$ is the heat transferred,
- $m$ is the mass of the substance,
- $c$ is the specific heat,
- $\Delta T$ is the temperature change.

The SI unit of specific heat is $\text{J/(kg·K)}$.

---

### 2. Physical Meaning of Specific Heat

A substance with a **large specific heat** requires a large amount of heat to change its temperature.  
A substance with a **small specific heat** changes temperature easily when heat is added or removed.

Water has an unusually high specific heat, which is why it is effective at storing thermal energy and moderating climate.

---

### 3. Specific Heat and Temperature Change

Rewriting the defining equation:

$$
\Delta T = \frac{Q}{mc}.
$$

For a given heat input:
- Large $c$ → small temperature change,
- Small $c$ → large temperature change.

---

### Example 19.3. Heating Copper and Water

Suppose $500~\text{J}$ of heat is added to:
- $0.50~\text{kg}$ of copper ($c = 390~\text{J/(kg·K)}$),
- $0.50~\text{kg}$ of water ($c = 4186~\text{J/(kg·K)}$).

For copper:

$$
\Delta T = \frac{500}{(0.50)(390)} = 2.6~\text{K}.
$$

For water:

$$
\Delta T = \frac{500}{(0.50)(4186)} = 0.24~\text{K}.
$$

Copper warms much more than water for the same heat input.

---

### 4. Molar Specific Heat

Specific heat can also be defined per mole rather than per kilogram.  
The **molar specific heat** $C$ is defined by

$$
Q = nC\Delta T,
$$

where $n$ is the number of moles.

The SI unit of molar specific heat is $\text{J/(mol·K)}$.

---

### 5. Relation Between Specific Heat and Internal Energy

When heat is added to a system and no work is done, the heat transfer increases internal energy:

$$
Q = \Delta U.
$$

For an ideal monatomic gas at constant volume:

$$
\Delta U = \frac{3}{2}nR\Delta T,
$$

so the molar specific heat at constant volume is

$$
C_V = \frac{3}{2}R.
$$

---

### 6. Specific Heat of Solids

For many solids at room temperature, the molar specific heat is approximately

$$
C \approx 3R.
$$

This empirical result is known as the **Dulong–Petit law** and reflects the vibrational degrees of freedom of atoms in a solid lattice.

---

### 7. Temperature Dependence

Specific heat is not always constant.  
It may vary with:
- Temperature,
- Phase of the material,
- Molecular structure.

At very low temperatures, specific heats of solids decrease significantly due to quantum effects.

---

### 8. Heat Capacity

The **heat capacity** $C_{\text{sys}}$ of an object is the total heat required to raise its temperature by $1~\text{K}$:

$$
C_{\text{sys}} = mc.
$$

Heat capacity depends on both the material and the amount of substance.

---

### Example 19.4. Heat Capacity of an Aluminum Block

An aluminum block of mass $2.0~\text{kg}$ has specific heat  
$c = 900~\text{J/(kg·K)}$.

$$
C_{\text{sys}} = (2.0)(900) = 1800~\text{J/K}.
$$

---

### Conceptual Notes

Specific heat quantifies a material’s resistance to temperature change when heat is added or removed. It depends on microscopic energy storage mechanisms such as translational, rotational, and vibrational motion. Water’s large specific heat plays a crucial role in climate regulation and thermal stability.

---

## 19.4. Calorimetry — Solving Problems

**Calorimetry** is the quantitative study of heat transfer.  
It is based on the principle of **energy conservation**, which states that, in an isolated system, the total energy remains constant.

In calorimetry problems, heat lost by one part of a system is equal to the heat gained by another part, assuming no heat is exchanged with the surroundings.

---

### 1. Fundamental Principle of Calorimetry

For an isolated system:

$$
\sum Q = 0.
$$

This means that the algebraic sum of all heat transfers within the system is zero.

Heat gained is taken as positive, and heat lost is taken as negative.

---

### 2. Temperature Equilibrium

When objects at different temperatures are placed in thermal contact:
- Heat flows from the hotter object to the cooler one,
- The process continues until all objects reach a common **final temperature** $T_f$,
- Thermal equilibrium is then established.

---

### 3. Heat Transfer for Temperature Changes

When no phase change occurs, the heat transferred to or from a substance is

$$
Q = mc\Delta T = mc(T_f - T_i).
$$

This formula applies to solids, liquids, and gases undergoing temperature change only.

---

### 4. Simple Calorimetry Setup

A typical calorimetry problem involves:
- Two or more substances,
- Known masses and specific heats,
- Known initial temperatures,
- An unknown final equilibrium temperature.

The calorimeter itself may absorb heat and must sometimes be included in the energy balance.

---

### Example 19.5. Mixing Hot and Cold Water

A mass of $0.20~\text{kg}$ of water at $80^\circ\text{C}$ is mixed with $0.30~\text{kg}$ of water at $20^\circ\text{C}$.  
Assume no heat loss to the surroundings.

Let $T_f$ be the final temperature.

Heat lost by hot water:
$$
Q_1 = (0.20)(4186)(T_f - 80).
$$

Heat gained by cold water:
$$
Q_2 = (0.30)(4186)(T_f - 20).
$$

Energy conservation:
$$
Q_1 + Q_2 = 0.
$$

Solving:

$$
0.20(T_f - 80) + 0.30(T_f - 20) = 0,
$$

$$
0.50T_f - 22 = 0,
$$

$$
T_f = 44^\circ\text{C}.
$$

---

### 5. Including the Calorimeter

If the calorimeter has heat capacity $C_{\text{cal}}$, it absorbs heat according to

$$
Q_{\text{cal}} = C_{\text{cal}}(T_f - T_i).
$$

The energy balance becomes:

$$
\sum Q_{\text{objects}} + Q_{\text{cal}} = 0.
$$

Neglecting the calorimeter can introduce significant error if its heat capacity is comparable to that of the substances involved.

---

### Example 19.6. Metal Sample in Water

A $0.10~\text{kg}$ metal sample at $200^\circ\text{C}$ is placed in $0.25~\text{kg}$ of water at $20^\circ\text{C}$.  
The final temperature is $24^\circ\text{C}$.  
Find the specific heat of the metal.

Heat gained by water:
$$
Q_w = (0.25)(4186)(24 - 20) = 4186~\text{J}.
$$

Heat lost by metal:
$$
Q_m = (0.10)c_m(24 - 200).
$$

Energy conservation:
$$
Q_w + Q_m = 0.
$$

Solving for $c_m$:

$$
c_m = \frac{4186}{0.10(200 - 24)} = 238~\text{J/(kg·K)}.
$$

---

### 6. Calorimetry with Phase Changes

When phase changes occur, heat transfer includes **latent heat** terms in addition to temperature-change terms.

The total heat is the sum of all contributions:

$$
Q_{\text{total}} = \sum mc\Delta T + \sum mL.
$$

Phase changes are treated in the order in which they occur.

---

### 7. Common Assumptions and Pitfalls

- Neglecting heat loss to the surroundings,
- Forgetting to include the calorimeter,
- Mixing units (Celsius vs. kelvin),
- Incorrect sign conventions.

Careful bookkeeping of heat flows avoids these errors.

---

### 8. Calorimetry as an Experimental Tool

Calorimetry is used to:
- Measure specific heats,
- Determine latent heats,
- Study chemical reactions,
- Analyze biological and industrial processes.

Its accuracy depends on insulation and careful measurement.

---

### Conceptual Notes

Calorimetry applies energy conservation to systems exchanging heat. Heat lost by hot components equals heat gained by cooler components. By tracking temperature changes and using specific heat relations, unknown thermal properties can be determined. Including all parts of the system, including the calorimeter, is essential for accurate results.

---

In the next section, we introduce **latent heat**, which describes heat transfer during phase changes at constant temperature.


## 19.5. Latent Heat

When a substance undergoes a **phase change**, such as melting or boiling, heat is transferred without any change in temperature.  
The energy involved in a phase change is called **latent heat**.  
This energy is used to change the microscopic arrangement of particles rather than to increase their average kinetic energy.

---

### 1. Phase Changes and Temperature

During a phase change:
- The temperature of the substance remains constant,
- Heat is either absorbed or released,
- The substance exists in two phases simultaneously.

Examples include:
- Ice melting into water at $0^\circ\text{C}$,
- Water boiling into steam at $100^\circ\text{C}$ (at standard pressure).

---

### 2. Definition of Latent Heat

The heat $Q$ required for a phase change of mass $m$ is given by

$$
Q = mL,
$$

where $L$ is the **latent heat** of the phase change.

Latent heat has units of $\text{J/kg}$.

---

### 3. Latent Heat of Fusion

The **latent heat of fusion** $L_f$ is the heat required to change a substance from solid to liquid at constant temperature.

For water:
$$
L_f = 3.34\times10^5~\text{J/kg}.
$$

Melting requires heat input, while freezing releases the same amount of heat.

---

### 4. Latent Heat of Vaporization

The **latent heat of vaporization** $L_v$ is the heat required to change a substance from liquid to gas at constant temperature.

For water at $100^\circ\text{C}$:
$$
L_v = 2.26\times10^6~\text{J/kg}.
$$

Vaporization requires much more energy than melting because molecules must fully overcome intermolecular attractions.

---

### Example 19.7. Melting Ice

How much heat is required to melt $0.50~\text{kg}$ of ice at $0^\circ\text{C}$?

$$
Q = mL_f = (0.50)(3.34\times10^5) = 1.67\times10^5~\text{J}.
$$

---

### 5. Heating Curves

A **heating curve** shows how the temperature of a substance changes as heat is added.

Key features:
- Sloped regions correspond to temperature changes,
- Flat regions correspond to phase changes,
- Heat input during flat regions goes entirely into latent heat.

---

### 6. Microscopic Interpretation

During melting:
- Molecular vibrations increase until bonds weaken,
- Particles gain freedom to move past one another.

During boiling:
- Molecules must completely escape attractive forces,
- Vapor bubbles form throughout the liquid.

Temperature remains constant because average kinetic energy does not change.

---

### 7. Cooling and Phase Changes

The same latent heats apply in reverse:
- Freezing releases heat equal to $mL_f$,
- Condensation releases heat equal to $mL_v$.

This released heat plays a major role in weather and climate processes.

---

### Example 19.8. Condensation Heating

When $0.10~\text{kg}$ of steam condenses at $100^\circ\text{C}$:

$$
Q = (0.10)(2.26\times10^6) = 2.26\times10^5~\text{J}.
$$

This heat is transferred to the surroundings.

---

### 8. Phase Change Sequences

Many problems involve multiple stages, such as:
- Heating ice to $0^\circ\text{C}$,
- Melting ice,
- Heating water to $100^\circ\text{C}$,
- Boiling water.

Each stage must be treated separately, and total heat is the sum of all contributions.

---

### Conceptual Notes

Latent heat describes energy transfer during phase changes at constant temperature. This energy changes molecular structure and separation rather than molecular speed. Vaporization requires far more energy than melting because molecules must fully escape intermolecular forces. Latent heat plays a critical role in atmospheric processes and energy transport.

---

## 19.6. The First Law of Thermodynamics

The **First Law of Thermodynamics** expresses the principle of **energy conservation** for thermodynamic systems.  
It relates changes in a system’s internal energy to the heat transferred to the system and the work done by or on the system.

---

### 1. Statement of the First Law

The First Law of Thermodynamics is written as

$$
\Delta U = Q - W,
$$

where:
- $\Delta U$ is the change in internal energy of the system,
- $Q$ is the heat added to the system,
- $W$ is the work done **by** the system.

This equation applies to all thermodynamic processes and systems.

---

### 2. Sign Conventions

By convention:
- $Q > 0$ if heat is added to the system,
- $Q < 0$ if heat leaves the system,
- $W > 0$ if the system does work on its surroundings,
- $W < 0$ if work is done on the system.

These conventions ensure consistency when applying the First Law.

---

### 3. Physical Meaning

The First Law states that energy added to a system as heat can either:
- Increase the internal energy of the system,
- Be converted into work done by the system,
- Or both.

Energy cannot be created or destroyed, only transferred or transformed.

---

### 4. Internal Energy Revisited

Internal energy includes all microscopic forms of energy associated with the system’s particles.  
A change in internal energy may result from:
- Heat transfer,
- Mechanical work,
- Electrical work,
- Other forms of energy transfer.

The First Law provides a unified framework for analyzing all such processes.

---

### 5. Work Done by a Gas

When a gas expands against an external pressure, it does work on its surroundings.  
For a gas expanding quasistatically at pressure $P$, the infinitesimal work is

$$
dW = P\,dV.
$$

For a finite volume change from $V_i$ to $V_f$:

$$
W = \int_{V_i}^{V_f} P\,dV.
$$

---

### Example 19.9. Work Done at Constant Pressure

A gas expands from $0.020~\text{m}^3$ to $0.030~\text{m}^3$ at constant pressure  
$P = 1.0\times10^5~\text{Pa}$.

$$
W = P(V_f - V_i)
  = (1.0\times10^5)(0.010)
  = 1000~\text{J}.
$$

The gas does $1000~\text{J}$ of work on its surroundings.

---

### 6. Special Cases of the First Law

#### No Heat Transfer ($Q = 0$)

$$
\Delta U = -W.
$$

Any work done by the system reduces its internal energy.

#### No Work Done ($W = 0$)

$$
\Delta U = Q.
$$

All added heat increases internal energy.

---

### 7. Cyclic Processes

In a cyclic process, the system returns to its initial state.  
Since internal energy is a state function:

$$
\Delta U = 0.
$$

Therefore,

$$
Q = W.
$$

The net heat absorbed equals the net work done.

---

### 8. Path Dependence

- Internal energy change $\Delta U$ depends only on initial and final states.
- Heat $Q$ and work $W$ depend on the **path** taken between those states.

Different processes connecting the same states can involve different amounts of heat and work.

---

### Example 19.10. Different Paths, Same $\Delta U$

A gas is taken between the same initial and final states by two different processes.  
Although $Q$ and $W$ differ for each path, the change in internal energy $\Delta U$ is the same.

---

### Conceptual Notes

The First Law of Thermodynamics applies energy conservation to thermodynamic systems. It connects heat transfer, work, and internal energy through a simple but universal equation. Internal energy is a state function, while heat and work depend on the process. This law provides the foundation for analyzing all thermal processes.

---

In the next section, we apply the First Law to specific **thermodynamic processes**, such as isothermal, isobaric, isochoric, and adiabatic processes.


## 19.7. Thermodynamic Processes and the First Law

A **thermodynamic process** describes how a system changes from one equilibrium state to another.  
Different processes are distinguished by which physical quantities are held constant during the change.  
Applying the **First Law of Thermodynamics** to these processes clarifies how heat, work, and internal energy are related.

---

### 1. General Framework

The First Law,

$$
\Delta U = Q - W,
$$

applies to all processes.  
What differs from one process to another is how $Q$, $W$, and $\Delta U$ are related through constraints on pressure, volume, or temperature.

---

### 2. Isothermal Process ($T = \text{constant}$)

In an **isothermal process**, the temperature remains constant.

For an ideal gas, internal energy depends only on temperature, so:

$$
\Delta U = 0.
$$

Therefore, the First Law becomes:

$$
Q = W.
$$

All heat added to the system is converted into work done by the system.

For a quasistatic isothermal expansion of an ideal gas:

$$
W = nRT \ln\!\left(\frac{V_f}{V_i}\right).
$$

---

### Example 19.11. Isothermal Expansion

A gas expands isothermally at $T = 300~\text{K}$ from $V_i = 0.010~\text{m}^3$ to  
$V_f = 0.020~\text{m}^3$ with $n = 1.0~\text{mol}$.

$$
W = (1.0)(8.31)(300)\ln(2) = 1730~\text{J}.
$$

The gas absorbs $1730~\text{J}$ of heat and does the same amount of work.

---

### 3. Isochoric Process ($V = \text{constant}$)

In an **isochoric process**, the volume is constant.

Because $dV = 0$:

$$
W = \int P\,dV = 0.
$$

Thus, the First Law reduces to:

$$
\Delta U = Q.
$$

All heat added to the system increases its internal energy.

For an ideal gas:

$$
Q = \Delta U = nC_V\Delta T.
$$

---

### 4. Isobaric Process ($P = \text{constant}$)

In an **isobaric process**, the pressure remains constant.

The work done by the gas is:

$$
W = P(V_f - V_i).
$$

The First Law gives:

$$
Q = \Delta U + W.
$$

For an ideal gas:

$$
Q = nC_P\Delta T,
$$

where $C_P$ is the molar specific heat at constant pressure.

---

### Example 19.12. Heating at Constant Pressure

One mole of an ideal gas is heated by $\Delta T = 50~\text{K}$ at constant pressure.

$$
Q = nC_P\Delta T.
$$

For a monatomic gas, $C_P = \frac{5}{2}R$, so:

$$
Q = (1.0)\frac{5}{2}(8.31)(50) = 1040~\text{J}.
$$

---

### 5. Adiabatic Process ($Q = 0$)

In an **adiabatic process**, no heat is exchanged with the surroundings:

$$
Q = 0.
$$

The First Law becomes:

$$
\Delta U = -W.
$$

Work done by the system comes entirely from its internal energy.

For an ideal gas undergoing a quasistatic adiabatic process:

$$
PV^\gamma = \text{constant},
$$

where $\gamma = C_P/C_V$.

---

### 6. Comparison of Processes

For the same initial and final volumes:

- Isothermal expansion produces less cooling than adiabatic expansion,
- Adiabatic expansion causes a temperature drop,
- Isochoric heating produces no work,
- Isobaric heating produces both temperature increase and work.

These differences arise from how energy is partitioned between heat, work, and internal energy.

---

### 7. Graphical Representation

On a $P$–$V$ diagram:
- Isothermal curves are hyperbolas,
- Adiabatic curves are steeper than isotherms,
- Isochoric processes are vertical lines,
- Isobaric processes are horizontal lines.

The area under the curve represents the work done.

---

### 8. Path Dependence of Heat and Work

Heat and work depend on the **path** taken between two states, not just on the endpoints.  
Different thermodynamic processes connecting the same initial and final states generally involve different values of $Q$ and $W$, even though $\Delta U$ is the same.

---

### Conceptual Notes

Thermodynamic processes are distinguished by constraints on temperature, volume, pressure, or heat flow. Applying the First Law to each process reveals how energy is redistributed between heat, work, and internal energy. Understanding these idealized processes provides a foundation for analyzing real engines and refrigeration cycles.

---

In the next section, we examine **molar specific heats for gases** and introduce the **equipartition of energy**, which explains why different gases store different amounts of internal energy.


## 19.8. Molar Specific Heats for Gases, and the Equipartition of Energy

The amount of heat required to raise the temperature of a gas depends not only on the amount of gas but also on the **process** by which heating occurs.  
For gases, it is therefore useful to define **molar specific heats** under well-defined conditions.  
The values of these specific heats can be understood using the **equipartition of energy**, which links macroscopic thermal behavior to microscopic degrees of freedom.

---

### 1. Molar Specific Heats of a Gas

The **molar specific heat** is the heat required to raise the temperature of one mole of gas by $1~\text{K}$.

Two molar specific heats are particularly important:

- $C_V$: molar specific heat at **constant volume**,  
- $C_P$: molar specific heat at **constant pressure**.

---

### 2. Molar Specific Heat at Constant Volume

At constant volume, no work is done by the gas:

$$
W = 0.
$$

From the First Law,

$$
\Delta U = Q.
$$

For an ideal gas, the internal energy change is proportional to the temperature change:

$$
\Delta U = n C_V \Delta T.
$$

Thus,

$$
Q = n C_V \Delta T.
$$

---

### 3. Molar Specific Heat at Constant Pressure

At constant pressure, the gas expands as it is heated and does work on its surroundings.

The heat added is used for:
- Increasing internal energy,
- Doing work.

From the First Law,

$$
Q = \Delta U + W.
$$

For an ideal gas:

$$
Q = n C_P \Delta T.
$$

Since additional energy is required to perform work, it follows that

$$
C_P > C_V.
$$

---

### 4. Relationship Between $C_P$ and $C_V$

For an ideal gas, the difference between the molar specific heats is

$$
C_P - C_V = R.
$$

This result holds for all ideal gases, regardless of molecular structure.

---

### 5. Monatomic Ideal Gas

A monatomic ideal gas (such as He or Ne) has only translational degrees of freedom.

Its internal energy is

$$
U = \frac{3}{2}nRT.
$$

Thus,

$$
C_V = \frac{3}{2}R,
$$

$$
C_P = C_V + R = \frac{5}{2}R.
$$

The ratio

$$
\gamma = \frac{C_P}{C_V} = \frac{5}{3}.
$$

---

### 6. Equipartition of Energy

The **equipartition theorem** states that, for a system in thermal equilibrium, each quadratic degree of freedom contributes an average energy of

$$
\frac{1}{2}kT
$$

per molecule.

---

### 7. Degrees of Freedom

A **degree of freedom** is an independent way in which a molecule can store energy.

Examples include:
- Translational motion (3 degrees),
- Rotational motion,
- Vibrational motion.

The number of active degrees of freedom determines the internal energy and specific heats.

---

### 8. Diatomic Gases

A diatomic molecule (such as N$_2$ or O$_2$) has:
- 3 translational degrees of freedom,
- 2 rotational degrees of freedom (at ordinary temperatures).

This gives a total of 5 active degrees of freedom.

The internal energy per mole is

$$
U = \frac{5}{2}nRT.
$$

Thus,

$$
C_V = \frac{5}{2}R,
$$

$$
C_P = \frac{7}{2}R,
$$

and

$$
\gamma = \frac{7}{5}.
$$

---

### 9. Vibrational Degrees of Freedom

Vibrational motion contributes additional degrees of freedom, but these become active only at sufficiently high temperatures.

Each vibrational mode contributes:
- One kinetic term,
- One potential term,

for a total of $kT$ per mode.

At room temperature, vibrational contributions are usually negligible for diatomic gases.

---

### 10. Experimental Agreement and Limitations

Measured values of molar specific heats generally agree with equipartition predictions at moderate temperatures.

However:
- At low temperatures, some degrees of freedom become inactive,
- Quantum effects modify the predictions.

Equipartition is therefore an approximation that works best at high temperatures.

---

### Example 19.13. Specific Heat of a Monatomic Gas

For a monatomic gas:

$$
C_V = \frac{3}{2}R = 12.5~\text{J/(mol·K)},
$$

$$
C_P = \frac{5}{2}R = 20.8~\text{J/(mol·K)}.
$$

---

### Conceptual Notes

Molar specific heats describe how gases absorb heat under different constraints. The equipartition theorem explains these values by counting the microscopic degrees of freedom available for energy storage. Monatomic gases store energy only in translation, while diatomic gases also store energy in rotation. Vibrational modes contribute at higher temperatures, and quantum effects limit equipartition at low temperatures.

---

In the next section, we apply these ideas to **adiabatic expansion of a gas**, where no heat is exchanged and temperature changes arise purely from work done.


## 19.9. Adiabatic Expansion of a Gas

An **adiabatic process** is a thermodynamic process in which **no heat is exchanged** between the system and its surroundings.  
Any change in the internal energy of the system arises solely from **work done** by or on the system.  
Adiabatic processes are especially important in rapidly occurring phenomena and in systems that are well insulated.

---

### 1. Definition of an Adiabatic Process

In an adiabatic process:

$$
Q = 0.
$$

Applying the First Law of Thermodynamics,

$$
\Delta U = Q - W,
$$

gives

$$
\Delta U = -W.
$$

Thus, when a gas expands adiabatically and does work on its surroundings, its internal energy decreases.

---

### 2. Temperature Change in Adiabatic Expansion

For an ideal gas, internal energy depends only on temperature.  
Therefore, adiabatic expansion causes a **decrease in temperature**, while adiabatic compression causes an increase in temperature.

This temperature change occurs even though no heat flows into or out of the system.

---

### 3. Adiabatic Condition for an Ideal Gas

For a quasistatic adiabatic process involving an ideal gas, the pressure and volume satisfy

$$
PV^\gamma = \text{constant},
$$

where

$$
\gamma = \frac{C_P}{C_V}.
$$

For common gases:
- Monatomic gas: $\gamma = \frac{5}{3}$,
- Diatomic gas (room temperature): $\gamma = \frac{7}{5}$.

---

### 4. Additional Adiabatic Relations

From $PV^\gamma = \text{constant}$ and the ideal gas law, several equivalent relations follow:

$$
TV^{\gamma - 1} = \text{constant},
$$

$$
T^\gamma P^{1-\gamma} = \text{constant}.
$$

These relations allow direct calculation of temperature, pressure, or volume changes.

---

### Example 19.14. Adiabatic Expansion

One mole of an ideal monatomic gas expands adiabatically from:
- $T_i = 400~\text{K}$,
- $V_i = 0.010~\text{m}^3$,

to:
- $V_f = 0.020~\text{m}^3$.

For a monatomic gas, $\gamma = 5/3$.

Using

$$
T_f = T_i \left(\frac{V_i}{V_f}\right)^{\gamma - 1},
$$

$$
T_f = 400 \left(\frac{0.010}{0.020}\right)^{2/3}
     = 400 (0.63)
     \approx 250~\text{K}.
$$

The gas cools significantly during expansion.

---

### 5. Work Done in an Adiabatic Process

The work done by an ideal gas during adiabatic expansion is

$$
W = \frac{P_i V_i - P_f V_f}{\gamma - 1}.
$$

This expression differs from the isothermal work formula and reflects the changing temperature.

---

### 6. Comparison with Isothermal Expansion

For the same initial and final volumes:

- **Isothermal expansion:** Temperature constant, heat flows into the gas, less cooling.
- **Adiabatic expansion:** No heat transfer, temperature drops more sharply.

On a $P$–$V$ diagram, adiabatic curves are steeper than isothermal curves.

---

### 7. Physical Realizations of Adiabatic Processes

Adiabatic processes occur when:
- The system is well insulated,
- The process happens very rapidly,
- Heat transfer is negligible during the process.

Examples include:
- Rapid expansion of gases in engines,
- Compression of air in a bicycle pump,
- Atmospheric processes such as rising air parcels.

---

### 8. Atmospheric Application

As air rises in the atmosphere, it expands due to decreasing pressure.  
This expansion is approximately adiabatic, leading to cooling with altitude.

This mechanism explains cloud formation and weather patterns.

---

### Conceptual Notes

In an adiabatic process, changes in internal energy arise solely from work done. For ideal gases, adiabatic expansion causes cooling and compression causes heating. The relations linking pressure, volume, and temperature differ from those of isothermal processes and depend on the heat capacity ratio $\gamma$. Adiabatic processes are central to understanding engines, refrigeration, and atmospheric physics.

---

## 19.10. Heat Transfer: Conduction, Convection, Radiation

Heat can be transferred between systems by three distinct physical mechanisms: **conduction**, **convection**, and **radiation**.  
All three mechanisms may operate simultaneously, but each has a different microscopic origin and dominates under different conditions.

---

### 1. Overview of Heat Transfer Mechanisms

- **Conduction:** Energy transfer through direct microscopic collisions and interactions within a material or between materials in contact.  
- **Convection:** Energy transfer by the bulk motion of a fluid.  
- **Radiation:** Energy transfer by electromagnetic waves, requiring no material medium.

These mechanisms explain heat flow in solids, fluids, and even across empty space.

---

### 2. Conduction

**Conduction** occurs when energy is transferred through a material due to temperature gradients, without macroscopic motion of the material itself.

In solids, conduction is primarily due to:
- Vibrational energy transfer between atoms or molecules,
- Motion of free electrons in metals.

---

### 3. Rate of Heat Conduction

For steady-state, one-dimensional heat flow through a material of thickness $L$ and cross-sectional area $A$, the rate of heat transfer is given by **Fourier’s law**:

$$
\frac{dQ}{dt} = kA \frac{T_H - T_C}{L},
$$

where:
- $k$ is the **thermal conductivity** of the material,
- $T_H$ and $T_C$ are the temperatures of the hot and cold sides.

A larger thermal conductivity corresponds to better heat conduction.

---

### Example 19.15. Heat Conduction Through a Wall

A brick wall of thickness $0.20~\text{m}$ and area $5.0~\text{m}^2$ separates indoor air at $20^\circ\text{C}$ from outdoor air at $0^\circ\text{C}$.  
If $k = 0.60~\text{W/(m·K)}$:

$$
\frac{dQ}{dt} = (0.60)(5.0)\frac{20}{0.20} = 300~\text{W}.
$$

---

### 4. Thermal Resistance

Thermal conduction can be treated analogously to electrical resistance.  
The **thermal resistance** is defined as

$$
R_{\text{th}} = \frac{L}{kA}.
$$

The heat flow rate becomes

$$
\frac{dQ}{dt} = \frac{T_H - T_C}{R_{\text{th}}}.
$$

Thermal resistances in series add directly.

---

### 5. Convection

**Convection** occurs when heat is transferred by the bulk motion of a fluid.

There are two main types:
- **Natural convection:** Driven by buoyancy forces due to density differences caused by temperature gradients,
- **Forced convection:** Driven by external means such as fans or pumps.

---

### 6. Convective Heat Transfer

The rate of convective heat transfer is often described empirically by

$$
\frac{dQ}{dt} = hA(T_s - T_f),
$$

where:
- $h$ is the **convective heat transfer coefficient**,
- $T_s$ is the surface temperature,
- $T_f$ is the fluid temperature.

The coefficient $h$ depends strongly on fluid properties and flow conditions.

---

### Example 19.16. Cooling by Airflow

A metal surface of area $0.50~\text{m}^2$ at $80^\circ\text{C}$ is cooled by air at $20^\circ\text{C}$.  
If $h = 15~\text{W/(m}^2\text{·K)}$:

$$
\frac{dQ}{dt} = (15)(0.50)(60) = 450~\text{W}.
$$

---

### 7. Radiation

**Radiation** is heat transfer by electromagnetic waves, primarily in the infrared region.  
Unlike conduction and convection, radiation does not require a medium and can occur through vacuum.

All objects emit radiation continuously due to thermal motion of charged particles.

---

### 8. Stefan–Boltzmann Law

The power radiated by a surface at absolute temperature $T$ is

$$
P = \varepsilon \sigma A T^4,
$$

where:
- $\varepsilon$ is the **emissivity** of the surface,
- $\sigma = 5.67\times10^{-8}~\text{W/(m}^2\text{·K}^4)$ is the **Stefan–Boltzmann constant**.

A perfect blackbody has $\varepsilon = 1$.

---

### 9. Net Radiative Heat Transfer

When a surface at temperature $T$ radiates to surroundings at temperature $T_s$, the net radiated power is

$$
P_{\text{net}} = \varepsilon \sigma A (T^4 - T_s^4).
$$

Radiative heat transfer increases rapidly with temperature due to the $T^4$ dependence.

---

### Example 19.17. Radiative Heat Loss

A surface of area $1.0~\text{m}^2$ with emissivity $\varepsilon = 0.8$ is at $400~\text{K}$ in surroundings at $300~\text{K}$.

$$
P_{\text{net}} = (0.8)(5.67\times10^{-8})(1.0)(400^4 - 300^4)
               \approx 620~\text{W}.
$$

---

### 10. Comparison of Heat Transfer Mechanisms

- Conduction dominates in solids and stationary fluids,
- Convection dominates in moving fluids,
- Radiation dominates at high temperatures and across vacuum.

Real systems often involve all three simultaneously.

---

### 11. Insulation and Thermal Control

Good thermal insulators:
- Have low thermal conductivity,
- Trap air to suppress convection,
- Reflect radiation to reduce radiative losses.

Examples include fiberglass, foam, and reflective coatings.

---

### Conceptual Notes

Heat transfer occurs by conduction, convection, and radiation. Conduction involves microscopic energy transfer within materials, convection involves bulk fluid motion, and radiation involves electromagnetic waves. Each mechanism obeys distinct physical laws and dominates under different conditions. Understanding these mechanisms is essential for thermal engineering, climate control, and energy efficiency.

## 20.1. The Second Law of Thermodynamics — Introduction

The **first law of thermodynamics** states that energy is conserved, but it does not specify which processes actually occur in nature. Many processes that satisfy energy conservation are never observed. The **second law of thermodynamics** provides the additional principle that determines the *direction* of natural processes.

Experience shows that certain processes occur spontaneously in only one direction. Heat flows naturally from a hot object to a cold one, never spontaneously in reverse. Gases expand to fill available volume, but they do not spontaneously compress themselves. Mechanical energy can be converted into thermal energy by friction, but thermal energy does not spontaneously convert entirely into mechanical work.

These observations motivate the need for a second fundamental law.

---

### Directionality of Natural Processes

The second law addresses the question: **why do some processes occur while their reverse never does**, even though both would conserve energy.

Examples of irreversible processes include:
- Heat flowing from a higher temperature to a lower temperature,
- A gas freely expanding into a vacuum,
- Mechanical energy being degraded into internal (thermal) energy by friction.

The reverse of these processes is not observed in nature.

---

### Kelvin–Planck Statement of the Second Law

One formulation of the second law concerns heat engines:

> It is impossible for any device operating in a cycle to absorb heat from a single thermal reservoir and convert it completely into work.

This means that **no heat engine can be 100% efficient**. Some heat must always be expelled to a colder reservoir.

If a heat engine absorbs heat $Q_H$ from a hot reservoir and produces work $W$, then it must reject heat $Q_L$ such that

$$
Q_H = W + Q_L.
$$

Because $Q_L > 0$, complete conversion of heat into work is impossible.

---

### Clausius Statement of the Second Law

An equivalent formulation focuses on heat transfer:

> No process is possible whose sole effect is to transfer heat from a colder body to a hotter body.

This explains why refrigerators and heat pumps require external work to operate. Heat does not naturally flow from cold to hot.

---

### Reversibility and Irreversibility

A **reversible process** is an idealized process that can be reversed without leaving any changes in the system or surroundings. Such processes proceed infinitely slowly and without friction or dissipation.

All **real processes are irreversible**, because they involve effects such as:
- Friction,
- Finite temperature differences,
- Turbulence,
- Inelastic deformation.

The second law applies most strongly to irreversible processes, which always proceed in a preferred direction.

---

### Heat Engines as Motivation

A **heat engine** operates between two reservoirs at temperatures $T_H$ and $T_L$, absorbing heat $Q_H$, performing work $W$, and rejecting heat $Q_L$.

The efficiency of a heat engine is defined as

$$
e = \frac{W}{Q_H} = 1 - \frac{Q_L}{Q_H}.
$$

The second law guarantees that $Q_L \neq 0$, and therefore $e < 1$.

---

### Conceptual Summary

The second law of thermodynamics introduces a fundamental **asymmetry in time**. While energy conservation allows many processes in principle, the second law determines which processes actually occur. It explains why heat engines cannot be perfectly efficient, why heat flows spontaneously from hot to cold, and why natural processes have a preferred direction.

This law will later be expressed quantitatively using the concept of **entropy**, which provides a general and powerful formulation of irreversibility.




## 20.2. Heat Engines

A **heat engine** is a device that converts thermal energy into mechanical work by operating between two thermal reservoirs at different temperatures.  
The operation of heat engines illustrates the practical implications of the **second law of thermodynamics** and explains why no engine can be perfectly efficient.

---

### 1. Basic Structure of a Heat Engine

A heat engine operates cyclically and consists of three essential elements:

- A **hot reservoir** at temperature $T_H$,
- A **working substance** (such as a gas or steam),
- A **cold reservoir** at temperature $T_L$, with $T_L < T_H$.

During one complete cycle:
- The engine absorbs heat $Q_H$ from the hot reservoir,
- It performs work $W$ on the surroundings,
- It rejects heat $Q_L$ to the cold reservoir.

---

### 2. Energy Balance for a Heat Engine

Because the engine operates in a cycle, its internal energy returns to its initial value after each cycle:

$$
\Delta U = 0.
$$

Applying the First Law of Thermodynamics:

$$
Q_H = W + Q_L.
$$

This relation shows that not all absorbed heat can be converted into work.

---

### 3. Efficiency of a Heat Engine

The **thermal efficiency** $e$ of a heat engine is defined as the fraction of absorbed heat converted into useful work:

$$
e = \frac{W}{Q_H}.
$$

Using the energy balance equation, efficiency can also be written as

$$
e = 1 - \frac{Q_L}{Q_H}.
$$

Because $Q_L > 0$, the efficiency is always less than 1.

---

### 4. Physical Interpretation of Efficiency

The efficiency expresses how effectively a heat engine converts heat into work.

- High efficiency means most of the absorbed heat becomes work,
- Low efficiency means much of the absorbed heat is wasted as rejected heat.

The second law guarantees that some heat must always be rejected to a colder reservoir.

---

### 5. Example 20.1. Simple Heat Engine

A heat engine absorbs $5000~\text{J}$ of heat from a hot reservoir and rejects $3500~\text{J}$ to a cold reservoir.

The work done is

$$
W = Q_H - Q_L = 5000 - 3500 = 1500~\text{J}.
$$

The efficiency is

$$
e = \frac{1500}{5000} = 0.30.
$$

The engine has an efficiency of **30%**.

---

### 6. Necessity of a Cold Reservoir

A cold reservoir is essential for heat engine operation.  
If $Q_L = 0$, the engine would convert all absorbed heat into work, violating the Kelvin–Planck statement of the second law.

Thus, **waste heat is unavoidable** in any cyclic engine.

---

### 7. Cyclic Nature of Heat Engines

Because the working substance returns to its initial state after each cycle:
- The net internal energy change is zero,
- All energy input and output must occur as heat or work.

This cyclic operation distinguishes heat engines from one-time processes.

---

### 8. Real Heat Engines

Examples of real heat engines include:
- Steam turbines in power plants,
- Internal combustion engines in cars,
- Jet engines,
- Stirling engines.

All real engines are irreversible and operate with efficiencies far below the theoretical maximum.

---

### 9. Sources of Inefficiency

Efficiency is reduced by:
- Friction,
- Turbulence,
- Heat loss to the environment,
- Finite temperature differences during heat transfer.

Even in the absence of these effects, a fundamental limit on efficiency remains, as shown by the Carnot engine.

---

### 10. Work and Power Output

The **power output** of a heat engine is the work done per unit time:

$$
P = \frac{W}{\Delta t}.
$$

High power output does not necessarily imply high efficiency; an engine can produce large power while wasting most of the input heat.

---

### Conceptual Notes

Heat engines convert heat into work by operating between two temperature reservoirs. The First Law governs the energy balance, while the Second Law limits efficiency and requires rejection of waste heat. No heat engine can be perfectly efficient, and real engines fall even further below the theoretical maximum due to irreversibility and losses.

---

In the next section, we introduce the **Carnot engine**, which establishes the maximum possible efficiency for any heat engine operating between two given temperatures.


## 20.3. The Carnot Engine; Reversible and Irreversible Processes

The **Carnot engine** is an idealized heat engine that operates in a completely **reversible** manner.  
Although no real engine can be perfectly reversible, the Carnot engine establishes the **maximum possible efficiency** that any heat engine can achieve when operating between two given temperatures.

---

### 1. Reversible and Irreversible Processes

A **reversible process** is an ideal process that can be reversed without leaving any net change in either the system or the surroundings.  
Such a process proceeds infinitely slowly, with no friction, no turbulence, and no finite temperature differences during heat transfer.

An **irreversible process** is any real process in which reversing the direction leaves changes in the system or surroundings.  
All real engines operate irreversibly due to friction, finite heat flow rates, and dissipative effects.

Reversible processes provide an upper bound on performance; irreversible processes always perform worse.

---

### 2. Carnot Cycle Overview

The Carnot engine operates in a cycle consisting of **four reversible steps**, using an ideal gas as the working substance:

1. Reversible **isothermal expansion** at temperature $T_H$,  
2. Reversible **adiabatic expansion** from $T_H$ to $T_L$,  
3. Reversible **isothermal compression** at temperature $T_L$,  
4. Reversible **adiabatic compression** from $T_L$ back to $T_H$.

After completing these four steps, the system returns to its initial state.

---

### 3. Isothermal Expansion at $T_H$

During the first step, the gas expands isothermally while in contact with the hot reservoir at temperature $T_H$.

- Heat $Q_H$ flows into the gas,
- The gas does work on the surroundings,
- Internal energy remains constant because temperature is constant.

Thus,

$$
Q_H = W_1.
$$

---

### 4. Adiabatic Expansion

In the second step, the gas expands adiabatically:

- No heat is exchanged ($Q = 0$),
- The gas continues to do work,
- The temperature drops from $T_H$ to $T_L$.

The decrease in internal energy supplies the work done during this step.

---

### 5. Isothermal Compression at $T_L$

In the third step, the gas is compressed isothermally while in contact with the cold reservoir at temperature $T_L$.

- Heat $Q_L$ is expelled to the cold reservoir,
- Work is done on the gas,
- Internal energy remains constant.

The heat rejected satisfies

$$
Q_L = W_3.
$$

---

### 6. Adiabatic Compression

In the final step, the gas is compressed adiabatically:

- No heat is exchanged,
- Work is done on the gas,
- The temperature rises from $T_L$ back to $T_H$.

The system is now ready to repeat the cycle.

---

### 7. Efficiency of the Carnot Engine

The efficiency of any heat engine is

$$
e = \frac{W}{Q_H} = 1 - \frac{Q_L}{Q_H}.
$$

For the Carnot engine, the ratio of heat transfers equals the ratio of reservoir temperatures:

$$
\frac{Q_L}{Q_H} = \frac{T_L}{T_H},
$$

where temperatures are measured in kelvin.

Thus, the **Carnot efficiency** is

$$
e_C = 1 - \frac{T_L}{T_H}.
$$

---

### 8. Significance of Carnot Efficiency

The Carnot efficiency depends **only on the temperatures** of the two reservoirs, not on:
- The working substance,
- The details of the engine design.

No engine operating between the same two temperatures can have an efficiency greater than the Carnot efficiency.

---

### Example 20.2. Carnot Efficiency

A Carnot engine operates between:
- $T_H = 600~\text{K}$,
- $T_L = 300~\text{K}$.

The maximum possible efficiency is

$$
e_C = 1 - \frac{300}{600} = 0.50.
$$

No real engine operating between these temperatures can exceed 50% efficiency.

---

### 9. Irreversibility and Real Engines

Real engines are always less efficient than the Carnot engine because of irreversibility caused by:
- Friction,
- Finite temperature differences during heat transfer,
- Turbulence and dissipation.

These effects generate entropy and reduce useful work output.

---

### 10. Carnot’s Theorem

**Carnot’s theorem** states:

- All reversible engines operating between the same two temperatures have the same efficiency.
- No irreversible engine can be more efficient than a reversible one operating between the same temperatures.

This theorem establishes an absolute upper limit on engine performance.

---

### Conceptual Notes

The Carnot engine represents an ideal limit rather than a practical device. Its efficiency depends only on reservoir temperatures and sets a fundamental bound on all heat engines. Reversibility is a theoretical ideal; real processes are always irreversible and therefore less efficient. The Carnot cycle clarifies why waste heat is unavoidable and why temperature differences drive energy conversion.

---

In the next section, we analyze **refrigerators, air conditioners, and heat pumps**, which operate by reversing the heat engine concept.


## 20.4. Refrigerators, Air Conditioners, and Heat Pumps

Devices such as **refrigerators**, **air conditioners**, and **heat pumps** operate by transferring heat from a colder region to a warmer one.  
Because this process does not occur spontaneously, **external work** is required.  
These devices are essentially **heat engines run in reverse** and are governed by the second law of thermodynamics.

---

### 1. Basic Operating Principle

All three devices operate between:
- A **cold reservoir** at temperature $T_L$,
- A **hot reservoir** at temperature $T_H$, with $T_H > T_L$.

During one cycle:
- Heat $Q_L$ is removed from the cold reservoir,
- Work $W$ is done on the system,
- Heat $Q_H$ is expelled to the hot reservoir.

Energy conservation requires:

$$
Q_H = Q_L + W.
$$

---

### 2. Refrigerator

A **refrigerator** is designed to remove heat from a low-temperature region, such as the interior of a refrigerator, and reject it to the warmer surroundings.

The goal is to maximize the amount of heat removed from the cold reservoir for a given work input.

---

### 3. Coefficient of Performance (Refrigerator)

The performance of a refrigerator is measured by its **coefficient of performance** (COP), defined as

$$
\text{COP}_{\text{ref}} = \frac{Q_L}{W}.
$$

Using the energy balance equation:

$$
\text{COP}_{\text{ref}} = \frac{Q_L}{Q_H - Q_L}.
$$

A larger COP indicates a more effective refrigerator.

---

### Example 20.3. Refrigerator Performance

A refrigerator removes $1200~\text{J}$ of heat from its interior while consuming $300~\text{J}$ of electrical energy.

$$
\text{COP}_{\text{ref}} = \frac{1200}{300} = 4.0.
$$

---

### 4. Air Conditioners

An **air conditioner** is functionally identical to a refrigerator.  
The difference lies in what is considered the system of interest.

- In a refrigerator, the cooled region is the interior cabinet.
- In an air conditioner, the cooled region is an entire room or building.

The same coefficient of performance applies.

---

### 5. Heat Pumps

A **heat pump** is designed to deliver heat to a warm space, such as a house in winter.

Instead of focusing on heat removal from the cold reservoir, the goal is to maximize heat delivered to the hot reservoir.

---

### 6. Coefficient of Performance (Heat Pump)

For a heat pump, the coefficient of performance is defined as

$$
\text{COP}_{\text{hp}} = \frac{Q_H}{W}.
$$

Using $Q_H = Q_L + W$:

$$
\text{COP}_{\text{hp}} = 1 + \text{COP}_{\text{ref}}.
$$

Heat pumps can therefore have COP values greater than 1.

---

### Example 20.4. Heat Pump Efficiency

If a heat pump uses $500~\text{J}$ of work to deliver $2500~\text{J}$ of heat into a house:

$$
\text{COP}_{\text{hp}} = \frac{2500}{500} = 5.0.
$$

---

### 7. Carnot Refrigerator and Heat Pump

The maximum possible performance is achieved by an ideal **Carnot refrigerator** or **Carnot heat pump**.

For a Carnot refrigerator:

$$
\text{COP}_{\text{ref,max}} = \frac{T_L}{T_H - T_L}.
$$

For a Carnot heat pump:

$$
\text{COP}_{\text{hp,max}} = \frac{T_H}{T_H - T_L}.
$$

Temperatures must be expressed in kelvin.

---

### Example 20.5. Maximum COP

A refrigerator operates between:
- $T_L = 270~\text{K}$,
- $T_H = 300~\text{K}$.

The maximum possible COP is

$$
\text{COP}_{\text{ref,max}} = \frac{270}{300 - 270} = 9.0.
$$

Real devices always have lower performance due to irreversibility.

---

### 8. Practical Limitations

Real refrigerators and heat pumps are less efficient than Carnot devices due to:
- Friction,
- Heat loss,
- Non-quasistatic operation,
- Finite temperature differences.

Nevertheless, heat pumps can be significantly more energy-efficient than direct electric heating.

---

### 9. Second Law Perspective

Refrigerators and heat pumps do not violate the second law because:
- External work is supplied,
- Total entropy of the system plus surroundings increases.

Heat is moved from cold to hot only at the cost of work.

---

### Conceptual Notes

Refrigerators, air conditioners, and heat pumps transfer heat against its natural direction by consuming work. Their effectiveness is measured by coefficients of performance rather than efficiency. Ideal Carnot devices set the upper limit on performance, which depends only on reservoir temperatures. Real devices always operate below this limit due to irreversibility.

---

In the next section, we introduce **entropy**, a central concept that provides a quantitative measure of irreversibility and underpins the second law of thermodynamics.


## 20.5. Entropy

The concept of **entropy** provides a precise, quantitative way to express the second law of thermodynamics.  
Entropy characterizes the direction of thermodynamic processes and distinguishes between reversible and irreversible behavior.

---

### 1. Motivation for Entropy

The second law states that some processes are irreversible, but it does not initially provide a numerical quantity that always increases.  
Entropy is introduced to fill this role.

Entropy allows us to:
- Predict whether a process can occur,
- Compare different processes quantitatively,
- Express the second law in a compact mathematical form.

---

### 2. Entropy Change in a Reversible Process

For a **reversible** process, the infinitesimal change in entropy $dS$ is defined as

$$
dS = \frac{\delta Q_{\text{rev}}}{T},
$$

where:
- $\delta Q_{\text{rev}}$ is the infinitesimal heat transfer in a reversible process,
- $T$ is the absolute temperature at which the transfer occurs.

Entropy is measured in $\text{J/K}$.

---

### 3. Entropy as a State Function

Entropy is a **state function**, meaning:
- Its value depends only on the state of the system,
- The change in entropy $\Delta S$ depends only on the initial and final states.

Even if a process is irreversible, the entropy change can be calculated by imagining a reversible path between the same states.

---

### 4. Finite Entropy Change

For a reversible process in which heat $Q_{\text{rev}}$ is transferred at constant temperature $T$:

$$
\Delta S = \frac{Q_{\text{rev}}}{T}.
$$

This expression applies, for example, to isothermal phase changes and reversible isothermal expansion of gases.

---

### Example 20.6. Entropy Change During Melting

When $0.50~\text{kg}$ of ice melts reversibly at $0^\circ\text{C}$:

$$
\Delta S = \frac{Q_{\text{rev}}}{T}
         = \frac{mL_f}{T}
         = \frac{(0.50)(3.34\times10^5)}{273}
         = 612~\text{J/K}.
$$

---

### 5. Entropy Change of an Ideal Gas (Isothermal)

For a reversible isothermal expansion of an ideal gas from $V_i$ to $V_f$:

$$
\Delta S = nR \ln\!\left(\frac{V_f}{V_i}\right).
$$

Entropy increases when the gas expands and decreases when it is compressed.

---

### 6. Entropy Change of an Ideal Gas (General Case)

For any reversible change of an ideal gas between states $(T_i, V_i)$ and $(T_f, V_f)$:

$$
\Delta S = nC_V \ln\!\left(\frac{T_f}{T_i}\right)
         + nR \ln\!\left(\frac{V_f}{V_i}\right).
$$

This formula is widely used in thermodynamics.

---

### 7. Physical Interpretation of Entropy

Entropy is often associated with:
- Energy dispersal,
- The spreading of energy over available microscopic states.

A higher-entropy state corresponds to a more uniform distribution of energy among microscopic degrees of freedom.

---

### 8. Entropy and Irreversibility

In irreversible processes:
- Friction converts work into internal energy,
- Heat flows across finite temperature differences,
- Turbulence and dissipation occur.

These processes increase entropy beyond what would occur in a reversible process between the same states.

---

### 9. Units and Scale

Entropy has units of $\text{J/K}$.  
Although the entropy of a single molecule is extremely small, macroscopic systems involve enormous numbers of particles, making entropy changes readily measurable.

---

### Conceptual Notes

Entropy is a state function that quantifies the directionality of thermodynamic processes. It is defined through reversible heat transfer and provides a precise mathematical formulation of the second law. Entropy increases when energy becomes more spread out among microscopic degrees of freedom and serves as a powerful criterion for determining whether a process can occur.

---

In the next section, we apply entropy to the **second law of thermodynamics** and show how it governs the behavior of isolated systems.


## 20.6. Entropy and the Second Law of Thermodynamics

The concept of entropy allows the **second law of thermodynamics** to be expressed in a precise and universal mathematical form.  
Instead of describing individual processes qualitatively, the second law can now be stated as a general rule governing the entropy of systems and their surroundings.

---

### 1. The Second Law in Terms of Entropy

The second law of thermodynamics can be stated as:

> In any natural process, the total entropy of an isolated system never decreases.

Mathematically,

$$
\Delta S_{\text{total}} \ge 0.
$$

Equality holds for **reversible** processes, and the inequality is strict for **irreversible** processes.

---

### 2. System and Surroundings

To apply the entropy form of the second law, we consider:
- The **system**, which undergoes a process,
- The **surroundings**, which interact with the system.

The total entropy change is

$$
\Delta S_{\text{total}} = \Delta S_{\text{system}} + \Delta S_{\text{surroundings}}.
$$

Even if the entropy of the system decreases, the total entropy can still increase if the surroundings’ entropy increases by a larger amount.

---

### 3. Isolated Systems

An **isolated system** exchanges neither heat nor work with its surroundings.

For an isolated system:

$$
\Delta S_{\text{isolated}} \ge 0.
$$

Entropy remains constant for reversible processes and increases for irreversible ones.

---

### 4. Heat Transfer Between Two Reservoirs

Consider heat $Q$ flowing from a hot reservoir at temperature $T_H$ to a cold reservoir at temperature $T_L$.

Entropy change of the hot reservoir:
$$
\Delta S_H = -\frac{Q}{T_H}.
$$

Entropy change of the cold reservoir:
$$
\Delta S_L = \frac{Q}{T_L}.
$$

Total entropy change:
$$
\Delta S_{\text{total}} = Q\left(\frac{1}{T_L} - \frac{1}{T_H}\right) > 0,
$$

since $T_L < T_H$.

This explains why heat flows spontaneously from hot to cold.

---

### 5. Reversible vs. Irreversible Heat Transfer

If heat is transferred reversibly through an infinitesimal temperature difference, the entropy change of the system and surroundings exactly cancel:

$$
\Delta S_{\text{total}} = 0.
$$

In real heat transfer, finite temperature differences exist, and entropy increases.

---

### 6. Entropy Production

The increase in total entropy in an irreversible process is called **entropy production**.

Entropy production arises from:
- Friction,
- Inelastic deformation,
- Electrical resistance,
- Finite-rate heat flow.

These effects are unavoidable in real systems.

---

### 7. Entropy and Heat Engines

For a heat engine operating between two reservoirs:

$$
\Delta S_{\text{total}} = -\frac{Q_H}{T_H} + \frac{Q_L}{T_L}.
$$

For a **reversible** (Carnot) engine,

$$
\Delta S_{\text{total}} = 0,
$$

which implies

$$
\frac{Q_L}{Q_H} = \frac{T_L}{T_H}.
$$

For a real engine,

$$
\Delta S_{\text{total}} > 0,
$$

and the efficiency is lower than the Carnot efficiency.

---

### 8. Spontaneity Criterion

A process is **thermodynamically possible** if

$$
\Delta S_{\text{total}} \ge 0.
$$

If $\Delta S_{\text{total}} < 0$, the process cannot occur spontaneously.

This criterion applies to all physical, chemical, and biological processes.

---

### 9. Entropy Decrease in a Subsystem

It is possible for a subsystem to experience a decrease in entropy, provided the surroundings experience a larger increase.

This explains how ordered structures can form locally without violating the second law.

---

### 10. Entropy and Time’s Arrow

The increase of entropy defines a preferred direction of time, often called the **arrow of time**.

Macroscopic processes are irreversible because they move toward states of higher total entropy.

---

### Conceptual Notes

Entropy provides a quantitative formulation of the second law of thermodynamics. In any process, the total entropy of an isolated system cannot decrease. Reversible processes conserve entropy, while irreversible processes produce entropy. This principle determines the direction of natural processes and explains why heat engines have limited efficiency and why spontaneous processes occur only in one direction.

---

## 20.7. Order to Disorder

The concept of entropy is often associated with the transition from **order to disorder**.  
While this language is qualitative, it provides useful intuition for understanding why natural processes tend to proceed in one direction and why highly ordered states are rarely observed without external intervention.

---

### 1. Order, Disorder, and Entropy

A highly ordered system is one in which the microscopic components are arranged in a very specific, constrained way.  
A disordered system allows many more possible microscopic arrangements consistent with the same macroscopic state.

Entropy measures the number of microscopic arrangements compatible with a macroscopic description.  
States with greater disorder generally correspond to higher entropy.

---

### 2. Expansion of a Gas

Consider a gas initially confined to one half of a container.

When the partition is removed:
- The gas expands to fill the entire container,
- The macroscopic pressure and temperature may remain unchanged,
- The number of accessible microscopic configurations increases enormously.

The entropy increases because the molecules have many more possible positions.

The reverse process, in which all molecules spontaneously return to one half of the container, is not observed.

---

### 3. Mixing of Substances

When two different gases are allowed to mix:
- They spread throughout the available volume,
- The final mixed state has higher entropy than the separated state.

This increase in entropy explains why mixing occurs spontaneously and why spontaneous unmixing does not occur.

---

### 4. Entropy Change for Free Expansion

For an ideal gas undergoing free expansion into a vacuum, no work is done and no heat is exchanged:

$$
Q = 0, \quad W = 0.
$$

From the First Law,

$$
\Delta U = 0.
$$

However, the entropy increases:

$$
\Delta S = nR \ln\!\left(\frac{V_f}{V_i}\right) > 0.
$$

This demonstrates that entropy change is not determined by energy change alone.

---

### 5. Microscopic Interpretation

At the microscopic level:
- Ordered states correspond to a small number of possible molecular arrangements,
- Disordered states correspond to a vast number of arrangements.

Natural processes overwhelmingly favor states with the largest number of accessible microscopic configurations.

---

### 6. Probability and Entropy

The tendency toward disorder is fundamentally a matter of probability.

Highly ordered arrangements are not forbidden by the laws of mechanics, but they are extremely improbable for systems containing enormous numbers of particles.

Entropy increase reflects the overwhelming statistical likelihood of disordered states.

---

### 7. Apparent Violations of Order Increase

Local decreases in entropy are possible and commonly observed.

Examples include:
- Crystal formation from a liquid,
- Growth of living organisms,
- Formation of stars and planets.

These processes do not violate the second law because they are accompanied by larger entropy increases in the surroundings.

---

### 8. Energy Input and Order

Maintaining or creating order requires a continuous input of energy.

Living systems, machines, and organized structures persist only because energy flows through them, increasing the entropy of the environment.

Without energy input, ordered systems decay toward disorder.

---

### 9. Order vs. Energy Conservation

Energy conservation alone does not prevent systems from becoming disordered.  
The second law, through entropy, explains why disorder increases even when total energy remains constant.

This distinction highlights why entropy is a fundamentally new concept beyond energy.

---

### Conceptual Notes

Entropy increase is often described as a transition from order to disorder. Ordered states correspond to few microscopic configurations, while disordered states correspond to many. Natural processes favor states of higher probability, which typically have higher entropy. Local decreases in entropy are possible but must be offset by greater increases elsewhere, ensuring that total entropy does not decrease.

---

## 20.8. Unavailability of Energy; Heat Death

Although energy is conserved, not all energy is equally useful for doing work.  
As entropy increases, a growing fraction of energy becomes **unavailable for conversion into mechanical work**.  
This limitation is a direct consequence of the second law of thermodynamics and leads to the concept of **heat death**.

---

### 1. Available and Unavailable Energy

Energy that can be converted into work is called **available energy**.  
Energy that is uniformly distributed and cannot be harnessed to perform work is called **unavailable energy**.

As entropy increases, energy becomes more dispersed and less capable of producing organized motion.

---

### 2. Energy Degradation

In many processes, energy is transformed from organized forms into disorganized thermal energy.

Examples include:
- Mechanical energy converted to thermal energy by friction,
- Electrical energy converted to heat in resistors,
- Turbulent flow converting kinetic energy into internal energy.

Although total energy remains constant, its **quality** decreases.

---

### 3. Entropy and Work Potential

The ability of a system to do work depends on the existence of gradients, such as:
- Temperature differences,
- Pressure differences,
- Chemical potential differences.

As these gradients disappear, entropy increases and the capacity to do work diminishes.

A system in complete equilibrium has maximum entropy and zero ability to do work.

---

### 4. Heat Engines and Energy Availability

A heat engine requires two reservoirs at different temperatures.

As heat flows from hot to cold:
- Some energy is converted into work,
- The rest is rejected as waste heat.

Once both reservoirs reach the same temperature, no further work can be extracted.

---

### 5. Entropy Balance and Energy Loss

For a process involving heat transfer $Q$ at temperature $T$:

$$
\Delta S = \frac{Q}{T}.
$$

The associated increase in entropy represents a loss of work potential.  
Even reversible processes convert some energy into unavailable forms when considered in a broader context.

---

### 6. The Concept of Heat Death

The **heat death of the universe** refers to a hypothetical final state in which:
- All energy is uniformly distributed,
- All temperature differences vanish,
- Entropy reaches its maximum possible value.

In this state:
- No heat engines can operate,
- No work can be performed,
- All macroscopic processes cease.

---

### 7. Heat Death and Cosmology

The concept of heat death arises from applying the second law to the universe as a whole, treated as an isolated system.

As entropy increases over cosmic time:
- Stars burn out,
- Temperature gradients diminish,
- Energy becomes increasingly unavailable.

The heat death scenario is one possible long-term fate of the universe.

---

### 8. Limitations and Open Questions

The application of thermodynamics to the entire universe involves assumptions:
- The universe is treated as an isolated system,
- Classical thermodynamics is extended to cosmological scales.

Modern cosmology introduces complexities such as:
- Expansion of the universe,
- Dark energy,
- Quantum effects.

These factors influence how entropy evolves on the largest scales.

---

### 9. Practical Implications

On human time scales, heat death is irrelevant.  
However, the concept highlights an important principle:

Even though energy is conserved, **useful energy is continually lost** due to entropy increase.

This principle underlies efficiency limits in engines and the necessity of continual energy input to sustain organized systems.

---

### Conceptual Notes

Entropy increase leads to a gradual degradation of energy quality. As entropy grows, energy becomes more uniformly distributed and less capable of doing work. The idea of heat death represents the ultimate consequence of this process, where all energy is unavailable for work. While speculative on cosmic scales, the concept reinforces the fundamental asymmetry introduced by the second law of thermodynamics.

---

## 20.9. Statistical Interpretation of Entropy and the Second Law

While entropy was originally introduced through macroscopic heat transfer, its deepest meaning emerges from a **statistical interpretation**.  
From this viewpoint, the second law of thermodynamics is not an absolute mechanical law but a statement about **probability** and the behavior of systems with enormous numbers of particles.

---

### 1. Microstates and Macrostates

A **macrostate** is specified by macroscopic variables such as pressure, volume, temperature, and energy.  
A **microstate** is a complete specification of the positions and momenta of all particles in the system.

Many different microstates can correspond to the same macrostate.

---

### 2. Multiplicity

The **multiplicity** $W$ of a macrostate is the number of distinct microstates consistent with that macrostate.

- Small $W$ → highly ordered, unlikely state,  
- Large $W$ → highly disordered, very likely state.

Macroscopic systems typically have astronomically large multiplicities.

---

### 3. Boltzmann’s Entropy Formula

The statistical definition of entropy is given by **Boltzmann’s formula**:

$$
S = k \ln W,
$$

where:
- $S$ is the entropy,
- $k = 1.38\times10^{-23}~\text{J/K}$ is Boltzmann’s constant,
- $W$ is the multiplicity.

This equation connects thermodynamics with probability.

---

### 4. Meaning of the Logarithm

The logarithm ensures that:
- Entropy is additive for independent systems,
- Extremely large values of $W$ correspond to manageable values of $S$.

If two independent systems have multiplicities $W_1$ and $W_2$, then

$$
W_{\text{total}} = W_1 W_2,
$$

and

$$
S_{\text{total}} = k \ln W_{\text{total}} = S_1 + S_2.
$$

---

### 5. Entropy Increase as Probability Increase

When an isolated system evolves spontaneously, it moves from macrostates with smaller multiplicity to macrostates with larger multiplicity.

Thus:
- Entropy increase corresponds to increasing probability,
- The most probable macrostate is the one with maximum entropy.

The second law reflects overwhelming statistical likelihood rather than mechanical impossibility.

---

### 6. Gas Expansion Revisited

Consider a gas initially confined to half a container.

- Initial macrostate: small volume, limited molecular positions, small $W$,
- Final macrostate: full volume, many more positions, enormous $W$.

The entropy increase is

$$
\Delta S = k \ln\!\left(\frac{W_f}{W_i}\right) = nR \ln\!\left(\frac{V_f}{V_i}\right).
$$

The reverse process is not forbidden but has an astronomically small probability.

---

### 7. Time Reversal and Irreversibility

The microscopic laws of mechanics are time-reversible.  
However, macroscopic irreversibility arises because:

- The number of microstates corresponding to high-entropy macrostates vastly exceeds those for low-entropy macrostates,
- Systems naturally evolve toward the most probable configurations.

This explains why entropy increases despite time-reversible microscopic dynamics.

---

### 8. Fluctuations and Small Systems

For very small systems or very short times:
- Entropy may occasionally decrease due to fluctuations,
- Such events are observable in microscopic experiments.

For macroscopic systems, fluctuations are negligibly small, and entropy increase appears absolute.

---

### 9. The Second Law as a Statistical Law

The second law is not violated when entropy briefly decreases in a small system.  
Instead, it reflects that entropy increase is overwhelmingly likely for large systems.

In this sense, the second law is a **law of statistics**, not a strict prohibition.

---

### 10. Connection to Information

Entropy is closely related to information theory.

- High entropy corresponds to less information about the exact microstate,
- Low entropy corresponds to more information and greater order.

This connection extends thermodynamics into fields such as computation and communication.

---

### Conceptual Notes

The statistical interpretation of entropy explains irreversibility as a consequence of probability and multiplicity. Entropy measures the number of microscopic configurations consistent with a macroscopic state. Systems evolve toward macrostates with overwhelmingly larger numbers of microstates, making entropy increase virtually inevitable for macroscopic systems.

---

## 20.10. Thermodynamic Temperature; Third Law of Thermodynamics

The concepts of entropy and temperature are deeply connected.  
The **thermodynamic temperature scale** emerges naturally from entropy considerations and leads to the **Third Law of Thermodynamics**, which defines the behavior of entropy at very low temperatures.

---

### 1. Thermodynamic Definition of Temperature

In thermodynamics, temperature can be defined in terms of entropy and internal energy.  
For a reversible process, the change in entropy is

$$
dS = \frac{\delta Q_{\text{rev}}}{T}.
$$

Rearranging gives a thermodynamic definition of temperature:

$$
\frac{1}{T} = \left(\frac{\partial S}{\partial U}\right)_{V,N}.
$$

This definition shows that temperature is a measure of how rapidly entropy changes with internal energy.

---

### 2. Absolute Temperature Scale

The thermodynamic temperature scale is **independent of the properties of any particular substance**.  
It coincides with the Kelvin scale introduced earlier using ideal gas behavior.

Key features:
- Temperature is measured in kelvin (K),
- $0~\text{K}$ corresponds to absolute zero,
- Temperature ratios have direct physical meaning.

---

### 3. Entropy at Low Temperatures

As temperature decreases, molecular motion becomes increasingly restricted.  
At very low temperatures, systems approach a unique microscopic configuration corresponding to minimum energy.

This behavior leads to a universal statement about entropy near absolute zero.

---

### 4. Statement of the Third Law of Thermodynamics

The **Third Law of Thermodynamics** states:

> The entropy of a perfect crystalline substance approaches zero as the temperature approaches absolute zero.

Mathematically,

$$
\lim_{T \to 0} S = 0.
$$

This provides an absolute reference point for entropy.

---

### 5. Perfect Crystals

A **perfect crystal** is one in which:
- Atoms are arranged in a perfectly ordered lattice,
- There is only one possible microscopic configuration at $T = 0$.

For such a system, the multiplicity is

$$
W = 1,
$$

and therefore

$$
S = k \ln W = 0.
$$

---

### 6. Entropy Differences and Absolute Entropy

Before the Third Law, only entropy differences could be measured.  
The Third Law allows absolute entropies to be assigned by integrating from $T = 0$ upward:

$$
S(T) = \int_0^T \frac{C(T')}{T'}\,dT' + \sum \frac{L}{T},
$$

where:
- $C(T')$ is the heat capacity,
- The sum accounts for entropy changes due to phase transitions.

---

### 7. Unattainability of Absolute Zero

Although entropy approaches zero as $T \to 0$, absolute zero cannot be reached in a finite number of steps.

This principle is sometimes called the **unattainability principle** and is considered an alternative formulation of the Third Law.

---

### 8. Residual Entropy

Some substances do not reach zero entropy at $T = 0$ due to structural disorder, even in the ground state.

This remaining entropy is called **residual entropy** and occurs in systems such as:
- Imperfect crystals,
- Substances with orientational disorder.

---

### 9. Experimental Consequences

The Third Law implies that:
- Heat capacities approach zero as $T \to 0$,
- Entropy changes become very small at low temperatures,
- Cooling systems become progressively less effective near absolute zero.

These predictions are confirmed by low-temperature experiments.

---

### 10. Importance of the Third Law

The Third Law:
- Completes the framework of classical thermodynamics,
- Provides an absolute reference for entropy,
- Explains the difficulty of reaching extremely low temperatures,
- Connects thermodynamics with quantum mechanics.

---

### Conceptual Notes

Thermodynamic temperature is defined through the relationship between entropy and internal energy. The Third Law establishes that entropy approaches zero at absolute zero for a perfect crystal, providing an absolute entropy scale. Although absolute zero cannot be reached, the Third Law governs the behavior of matter at very low temperatures and underlies modern low-temperature physics.
